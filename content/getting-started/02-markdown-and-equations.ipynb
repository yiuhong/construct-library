## A formula

A pair of dollar signs sets a formula inside a sentence. The mean of 2, 4, and 7 is $\bar{x} = (2 + 4 + 7) / 3$.

A pair of double dollar signs sets a formula on its own line.

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$
