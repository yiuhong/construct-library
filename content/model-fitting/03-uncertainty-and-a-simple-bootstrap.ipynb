# Uncertainty and a simple bootstrap

How can I picture how a summary would move on a new draw of the same values?

A fitted number is one summary of one sample. A new sample would move it. Two small pictures of that movement: the spread of the values themselves, and a bootstrap that redraws those values.

The list is $[2, 4, 4, 8]$. The mean is one summary. The standard deviation, the square root of the variance, is the spread of these four numbers. A rough band is the mean minus that spread through to the mean plus that spread. It describes this list. It is not, by itself, a claim about a new experiment.


```python
# The mean, the spread, and a band one spread either side of the mean.
data = [2, 4, 4, 8]
mean = sum(data) / len(data)
variance = sum((v - mean) ** 2 for v in data) / len(data)
print(mean)
print(round(variance ** 0.5, 3))
print(round(mean - variance ** 0.5, 3), round(mean + variance ** 0.5, 3))
```

## Redraw the sample

The bootstrap redraws the observed values with replacement, recomputes the summary, and repeats. The spread of those recomputed summaries is a picture of how the summary moves when the sample is redrawn (Efron, 1979). Eight redraws are enough to see the idea. They are not a finished interval.

Efron, B. (1979). Bootstrap methods: Another look at the jackknife. *The Annals of Statistics, 7*(1), 1–26.

The next cell uses a fixed seed, so the same eight means appear each time you run it. The smallest and the largest show the range of this short run.


```python
# Redraw samples of the same size, allowing repeats, and keep each sample's mean.
import numpy as np

data = np.array([2, 4, 4, 8])
rng = np.random.default_rng(0)
means = []
for i in range(8):
    sample = rng.choice(data, size=4, replace=True)
    means.append(float(sample.mean()))
print([round(m, 3) for m in means])
print(round(min(means), 3), round(max(means), 3))
```
