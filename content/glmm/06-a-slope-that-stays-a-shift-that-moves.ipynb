# A slope that stays, a shift that moves

What stays the same across clusters when only the intercept shifts?

$$
\text{prediction} = (\text{intercept} + \text{shift}) + \text{slope} \times x
$$

The intercept is 3 and the slope is 2. Cluster A's shift is $-1$, so A's line is $2 + 2x$. At $x = 0$ the prediction is 2. At $x = 2$ the prediction is 6. The change from $x = 0$ to $x = 2$ is 4, which is the slope times 2.

Cluster B can have a different shift and the same slope. The shift moves the line up or down. It does not tilt it. The slope is the fixed effect in this illustration: it is shared. The lines below are drawn from these chosen numbers, not from a fitted model.


## One slope, two heights

The cell predicts for cluster A at two values of $x$, and draws both clusters with the same slope. The first plot in a session can take a moment while the plotting library downloads.


```python
# The cluster shift changes the height of the line. The slope stays shared.
import matplotlib.pyplot as plt

intercept = 3
slope = 2
shift_a = -1
shift_b = 1
# A's height at x = 0 is the intercept plus A's shift.
prediction_a_0 = intercept + shift_a + slope * 0
# At x = 2 the shared slope has added 4.
prediction_a_2 = intercept + shift_a + slope * 2
print(prediction_a_0)
print(prediction_a_2)

# Two lines with the same slope and different shifts.
xs = [0, 1, 2]
line_a = []
line_b = []
for x in xs:
    line_a.append(intercept + shift_a + slope * x)
    line_b.append(intercept + shift_b + slope * x)
# Two lines with the same slope and different heights.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(xs, line_a, marker="o", label="Cluster A")
ax.plot(xs, line_b, marker="o", label="Cluster B")
# The horizontal axis is x.
ax.set_xlabel("x")
# The vertical axis is the prediction.
ax.set_ylabel("Prediction")
# A key for the two clusters.
ax.legend()
# Show the figure under this cell.
plt.show()
```
