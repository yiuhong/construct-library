# A logit plus a cluster shift

How does a cluster shift enter a model for a probability?

The yes-or-no model used a line on the logit scale. A generalised linear mixed model adds the cluster shift on that same scale:

$$
\eta = \text{fixed linear predictor} + \text{cluster shift}
$$

$$
p = \frac{1}{1 + e^{-\eta}}
$$

Set the fixed linear predictor to 0. A shift of 0 leaves the probability at 0.5. A shift of $\log 3$ makes the odds 3, so the probability is 0.75. The shift moves the probability. It is not itself a probability.

The shift is chosen so the probability can be read. This lesson does not fit a mixed model, and a shift estimated from two clusters would be a rough description of those clusters.

Teaching note. Breslow, N. E., & Clayton, D. G. (1993). Approximate inference in generalized linear mixed models. *Journal of the American Statistical Association, 88*(421), 9-25. https://doi.org/10.1080/01621459.1993.10594284


## Read the probability

`probability` adds the chosen shift to a fixed linear predictor of 0, then maps that line to a probability.


```python
# A cluster shift on the logit scale. The fixed linear predictor is 0 in this illustration.
import math

def probability(cluster_shift):
    # The line is the fixed part plus the cluster shift.
    linear_predictor = 0 + cluster_shift
    # Map the line to a probability.
    return 1 / (1 + math.exp(-linear_predictor))

print(probability(0))
print(probability(math.log(3)))
```
