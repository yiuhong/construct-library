# Reading an estimate with care

What can the shared slope support on this panel, and what can it not support?

You will read the slope of 1 as a change in the stated week equation, set it beside one person's own rise, and count the clusters the illustration uses. By the end you should be able to keep the slope attached to that equation, and to say why six people do not make it a population estimate.

A number written into an equation reports what that equation does. It does not, on its own, measure a wider set of people. The panel was chosen so the arithmetic could be checked by hand.

The week equation used a slope of 1, chosen so each extra week added 1 to the stated score. P1's first and last weeks sat on that line. P4's did not. The practice shifts were chosen in the same way. None of those numbers was estimated from the six people.


## The slope in the stated equation

The slope is 1. Increasing the week by 1 adds 1 to the stated prediction, after the shift has set the height. From week 1 to week 4 is three steps, so the stated score rises by 3 for every person.

P1's quiz goes from 1 to 4, a rise of 3, which matches that stated rise. P4's quiz goes from 2 to 7, a rise of 5, which does not. The shared slope is not a description of every person's own change. It is the step that was written into the equation.


```python
# The stated rise over three weeks, set beside two people's own rises.
slope = 1
# Week 4 is three steps on from week 1.
stated_rise = slope * 3
# P1's quiz scores in week 1 and week 4.
p1_rise = 4 - 1
# P4's quiz scores in week 1 and week 4.
p4_rise = 7 - 2
print(stated_rise)
print(p1_rise)
print(p4_rise)
```

## Six clusters, and not a population

The panel has six people, so the illustration uses six clusters. Hold P6 out and five clusters remain. The written one-step change is still 1, because nothing was estimated from the list of people. Dropping a person does not turn the slope into an estimate, and neither does keeping all six.

What the 1 says here: in the stated equation, one extra week adds 1 to the prediction.

What it does not say: it does not measure a population of many clusters. Six chosen shifts do not estimate how spread out person shifts are. A cautious reading keeps the slope attached to this repeated-quiz panel. A published mixed model would estimate the slope and the spread of the shifts, and it would report uncertainty. Those steps are not done here. A slope from this panel would still be a small illustration if they were.


```python
# How many clusters the illustration uses, and the written one-step change.
people = ["P1", "P2", "P3", "P4", "P5", "P6"]
n_clusters = len(people)
# Holding one person out leaves the others. The slope is not re-estimated.
held_out = ["P6"]
n_remaining = n_clusters - len(held_out)
slope = 1
one_step = slope * 1
print(n_clusters)
print(n_remaining)
print(one_step)
```
