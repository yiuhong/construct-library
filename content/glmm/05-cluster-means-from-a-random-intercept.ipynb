# Cluster means from a random intercept

What does a random intercept change in the cluster means?

A random intercept is the cluster shift from the previous notebook, added to a fixed intercept. The fixed intercept is the shared part. Each cluster mean is that intercept plus its own shift.

The fixed intercept is 10. Cluster A's shift is $-2$, so A's cluster mean is 8. Cluster B's shift is $+2$, so B's cluster mean is 12. The difference of the cluster means is 4, which is the gap between the two shifts. The intercept itself is shared, so it cancels in the difference.

Again the shifts are chosen for the illustration. They are not the output of a fit.


## Intercept plus shift

The cell builds each cluster mean and subtracts them.


```python
# Each cluster mean is the fixed intercept plus a chosen shift.
fixed_intercept = 10
shift_a = -2
shift_b = 2
# Cluster means.
mean_a = fixed_intercept + shift_a
mean_b = fixed_intercept + shift_b
# The shared intercept cancels, leaving the gap between the shifts.
difference = mean_b - mean_a
print(mean_a)
print(difference)
```
