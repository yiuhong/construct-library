# A yes-or-no outcome inside clusters

How do I summarise a 0/1 outcome when rows share a cluster?

This notebook is in Advanced modelling with GLMM, in the module Generalised mixed models. A yes is 1 and a no is 0. Person A has `[1, 1, 0]`, so A has 2 yes outcomes and the proportion is $2/3$. Person B has `[0, 0, 0]`, so the proportion is 0.

The mean of the two person proportions is $(2/3 + 0) / 2 = 1/3$. Pooling the six rows gives $2/6 = 1/3$ as well, because the clusters are the same size. The pooled proportion still comes from two people. The next notebook puts a cluster shift on the line that states such a probability.


## A proportion per person

The cell counts the yes outcomes and averages the two proportions.


```python
# A yes count and a proportion for each person, then the mean of the two proportions.
person_a = [1, 1, 0]
person_b = [0, 0, 0]
# How many yes outcomes person A has. A yes is stored as 1.
yes_a = sum(person_a)
# Each proportion divides that person's yes count by that person's number of rows.
proportion_a = yes_a / len(person_a)
proportion_b = sum(person_b) / len(person_b)
# Average the cluster proportions, one per person.
mean_proportion = (proportion_a + proportion_b) / 2
print(yes_a)
print(proportion_a)
print(mean_proportion)
```
