# Rows, clusters, and two kinds of variation

What is a cluster when the same person has a quiz row each week?

You will count the people and the rows in one repeated-quiz panel, compare a mean of rows with a mean of people, and split a gap between people from a gap inside one person. By the end you should be able to say how many clusters the table holds, and which of those two gaps sits between clusters.

A row here is one person in one week. Several rows share a person, so a count of rows is not a count of people. The gap between two people's means is also a different object from the gap between two weeks of one person.

Seven people each had one questionnaire rating, and the row was the person. This panel repeats a short quiz: the same person has a row in each of four weeks, and the person is the cluster.


## Six people, four weeks

Six people, P1 to P6, took the same quiz once a week for four weeks. The score is out of 10. A score of 5 or more is recorded as a pass. Practice is the number of practice items completed that week, and a zero means none. The numbers were chosen so the arithmetic in this course stays on whole numbers. P1's four scores are 1, 3, 4, and 4. There are 24 rows and 6 people. Each person is one cluster.


```python
# Six people, four weekly scores each. The numbers were chosen for teaching.
scores = {
    "P1": [1, 3, 4, 4],
    "P2": [2, 4, 5, 5],
    "P3": [3, 5, 6, 6],
    "P4": [2, 5, 6, 7],
    "P5": [4, 6, 7, 7],
    "P6": [5, 7, 8, 8],
}
# One cluster per person.
n_clusters = len(scores)
# One row per week.
n_rows = 0
for person in scores:
    n_rows = n_rows + len(scores[person])
print(n_clusters)
print(n_rows)
```

## The mean of the rows, and the mean of the people

The 24 scores sum to 120, so the mean of the rows is $120/24 = 5$. The person means are 3, 4, 5, 5, 6, and 7. Their mean is also 5, because every person has four weeks.

The matching averages do not mean there are 24 people. A statement about people uses six clusters. Treating each row as its own person would overstate how many people were measured.


```python
# The mean of every row, and the mean of the six person means.
scores = {
    "P1": [1, 3, 4, 4],
    "P2": [2, 4, 5, 5],
    "P3": [3, 5, 6, 6],
    "P4": [2, 5, 6, 7],
    "P5": [4, 6, 7, 7],
    "P6": [5, 7, 8, 8],
}
all_scores = []
person_means = []
for person in scores:
    rows = scores[person]
    # Keep every weekly score for the row mean.
    all_scores.extend(rows)
    # One mean per person.
    person_means.append(sum(rows) / len(rows))
row_mean = sum(all_scores) / len(all_scores)
mean_of_people = sum(person_means) / len(person_means)
print(person_means)
print(row_mean)
print(mean_of_people)
```

## Between the people, and inside one person

P6's mean is 7 and P1's mean is 3, so the gap between those clusters is $7 - 3 = 4$. Inside P1 the scores run from 1 to 4, so the within-cluster range is $4 - 1 = 3$. The gap between the people is not the range inside one person. A summary that ignores the person mixes the two.

The first plot in a session can take a moment while the plotting library downloads.


```python
# A gap between two person means, and the range inside P1.
import matplotlib.pyplot as plt

score_p1 = [1, 3, 4, 4]
score_p6 = [5, 7, 8, 8]
# Each person mean.
mean_p1 = sum(score_p1) / len(score_p1)
mean_p6 = sum(score_p6) / len(score_p6)
# Between clusters: how far the two means sit apart.
between = mean_p6 - mean_p1
# Within P1: the larger score minus the smaller.
within_p1 = max(score_p1) - min(score_p1)
print(between)
print(within_p1)

# Week on the horizontal axis, so each person's four scores stay together.
weeks = [1, 2, 3, 4]
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter(weeks, score_p1, label="P1")
ax.scatter(weeks, score_p6, label="P6")
ax.set_xticks(weeks)
# The horizontal axis is the week inside the person.
ax.set_xlabel("Week")
# The vertical axis is the quiz score.
ax.set_ylabel("Score")
ax.legend()
plt.show()
```
