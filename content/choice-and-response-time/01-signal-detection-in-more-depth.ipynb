# Signal detection in more depth

How can I see hits and false alarms move together when the criterion moves?

A yes/no task has two classes of trial: noise alone, and noise plus a signal. The observer answers yes when an internal score passes a criterion. Moving the criterion changes both kinds of yes.

A hit is a yes when the signal was present. A false alarm is a yes when it was absent. From counts, 12 hits in 16 signal trials is a hit rate of 0.75. Four false alarms in 16 noise trials is a false-alarm rate of 0.25.

Equal-variance signal detection puts the noise scores on a normal curve with mean 0 and standard deviation 1, and the signal scores on a normal curve with mean $d'$ and the same standard deviation. $d'$ is the distance between the two means, in units of that standard deviation. Treating the yes/no answer as a criterion on two overlapping distributions is signal detection theory (Green & Swets, 1966).

Green, D. M., & Swets, J. A. (1966). *Signal detection theory and psychophysics*. Wiley.

The next cell uses $d' = 1$, so the signal mean is 1. A score passes a criterion $c$ when it is above $c$. The false-alarm rate is the area of the noise curve above $c$. The hit rate is the area of the signal curve above $c$. The first import of `scipy` in a session downloads that library.


```python
from scipy.stats import norm

d_prime = 1
for criterion in [0.0, 0.5, 1.0]:
    false_alarm = norm.sf(criterion)
    hit = norm.sf(criterion - d_prime)
    print(criterion, round(float(false_alarm), 3), round(float(hit), 3))
```

## Three points on an ROC

As the criterion moves up, both rates fall. The observer says yes less often on signal trials and less often on noise trials. $d'$ did not change. The curves stayed where they were. Only the cutoff moved.

A plot of hit rate against false-alarm rate, one point per criterion, is a short ROC. The dashed line is where the two rates are equal, which is what you get if the signal curve sits on top of the noise curve ($d' = 0$). With $d' = 1$ the points sit above that line: at a given false-alarm rate, the hit rate is higher. The first plot in a session can take a moment.


```python
import matplotlib.pyplot as plt
from scipy.stats import norm

criteria = [0.0, 0.5, 1.0]
false_alarms = [float(norm.sf(c)) for c in criteria]
hits = [float(norm.sf(c - 1)) for c in criteria]

fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(false_alarms, hits, marker="o")
ax.plot([0, 1], [0, 1], linestyle="--")
ax.set_xlabel("False-alarm rate")
ax.set_ylabel("Hit rate")
plt.show()
```
