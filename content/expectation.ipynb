# Expected value of a fair coin

Let X be the payoff of a fair coin: heads X = 1, tails X = 0, each with probability 1/2.

E[X] = 1 · ½ + 0 · ½ = 0.5


```python
import numpy as np
import matplotlib.pyplot as plt

x, p = np.array([0, 1]), np.array([0.5, 0.5])
E = np.sum(x * p)
print(f"E[X] = {E}")

fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.bar(["Tails (0)", "Heads (1)"], p)
ax.set_ylabel("Probability")
ax.set_ylim(0, 1)
ax.set_title("Fair coin: P(X) and expected value")
ax.axhline(E, color="C3", ls="--", label=f"E[X] = {E}")
ax.legend(loc="upper right")
plt.show()
```
