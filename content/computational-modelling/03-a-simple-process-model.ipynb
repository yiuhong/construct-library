# A simple process model

How can I get a choice and a response time from a small race?

Some models do not start from a line through a cloud of points. They state a process: how a choice and a response time could be produced.

This lesson uses a small sketch with two counters, one for choice A and one for choice B. Each counter starts at 0 and moves in a straight line. The slope of that line is the counter's rate. The choice is whichever counter reaches a shared threshold first. The response time is the time of that crossing, plus a short delay that is not part of the race.

$$
\text{time} = \frac{\text{threshold}}{\text{rate}} + \text{delay}
$$

The rates in this lesson are positive, so each counter does finish. A larger rate finishes sooner.


## One trial

Threshold 2, delay 0.2, rate A of 2, and rate B of 1. The next cell prints both finish times, then the choice and the response time. The response time is the smaller finish time.


```python
# Time to reach the threshold, then which counter finishes first. The counter with the smaller finish time wins.
# Finish time is the threshold divided by the rate, plus a delay before the race starts.
def finish_time(threshold, rate, delay):
    return threshold / rate + delay

# How much evidence a counter needs.
threshold = 2
# Time before either counter starts, shared by both.
delay = 0.2
# Counter A, with the higher rate.
time_a = finish_time(threshold, 2, delay)
# Counter B, with the lower rate.
time_b = finish_time(threshold, 1, delay)
# A's finish time, then B's.
print(time_a)
print(time_b)
# The smaller finish time wins.
if time_a < time_b:
    # Name the winner and its finish time.
    print("A", time_a)
else:
    print("B", time_b)
```

## The race on one set of axes

The next cell plays the race. Drag the slider under the figure, or press Play. The grey line is the threshold. The delay is added after a counter crosses it, so it is not on this axis. The first plot in a session can take a moment while the plotting library downloads.


```python
# Both counters filling towards the threshold. A rises faster than B. The grey line is the threshold they are racing to.
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# The evidence each counter is racing towards.
threshold = 2
# Time points for the frames, from 0 to 2.
times = []
# Eleven frames. i/5 runs from 0 to 2.
for i in range(0, 11):
    times.append(i / 5)

# The figure the player draws into.
fig, ax = plt.subplots(figsize=(6.4, 3.4), dpi=80)
# Fix the time axis.
ax.set_xlim(0, 2)
ax.set_ylim(0, 4.2)
# The threshold, as a grey horizontal line.
ax.axhline(threshold, color="#cccccc", linewidth=1)
# An empty line for counter A. The frames fill it.
line_a, = ax.plot([], [], label="A, rate 2")
# An empty line for counter B.
line_b, = ax.plot([], [], label="B, rate 1")
# Name the horizontal axis.
ax.set_xlabel("Time in the race")
# Name the vertical axis.
ax.set_ylabel("Evidence")
# A key, kept to the left so it does not cover the lines.
ax.legend(loc="upper left")

# Show both counters from the start up to this frame.
def draw(frame):
    # The time points revealed so far.
    shown = times[: frame + 1]
    # A gains 2 units of evidence per unit of time.
    line_a.set_data(shown, [2 * t for t in shown])
    # B gains 1 unit per unit of time.
    line_b.set_data(shown, [t for t in shown])
    return line_a, line_b

# Build a player under the figure. Press play, or drag the slider, to step through the frames.
anim = FuncAnimation(fig, draw, frames=len(times), interval=250, blit=False)
player = HTML(anim.to_jshtml())
# Close the static figure so only the player is shown.
plt.close(fig)
# The player under the cell.
player
```

Counter A reaches the threshold at time 1. Counter B reaches it at time 2. Adding the delay of 0.2 gives the finish times from the cell above.

The counters in this lesson are a teaching sketch: fixed rates, a shared threshold, and a delay. The published model in this family is the linear ballistic accumulator, in which independent accumulators race in straight lines.

Teaching note. Brown, S. D., & Heathcote, A. (2008). The simplest complete model of choice response time: Linear ballistic accumulation. *Cognitive Psychology, 57*(3), 153-178. https://doi.org/10.1016/j.cogpsych.2007.12.002
