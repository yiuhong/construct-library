# Linear regression as a model

How can I treat a fitted straight line as a model?

A linear regression model predicts $y$ from $x$ with two parameters. The slope $a$ is how much the prediction changes when $x$ increases by 1. The intercept $b$ is the prediction when $x$ is 0.

$$
\hat{y} = b + ax
$$

The residual is still observed minus predicted. The model is the line. The data are the points. The points do not have to sit on the line.


## A line you choose, and the line that fits

Three points: $(1, 4)$, $(2, 4)$, and $(3, 8)$. The hand-chosen line $b = 1$, $a = 2$ predicts 3, 5, and 7. The residuals are 1, $-1$, and 1. The sum of squared residuals is 3.

Least squares picks the slope and intercept that make that sum as small as possible.

$$
a = \frac{n\sum xy - (\sum x)(\sum y)}{n\sum x^2 - (\sum x)^2}
$$

$$
b = \bar{y} - a\bar{x}
$$

The next cell fits that line, prints both sums of squares, and draws the points and the two lines. The first plot in a session can take a moment while the plotting library downloads.


```python
# A line chosen by hand, set beside the line that fits these points. The fit is the line with the smaller sum of squared misses.
import matplotlib.pyplot as plt

# The inputs.
xs = [1, 2, 3]
# The observed outcomes.
ys = [4, 4, 8]
# How many points. The formulas below use this count.
n = 3

# Running totals for the least-squares formulas.
sum_x = 0
sum_y = 0
sum_xy = 0
sum_xx = 0
# Add each point into those totals.
for i in range(n):
    sum_x = sum_x + xs[i]
    sum_y = sum_y + ys[i]
    sum_xy = sum_xy + xs[i] * ys[i]
    sum_xx = sum_xx + xs[i] ** 2

# The fitted slope, from those totals.
slope = (n * sum_xy - sum_x * sum_y) / (n * sum_xx - sum_x ** 2)
# The fitted intercept, from the two means and that slope.
intercept = sum_y / n - slope * (sum_x / n)

# Squared misses for a line with intercept b and slope a.
def sum_of_squares(b, a):
    # A running total of those squared misses.
    total = 0
    # One point at a time.
    for i in range(n):
        residual = ys[i] - (b + a * xs[i])
        total = total + residual ** 2
    return total

# The fitted intercept and slope, then the two sums of squares.
print(round(intercept, 3), round(slope, 3))
print(round(sum_of_squares(1, 2), 3))
print(round(sum_of_squares(intercept, slope), 3))

# The observed points, the hand-chosen line, and the fitted line, on one pair of axes.
# A figure for the points and both lines.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The observed points, not joined.
ax.plot(xs, ys, marker="o", linestyle="none", label="Observed")
# Two x values, enough to draw each line across the figure.
line_x = [0, 4]
# The line that was chosen by hand.
ax.plot(line_x, [1 + 2 * x for x in line_x], label="Hand model")
# The fitted line.
ax.plot(line_x, [intercept + slope * x for x in line_x], label="Least squares")
# Name the horizontal axis.
ax.set_xlabel("x")
# Name the vertical axis.
ax.set_ylabel("y")
# A key for the points and the two lines.
ax.legend()
# Show the figure under this cell.
plt.show()
```

The fitted line has the smaller sum of squares. That is what least squares means: of the lines this rule considers, it keeps the one with the smallest squared error. A different rule could keep a different line.
