# Simulating data from a model

How can I run a model forward and write down what it produces?

To simulate is to run the model forward. You set the parameters, you supply the inputs, and you record the observations the model produces. Those observations are not measurements from an experiment. They are what the model says could have happened.


## Several trials of the race

Use the race from the process-model lesson. The threshold is 1 and the delay is 0.2 on every trial. Each trial brings its own rates. The smaller finish time wins, and that finish time is the response time.


```python
# Four races. In each race the counter that finishes sooner wins, and the print names that winner and its time.
# The same finish-time rule as the process-model lesson.
def finish_time(threshold, rate, delay):
    return threshold / rate + delay

# Evidence needed to finish.
threshold = 1
# Delay before each race starts.
delay = 0.2
# A's rate in each of the four races.
rates_a = [2, 1, 2, 0.5]
# B's rate in those same races.
rates_b = [1, 2, 1, 2]

# One race at a time.
for i in range(4):
    # How long A takes in this race.
    time_a = finish_time(threshold, rates_a[i], delay)
    # How long B takes.
    time_b = finish_time(threshold, rates_b[i], delay)
    # The smaller time wins.
    if time_a < time_b:
        print("A", round(time_a, 3))
    else:
        print("B", round(time_b, 3))
```

## A prediction plus noise

A line plus a noise term is another model you can run forward.

$$
\text{score} = b + ax + \text{noise}
$$

The same intercept and slope can produce different scores, because the noise need not be the same each time. The next cell adds a supplied noise to three points. It then draws four noise values from a normal distribution centred at 0, with spread 0.2. The seed makes a later run of this cell draw the same four values. The first time this session imports numpy, that library can take a moment to download.


```python
# A straight line plus a listed disturbance, then a few random disturbances. The listed ones are fixed. The random ones change only if the start value changes.
import numpy as np

# Three inputs.
xs = [1, 2, 3]
# A chosen disturbance for each input.
noises = [-0.2, 0.1, 0.4]
# Height of the line at x = 0.
intercept = 1
# Rise of the line per step in x.
slope = 2
# One input at a time.
for i in range(3):
    # Line plus the listed disturbance.
    print(intercept + slope * xs[i] + noises[i])

# Draw four disturbances from a normal curve. The fixed start makes the same draw happen every time this cell runs.
# A random source with a fixed start, so the draw repeats.
rng = np.random.default_rng(0)
# Four disturbances, centred on 0, with a typical size of 0.2.
drawn = rng.normal(loc=0.0, scale=0.2, size=4)
# Those four disturbances, to three decimals.
print(np.round(drawn, 3))
```

## One score at a time

The model line has intercept 1 and slope 2. Each step of the player adds one simulated score. The first three noises are the ones printed above. Drag the slider under the figure, or press Play.


```python
# One simulated score at a time, along the line. Each score is the line plus a listed disturbance, and each frame reveals one more score.
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Eight inputs.
xs = [1, 2, 3, 4, 5, 6, 7, 8]
# One disturbance per input.
noises = [-0.2, 0.1, 0.4, -0.3, 0.2, -0.1, 0.3, 0.0]
# The line's height at 0.
intercept = 1
# The line's rise per step in x.
slope = 2
# The simulated scores, filled before the player starts.
scores = []
# Build every score first. The player only reveals them.
for i in range(len(xs)):
    # Line plus disturbance at this input.
    scores.append(intercept + slope * xs[i] + noises[i])

# The figure the player draws into.
fig, ax = plt.subplots(figsize=(6.4, 3.4), dpi=80)
# The line itself, in grey, across the whole axis.
ax.plot([0, 9], [intercept, intercept + slope * 9], color="#cccccc", label="Model")
# The simulated scores, starting empty.
points, = ax.plot([], [], marker="o", linestyle="none", label="Simulated")
# Fix the axes so they stay still while points appear.
ax.set_xlim(0, 9)
ax.set_ylim(0, 20)
# Name the horizontal axis.
ax.set_xlabel("x")
# Name the vertical axis.
ax.set_ylabel("score")
# A key for the line and the points.
ax.legend()

# Reveal one more score on each frame.
def draw(frame):
    # How many scores to show.
    k = frame + 1
    # The first k inputs and their scores.
    points.set_data(xs[:k], scores[:k])
    return points,

# Build a player under the figure. Press play, or drag the slider, to step through the frames.
anim = FuncAnimation(fig, draw, frames=len(xs), interval=400, blit=False)
player = HTML(anim.to_jshtml())
# Close the static figure so only the player is shown.
plt.close(fig)
# The player under the cell.
player
```
