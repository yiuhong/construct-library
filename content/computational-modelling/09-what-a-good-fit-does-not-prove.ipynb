# What a good fit does not prove

How can I tell what a close fit still leaves open?

A fit can match a number closely and still leave the parameters unsettled. It can be perfect on the fitted points and still say nothing about the next point.


## Two settings, one prediction

The race in the process-model lesson finishes at

$$
\text{time} = \frac{\text{threshold}}{\text{rate}} + \text{delay}
$$

One setting is threshold 2, rate 2, and delay 0.2. Another is threshold 4, rate 4, and the same delay. The threshold doubled and the rate doubled.


```python
# Doubling the threshold and the rate leaves the finish time unchanged. The two calls are different settings with the same race.
# Finish time is threshold divided by rate, plus the delay.
def finish_time(threshold, rate, delay):
    return threshold / rate + delay

# Threshold 2 and rate 2.
print(finish_time(2, 2, 0.2))
# Both doubled. The delay is unchanged.
print(finish_time(4, 4, 0.2))
```

The two settings finish together. From that one finish time, with the delay known, you can see the ratio of threshold to rate. You cannot see the threshold and the rate as separate numbers. They are not identifiable from that one time. Matching the time does not prove that the threshold was 2 rather than 4.

If the other counter's rate scales in the same way, the choice stays the same as well.


## A perfect fit on the fitted points

Three scores, at $x = 1, 2, 3$, are 3, 5, and 8. A line with intercept $1/3$ and slope 2.5 has a small sum of squared residuals, and it makes a prediction at $x = 4$.

A memory model stores the three scores and predicts those $x$ values by lookup. Its error on the three fitted points is 0. It uses one stored score per point. It has no rule for $x = 4$.

The next cell prints the line's sum of squares, the line's prediction at $x = 4$, and the memory model's error on the three fitted points.


```python
# The line's error on the fitted points, its prediction one step further, and the memory model's error. The memory model stores the fitted scores and has nothing to say about a new x except by looking them up.
# The fitted inputs.
xs = [1, 2, 3]
# The fitted outcomes.
ys = [3, 5, 8]
# Intercept of the fitted line.
intercept = 1 / 3
# Slope of the fitted line.
slope = 2.5
# Squared error of that line on the fitted points.
total = 0
# One fitted point at a time.
for i in range(3):
    total = total + (ys[i] - (intercept + slope * xs[i])) ** 2
# The line's error on the fitted points.
print(round(total, 3))
# The line's prediction one step past the last fitted x.
print(round(intercept + slope * 4, 3))
# On the fitted points the memory model looks the scores up, so its error there is zero. It has no rule for a new x.
# The memory model's error on the fitted points.
print(0)
```

An error of 0 on the fitted points does not prove that the memory model is the process that produced the scores. Fitting the line does not prove that the line is that process either. A fit is a score under a rule you chose. It is not, by itself, evidence that the rule is the one that generated the data.
