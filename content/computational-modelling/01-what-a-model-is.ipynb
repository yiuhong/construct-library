# What a model is

How can I write a model as a rule that returns a prediction?

A model is a rule. It takes a parameter and an input, and it returns a prediction. The parameter is a number you choose. Change it, and the prediction changes.

Take a score that grows with hours of practice. One model is

$$
\text{predicted score} = a \times \text{hours}
$$

The input is the hours. The parameter is $a$. With $a = 2$, one hour predicts a score of 2, and three hours predict a score of 6.


```python
def predict(hours, a):
    return a * hours

for hours in [1, 2, 3]:
    print(hours, predict(hours, 2))
```

## The leftover

The residual is the leftover: observed score minus predicted score. A residual of 0 means the prediction matched that observation. A positive residual means the score was higher than the model said. A negative residual means it was lower.

Squaring a residual makes it positive, so a miss on either side counts. The sum of the squared residuals is one measure of how far the model is from these scores.

The next cell keeps $a = 2$. The scores for 1, 2, and 3 hours are 3, 5, and 6.


```python
hours = [1, 2, 3]
scores = [3, 5, 6]
a = 2

residuals = []
for i in range(3):
    predicted = a * hours[i]
    residuals.append(scores[i] - predicted)

print(residuals)

total = 0
for residual in residuals:
    total = total + residual ** 2
print(total)
```
