# Hits, false alarms, and confidence

How can I count hits, false alarms, and a stricter yes?

A yes-or-no judgment can be wrong in two directions. The signal was present and the response was yes: a hit. The signal was absent and the response was yes: a false alarm. A miss is a present signal called no. A correct rejection is an absent signal called no.

The hit rate is the hits divided by the number of signal trials. The false-alarm rate is the false alarms divided by the number of noise trials. Those two rates are different facts. A single "percent correct" throws the difference away.


## A criterion

In the signal-detection account, evidence on a signal trial tends to sit higher than evidence on a noise trial, and the two piles overlap. The criterion is a cutoff. Evidence above it is called yes. Move the cutoff up, and both the hit rate and the false-alarm rate fall. How far apart the piles sit is a different fact from where the cutoff is placed.

The next cell counts one small table. `signal` is 1 when the signal was present. `yes` is 1 when the response was yes. `sure` is 1 when that yes also cleared a stricter cutoff.

Teaching note. Hits, false alarms, and a criterion are the signal-detection account of a yes-or-no judgment. Green, D. M., & Swets, J. A. (1966). *Signal detection theory and psychophysics*. Wiley.


```python
# Count hits on signal trials and false alarms on noise trials. A hit is a yes on a signal trial. A false alarm is a yes on a noise trial.
# 1 when the trial was a signal, 0 when it was noise.
signal = [1, 1, 1, 1, 0, 0, 0, 0]
# 1 when the response was yes.
yes = [1, 1, 0, 1, 0, 1, 0, 0]
# 1 when that yes was given with high confidence.
sure = [1, 0, 0, 1, 0, 0, 0, 0]

# Counts, all starting at zero before the loop.
hits = 0
signal_trials = 0
false_alarms = 0
noise_trials = 0
sure_hits = 0
# One trial at a time.
for i in range(8):
    # A signal trial.
    if signal[i] == 1:
        # Count the signal trial.
        signal_trials = signal_trials + 1
        # A yes on this signal trial is a hit.
        if yes[i] == 1:
            # Count the hit.
            hits = hits + 1
            # And, among hits, count the ones given as sure.
            if sure[i] == 1:
                sure_hits = sure_hits + 1
    # A noise trial.
    else:
        # Count the noise trial.
        noise_trials = noise_trials + 1
        # A yes on noise is a false alarm.
        if yes[i] == 1:
            false_alarms = false_alarms + 1

# Hits, how many signal trials, and the hit rate. Then the false alarms, and the sure hits.
print(hits, signal_trials, hits / signal_trials)
print(false_alarms, noise_trials, false_alarms / noise_trials)
print(sure_hits)
```

## The stricter cutoff is the confidence

`sure` is not a new kind of evidence. It is a second cutoff, farther into the yes region. A hit that also clears it is a high-confidence hit. In this table, 2 of the 3 hits are high confidence, and the false alarm is not. Confidence, in this sketch, is how far the evidence sits past the first cutoff.

One picture of the overlap puts a noise pile at 0 and a signal pile at 1. Both have spread 1, and the yes cutoff is at 0. The false-alarm rate is the part of the noise pile above the cutoff, which is 0.5, so its $z$ score is 0. The hit rate is the part of the signal pile above the same cutoff. With spread 1, its $z$ score is the distance from the cutoff to the centre of the signal pile, which is 1.

$$
d' = z(\text{hit rate}) - z(\text{false-alarm rate}) = 1 - 0 = 1
$$

$d'$ reads the separation back out of the two rates. Here it returns the separation that was put in. The next cell prints the hit rate for that arrangement, then the false-alarm rate.


```python
# How much of a normal curve sits above a cutoff. The curve is centred at the centre you pass in, and the cutoff is measured from there.
import math

# The share of the curve that sits above the cutoff.
def proportion_above(cutoff, centre):
    # How far the cutoff is above the centre, in standard units.
    z = cutoff - centre
    return 1 - 0.5 * (1 + math.erf(z / math.sqrt(2)))

# A cutoff at 0 on a curve centred at 1, then a cutoff sitting on the centre.
print(round(proportion_above(0, 1), 3))
print(proportion_above(0, 0))
```
