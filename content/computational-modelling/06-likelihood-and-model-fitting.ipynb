# Likelihood and model fitting

Likelihood and fitting, in Basic mathematics, picks the parameter of a stated rule. The rule there is a coin. Here the rule is a small model of a choice, and the parameter is a rate.

Two choices, A and B. Rate B is fixed at 1. Rate A is the parameter, written $v$. The model says

$$
P(A) = \frac{v}{v + 1}
$$

$$
P(B) = \frac{1}{v + 1}
$$

A higher rate for A makes A more probable. It does not make A certain.


## The probability of these choices

Four independent choices came out A, A, A, B. The likelihood is the probability of that sequence if $v$ takes a chosen value.

$$
L(v) = P(A)^3 \times P(B)
$$

The next cell evaluates $L$ at $v = 1$, $v = 2$, and $v = 3$. The fit on this grid is the value with the largest likelihood.


```python
def likelihood(v):
    p_a = v / (v + 1)
    p_b = 1 / (v + 1)
    return p_a ** 3 * p_b

for v in [1, 2, 3]:
    print(v, round(likelihood(v), 4))
```

## The value the rule prefers

On this grid the likelihood is highest at $v = 3$. That matches the observed proportion, written as a rate. $P(A) = 3/4$, so $v / (v + 1) = 3/4$, and $v = 3$.

A different grid, or a different model, could prefer a different value. "Best fit" means the winner under the rule you stated.
