# Incremental learning

A learner holds a prediction $V$. After an outcome, the prediction moves part of the way toward that outcome. The gap between the outcome and the prediction is the prediction error. A step size between 0 and 1 sets how much of that gap is kept.

$$
V_{\text{new}} = V + \alpha (\lambda - V)
$$

$\alpha$ is the step size. $\lambda$ is the outcome. For one cue that is present on the trial, this is the Rescorla–Wagner update (Rescorla & Wagner, 1972).

Rescorla, R. A., & Wagner, A. R. (1972). A theory of Pavlovian conditioning: Variations in the effectiveness of reinforcement and nonreinforcement. In A. H. Black & W. F. Prokasy (Eds.), *Classical conditioning II: Current research and theory* (pp. 64–99). Appleton-Century-Crofts.

Start at $V = 0$, with $\alpha = 0.5$ and $\lambda = 1$ on every trial. The updates get smaller because the error gets smaller. The prediction approaches 1. It does not jump there in one trial when $\alpha$ is below 1.


```python
V = 0.0
alpha = 0.5
for trial in range(1, 5):
    error = 1 - V
    V = V + alpha * error
    print(trial, error, V)
```

## What the step size does

After trial 1 the prediction is 0.5 and the error was 1. After trial 2 the prediction is 0.75 and the error was 0.5. A larger $\alpha$ would have moved further on trial 1. An $\alpha$ of 1 would replace the prediction with the outcome in one step. An $\alpha$ of 0 would leave the prediction where it started.

The next cell is the same update with $\alpha = 1$ and with $\alpha = 0$, for one trial, starting from 0, with outcome 1.


```python
print(0 + 1 * (1 - 0))
print(0 + 0 * (1 - 0))
```
