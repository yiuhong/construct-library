# A basic Bayesian update

How can I update the probability of a hypothesis after an observation?

Two hypotheses, A and B, start equally likely. An observation is more likely under one of them. Multiply each prior by the likelihood of the observation, then divide by the total, so the two results still add to 1.

$$
P(A \mid \text{data}) = \frac{P(\text{data} \mid A)\, P(A)}{P(\text{data} \mid A)\, P(A) + P(\text{data} \mid B)\, P(B)}
$$

Prior $1/2$ each. The observation has likelihood $3/4$ under A and $1/4$ under B.

$$
P(A \mid \text{data}) = \frac{(3/4)(1/2)}{(3/4)(1/2) + (1/4)(1/2)} = \frac{3}{4}
$$

The prior was $1/2$. One observation that is three times as likely under A moves the probability of A to $3/4$.


```python
def posterior(prior_a, like_a, like_b):
    prior_b = 1 - prior_a
    weight_a = like_a * prior_a
    weight_b = like_b * prior_b
    return weight_a / (weight_a + weight_b)

print(posterior(0.5, 0.75, 0.25))
```

## A second observation

A second observation, independent of the first and with the same two likelihoods, uses $3/4$ as the new prior for A. The prior for B is $1/4$.

$$
P(A \mid \text{both}) = \frac{(3/4)(3/4)}{(3/4)(3/4) + (1/4)(1/4)} = \frac{9}{10}
$$

The probability of A moves from 0.5 to 0.75 to 0.9. Each observation pulls in the same direction, and the pull is smaller once A is already likely. This is two hypotheses and a stated likelihood. It is not a fit of a likelihood the notebook has estimated from a sample.


```python
def posterior(prior_a, like_a, like_b):
    prior_b = 1 - prior_a
    weight_a = like_a * prior_a
    weight_b = like_b * prior_b
    return weight_a / (weight_a + weight_b)

first = posterior(0.5, 0.75, 0.25)
second = posterior(first, 0.75, 0.25)
print(first, second)
```
