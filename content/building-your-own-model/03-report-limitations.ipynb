# Report limitations

How can I say what a fitted number does not show?

A result sentence should say what was done and what was not. Next to a fitted number, three limits belong in the same paragraph:

- how many points were used
- whether a second start was tried
- whether the parameter was scored on points that were not used to fit it

Example: "The slope is 1, from 3 points, from one start, with no held-out point." That sentence can be true and still be thin. Three points and one start do not show that a second valley was absent. No held-out point means the fit has not been scored as a prediction.

If a neural summary in the same report was simulated, the sentence should say it was simulated.


```python
xs = [1, 2, 3]
ys = [1, 2, 2]
reported = 1

sse_reported = 0
for i in range(3):
    sse_reported = sse_reported + (ys[i] - reported * xs[i]) ** 2

total_xy = 0
total_xx = 0
for i in range(3):
    total_xy = total_xy + xs[i] * ys[i]
    total_xx = total_xx + xs[i] ** 2
best = total_xy / total_xx
print(round(sse_reported, 3), round(best, 3))
print(len(ys))
```

## Check the sentence against the points

The cell prints the sum of squared errors at the reported slope, then the least-squares slope through the origin. The reported slope is 1. The least-squares slope is not 1. A sentence that calls 1 the least-squares answer does not match these points, even before the limits above.

The print of `len(ys)` is the number of points, 3. The report already said three points. That part can be checked by counting. The parts that were not done, a second start and a held-out point, cannot be checked from this list. They have to be named as not done.
