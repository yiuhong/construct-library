# After a significant rank test

After a rank test that compares several groups or several conditions, what is the next question, and what does a small sketch still not do?

This notebook is in Non-parametric tests, in the module Comparing groups. Kruskal–Wallis and Friedman ask whether the rank totals differ as a set. They do not say which pair differs. A follow-up looks at pairs. This notebook sketches that look for the tables from the two previous lessons. It is not a post-hoc package, and it does not adjust for how many pairs are inspected.

The three groups are the chosen scores from Several independent groups. There are three pairs. On that table, `kruskal` printed a p-value of about 0.039 from a thin chi-square approximation.

| group | scores |
| --- | --- |
| no notes | 2, 3, 5 |
| short notes | 4, 6, 7 |
| worked example | 8, 9, 10 |

No notes against the worked example: every no-notes score is below every worked-example score, so the win count is 0. The exact two-sided p-value for a complete split of three against three is 0.1. Short notes against the worked example is the same kind of split. No notes against short notes has one win, and a p-value of 0.2. Those pairwise p-values are larger than the omnibus printout. With three scores in a group, even a complete split does not produce a very small exact p-value. None of the three calls is adjusted for being one of three.

A Friedman follow-up stays inside the person. The same four people as in Several related samples are compared on list against practice only, with `wilcoxon`. Every practice rating is higher, so the negative rank sum is 0. Two of the positive changes are tied at 4. The two-sided p-value on four pairs is 0.125. That is one pair of three, uncorrected. The other two condition pairs are not run.

Nine scores and four people, uncorrected pairs, and no adjustment differ from a published follow-up. A published analysis would name the pairs in advance or account for how many pairs were inspected, and it would use more than three scores in a group. This sketch only shows which comparison comes next.


## Three group pairs, uncorrected

The cell counts wins and calls `mannwhitneyu` once for each pair. The three p-values are printed side by side. They are not adjusted.


```python
# The same three chosen groups. Three pairs, and no adjustment for looking at all three.
from scipy.stats import mannwhitneyu

no_notes = [2, 3, 5]
short_notes = [4, 6, 7]
worked_example = [8, 9, 10]
pairs = [
    ("no notes vs short notes", no_notes, short_notes),
    ("no notes vs worked example", no_notes, worked_example),
    ("short notes vs worked example", short_notes, worked_example),
]
for name, left, right in pairs:
    wins = 0
    for left_score in left:
        for right_score in right:
            if left_score > right_score:
                wins = wins + 1
    result = mannwhitneyu(left, right, alternative="two-sided")
    # The win count, then the statistic, then the uncorrected p-value.
    print(name, wins, result.statistic, result.pvalue)
```

## One within-person pair

List against practice, for the same four people. The other condition pairs are left unrun, so the sketch does not look like a complete set.


```python
# One of the three condition pairs. Positive means practice was rated higher than list.
from scipy.stats import wilcoxon

list_scores = [2, 3, 1, 2]
practice_scores = [5, 4, 5, 6]
changes = []
for listed, practised in zip(list_scores, practice_scores):
    changes.append(practised - listed)
print(changes)
# All four changes are positive. Two of them are tied at 4. The negative rank sum is 0.
result = wilcoxon(changes)
print(result.statistic)
print(result.pvalue)
```
