# A sign test

If only the direction of each change is kept, what is left to count?

This notebook is in Non-parametric tests, in the module Ranks and related scores. The same seven people rated a short item before and after a practice worksheet. The ratings were chosen. A plus means the rating rose. A minus means it fell.

| person | before | after | change |
| --- | --- | --- | --- |
| A | 2 | 3 | +1 |
| B | 4 | 4 | 0 |
| C | 3 | 1 | -2 |
| D | 2 | 5 | +3 |
| E | 1 | 3 | +2 |
| F | 5 | 3 | -2 |
| G | 2 | 4 | +2 |

Person B did not change, so that row is neither a plus nor a minus. Among the six changes that are not zero, four are pluses. A's rise of 1 and D's rise of 3 both count as one plus. The size is ignored on purpose. `binomtest` asks how unusual four pluses out of six would be if each non-zero change were equally likely to go either way. Four out of six is not a rare count under that fair split. The p-value on this table is 0.6875.

Seven chosen pairs, one zero, and no record of who the people are, differ from a published sign test. The test assumes that under no change each non-zero sign is a fair coin, and these seven rows are not a sample on which to check that. A published analysis would name the pairing, state that zeros are set aside, and use a sample sized for the question.


## Pluses among the non-zero changes

The cell counts pluses, minuses, and zeros, then calls `binomtest` on the pluses and the non-zero total. The chance on each non-zero change, under the fair split, is 0.5.


```python
# The same seven people, before and after a worksheet. The ratings were chosen.
from scipy.stats import binomtest

before = [2, 4, 3, 2, 1, 5, 2]
after = [3, 4, 1, 5, 3, 3, 4]
# A positive change is a plus. A negative change is a minus. Zero is neither.
changes = []
for earlier, later in zip(before, after):
    changes.append(later - earlier)
print(changes)
pluses = 0
minuses = 0
zeros = 0
for change in changes:
    if change > 0:
        pluses = pluses + 1
    elif change < 0:
        minuses = minuses + 1
    else:
        zeros = zeros + 1
print(pluses)
print(minuses)
print(zeros)
# Four pluses out of the six changes that are not zero, against a fair split.
non_zero = pluses + minuses
result = binomtest(pluses, non_zero, 0.5, alternative="two-sided")
print(result.k)
print(result.n)
print(result.pvalue)
```
