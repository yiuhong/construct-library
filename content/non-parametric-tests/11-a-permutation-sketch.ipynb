# A permutation sketch

How unusual is a gap if the only thing that changes is which scores wear which label?

This notebook is in Non-parametric tests, in the module Association and limits. Six quiz scores were chosen and split into two groups of three. The summary is the difference of the two means, second group minus first. Introductory statistics used a standard error around a mean. This sketch does not. It asks how often a reassignment of these six scores produces a difference at least as far from zero.

| group | scores | mean |
| --- | --- | --- |
| first | 1, 2, 3 | 2 |
| second | 4, 6, 8 | 6 |

The observed difference is 4. There are 20 ways to choose three of the six scores for the first group. The six scores are all different, so no split is ambiguous. Only two of the 20 ways put the difference at 4 or at -4: the observed split, and the split that swaps the groups. The share is $2 / 20 = 0.1$.

`permutation_test` enumerates the distinct reassignments when `n_resamples` is set to infinity. That is feasible only because 20 is small. The statistic function takes one split or a batch of them, which is what the enumeration needs. The printed p-value is 0.1, the same share. The permutation type is `independent`, because the two groups are separate students, not pairs.

Six chosen scores, with the three largest typed into one group, differ from a published permutation test. The sketch assumes the labels on these six scores can be exchanged. It does not assume a named shape. A published analysis would justify the summary, the grouping, and that exchange. Shuffling these labels does not repair a confounded design, and it does not create a population beyond the six numbers.


## The observed gap, then every split of six scores

The cell prints the observed difference and the number of ways to choose the first group. `permutation_test` then enumerates the reassignments.


```python
# Six distinct quiz scores, already split into two groups of three.
import math

import numpy as np
from scipy.stats import permutation_test

first = np.array([1.0, 2.0, 3.0])
second = np.array([4.0, 6.0, 8.0])

def mean_gap(a, b, axis=-1):
    # Second-group mean minus first-group mean. axis lets one call serve many splits.
    return np.mean(b, axis=axis) - np.mean(a, axis=axis)

observed = mean_gap(first, second)
print(observed)
# Ways to choose 3 scores out of 6 for the first group.
ways = math.comb(6, 3)
print(ways)
# np.inf asks for every distinct reassignment. Twenty splits is small enough to enumerate.
result = permutation_test(
    (first, second),
    mean_gap,
    permutation_type="independent",
    n_resamples=np.inf,
    alternative="two-sided",
)
print(result.statistic)
print(result.pvalue)
```
