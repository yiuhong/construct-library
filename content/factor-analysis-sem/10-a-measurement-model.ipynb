# A measurement model

How do several items stand in for one latent variable?

A measurement model gives the latent variable no arrow from the items. The arrows point the other way. One latent, called verbal, has three items. The factor variance is fixed at 1. The unique variances sit on the items and are not shared.

| item | loading | unique variance |
| --- | --- | --- |
| vocabulary | 2 | 1 |
| reading | 1 | 1 |
| spelling | 2 | 1 |

The implied covariance of two items is the product of their loadings. Vocabulary and reading imply $2 \times 1 = 2$. The implied variance of an item is the loading squared plus the unique variance. Vocabulary implies $2^2 + 1 = 5$.

There are no people, no standard errors, and no maximum-likelihood fit. The loadings and the unique variances were chosen. Three items and one latent are enough to read the products, and not more than that.


## From the loadings to a covariance and a variance

The cell multiplies the vocabulary and reading loadings, then adds vocabulary's unique variance to the square of its loading.


```python
# A measurement model with one latent. Factor variance is fixed at 1. The numbers are chosen.
loading_vocab = 2
loading_reading = 1
loading_spelling = 2
unique_vocab = 1
# Covariance of two items: the product of their loadings.
cov_items = loading_vocab * loading_reading
# Variance of vocabulary: the shared part plus the unique variance.
var_vocab = loading_vocab ** 2 + unique_vocab
# Spelling uses the same loading as vocabulary, so it implies the same covariance with reading.
cov_reading_spelling = loading_reading * loading_spelling
print(cov_items)
print(var_vocab)
print(cov_reading_spelling)
```
