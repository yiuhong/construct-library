# How many factors, and a hypothesized pattern

How is a guessed pattern different from leaving every loading free?

Three covariances are written down, from loadings 0.8, 0.6, and 0.4 on one factor: vocabulary with reading is 0.48, vocabulary with arithmetic is 0.32, and reading with arithmetic is 0.24. A one-factor search that may use all three loadings can recover 0.8, 0.6, and 0.4, and the gaps fall to about zero.

A hypothesised pattern fixes arithmetic's loading at 0 before the search. Vocabulary and reading can still match 0.48. The two covariances with arithmetic cannot: those gaps stay 0.32 and 0.24. The sum of squared gaps is $0.32^2 + 0.24^2 = 0.16$. That 0.16 is the cost of the zero on this chosen matrix. It is not a chi-square, and it is not a published fit index.

A second factor could be drawn for arithmetic. This notebook does not estimate one. Three items would not pin a second factor down here, and there is no sample of people. Deciding how many factors in a full analysis uses a larger sample and a stated rule. Here the only comparison is whether that one zero is costly.


## A free pattern and a pattern with a zero

Both searches start at 1. The fixed pattern holds arithmetic at 0. The cell prints the free loadings, then the sum of squared gaps for the pattern with the zero.


```python
# Compare a free one-factor pattern with a pattern that fixes arithmetic at 0.
from scipy.optimize import minimize

# Vocabulary-reading, vocabulary-arithmetic, reading-arithmetic.
observed = [0.48, 0.32, 0.24]

def loss_free(loadings):
    # Products of the three loadings, then the squared gaps.
    implied = [
        loadings[0] * loadings[1],
        loadings[0] * loadings[2],
        loadings[1] * loadings[2],
    ]
    total = 0
    for k in range(3):
        gap = implied[k] - observed[k]
        total = total + gap ** 2
    return total

def loss_fixed(free):
    # Arithmetic's loading is the third entry, and it stays 0.
    return loss_free([free[0], free[1], 0.0])

# The loading the hypothesised pattern refuses to estimate.
fixed_loading = 0
result_free = minimize(loss_free, x0=[1.0, 1.0, 1.0])
result_fixed = minimize(loss_fixed, x0=[1.0, 1.0])
constrained_gap = float(result_fixed.fun)
print([round(float(value), 3) for value in result_free.x])
print(round(float(result_free.fun), 6))
print(fixed_loading)
print(round(constrained_gap, 3))
```
