# Loadings in a common factor model

What does a loading say about the covariance of two items?

Set the factor's variance to 1, and keep the unique parts uncorrelated with the factor and with each other. Then the covariance of two items is the product of their loadings. The previous notebook wrote the factor scores out in full and did not need this setting.

Vocabulary, reading, and spelling have loadings 0.8, 0.6, and 0.5. The three covariances are $0.8 \times 0.6 = 0.48$, $0.8 \times 0.5 = 0.40$, and $0.6 \times 0.5 = 0.30$.

The second cell searches for three loadings whose products match those covariances. It starts at 1, 1, and 1. From that start it returns 0.8, 0.6, and 0.5. A start with the opposite sign can return the same numbers with every sign flipped. Both point along the same factor.

There are no people in this search. The covariances were written down. `scipy.optimize` walks a loss written in the cell. It is not a factor-analysis package. The first time a cell imports `scipy`, that library downloads.


## Products of loadings

The cell multiplies the three pairs. Factor variance is fixed at 1.


```python
# Covariances implied by three loadings, with the factor variance fixed at 1.
loading_vocab = 0.8
loading_reading = 0.6
loading_spelling = 0.5
# Each covariance is a product of two loadings.
cov_vocab_reading = loading_vocab * loading_reading
cov_vocab_spelling = loading_vocab * loading_spelling
cov_reading_spelling = loading_reading * loading_spelling
# Two decimal places, so a product such as 0.40 is not printed as 0.4.
print("%.2f" % cov_vocab_reading)
print("%.2f" % cov_vocab_spelling)
print("%.2f" % cov_reading_spelling)
```

## A search for the loadings

The loss is the sum of squared gaps between those products and the loadings being tried. The print is the loadings the search settles on, then the loss there.


```python
# Search for loadings whose products match the three covariances. A start of 1, 1, 1 keeps the positive sign.
from scipy.optimize import minimize

# The three covariances, in the order vocabulary-reading, vocabulary-spelling, reading-spelling.
targets = [0.48, 0.40, 0.30]
pair_i = [0, 0, 1]
pair_j = [1, 2, 2]

def loss(loadings):
    # Add the squared gap for each pair.
    total = 0
    for k in range(3):
        implied = loadings[pair_i[k]] * loadings[pair_j[k]]
        gap = implied - targets[k]
        total = total + gap ** 2
    return total

# Start the search at 1 for every loading.
result = minimize(loss, x0=[1.0, 1.0, 1.0])
print([round(float(value), 3) for value in result.x])
print(round(float(result.fun), 6))
```
