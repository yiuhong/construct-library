# What PCA does not claim

What does a large first component not prove?

A component is a direction of spread. It is not, by itself, a cause, a named trait, or a factor with a unique part set aside.

Five students have two tests that are copies of each other, and a third test that is not a copy. The first component puts the same weight on the two copies. That is what the decomposition can see. It does not show that a trait produced the copies. The copies were typed in twice.

PCA also has no unique-variance term. Whatever a direction does not collect is only a leftover in the scores. The second cell splits one made-up item by hand so that leftover can be counted. Four centred scores, written down for the arithmetic, are not a fitted model.


## Two copies and a third test

The cell centres the three tests and prints the first component's loadings. The sign is chosen so the first loading is positive.


```python
# Two identical tests and one test that is not a copy. The first component's loadings show which tests share a direction.
import numpy as np

# Five students. The first two columns are the same list.
copy = [1, 2, 3, 4, 5]
other = [5, 1, 4, 2, 3]
table = np.column_stack([copy, copy, other])
# Centre each column.
centred = table - table.mean(axis=0)
u, s, vt = np.linalg.svd(centred, full_matrices=False)
# The first component. Flip the sign so the first loading is positive.
loadings = vt[0].copy()
if loadings[0] < 0:
    loadings = -loadings
print([round(float(value), 3) for value in loadings])
```

## A leftover is not a unique term

Take the centred scores `[2, 0, 0, -2]`. Suppose one component accounts for the part `[1, -1, 1, -1]`. The leftover is the difference. Those two parts are uncorrelated, so their sums of squares add to the item's sum of squares. Nothing in the subtraction wrote a unique-variance term.


```python
# Split one centred item into a part along a component and a leftover. The leftover is not a unique-variance term.
centred_item = [2, 0, 0, -2]
along_component = [1, -1, 1, -1]
# What the component part does not collect.
leftover = []
for i in range(4):
    leftover.append(centred_item[i] - along_component[i])

def sum_of_squares(values):
    # Square each value and add.
    total = 0
    for value in values:
        total = total + value ** 2
    return total

item_ss = sum_of_squares(centred_item)
component_ss = sum_of_squares(along_component)
leftover_ss = sum_of_squares(leftover)
print(leftover)
print(item_ss)
print(component_ss)
print(leftover_ss)
# The two parts are uncorrelated, so these two sums add to the item's sum of squares.
print(component_ss + leftover_ss)
```
