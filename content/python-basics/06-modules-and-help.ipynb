# Modules and help

How can I load a module and read a short description of a name?

A module is a file of names you can load. `import` loads it. The math module comes with this Python. After import, use its names as `math.sqrt` and `math.pi`.


```python
# Load the math tools, then use the square root and pi. The names come from the module, so each one is written math.something.
# Load the math module.
import math

# Square root of 16.
print(math.sqrt(16))
# pi, rounded to two decimals.
print(round(math.pi, 2))
```

## A short description

A function's `__doc__` is a short text about what it does. `help(math.sqrt)` prints that kind of text too. Run the next cell and read the description of `sqrt`.


```python
# The short description stored with the square-root function. That text travels with the function, and it is what help shows.
# Load math, so the square-root function is available.
import math

# The description stored on the function itself.
print(math.sqrt.__doc__)
```

## Import one name

`from math import sqrt` loads that one name. You can call `sqrt` without the `math.` prefix.


```python
# Load one name on its own, so the call does not need the math. prefix. Only the name you imported is brought in.
# Bring in sqrt by itself.
from math import sqrt

# Call it without a module prefix.
print(sqrt(9))
```
