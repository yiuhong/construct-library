# Conditionals and loops

How can I choose one branch, and repeat a step that stops?

A conditional chooses a branch. A loop repeats a step. Both use a condition that can be true or false.


## A choice

`if` runs its block only when the comparison is true. `else` runs when it is not. Comparisons use `>`, `<`, `>=`, `<=`, and `==`.

`==` asks whether two values are equal. `=` stores a value. It is not a comparison.


```python
# One branch when the score is high enough, and the other when it is not. Only one of the two prints runs.
# The score being tested.
score = 7
# This branch runs when the score is 5 or more.
if score >= 5:
    # The word for a high enough score.
    print("pass")
# This branch runs in every other case.
else:
    # The word when the score was too low.
    print("again")
```

## A for loop

`for` takes each value in a list, once. This loop adds the scores that are greater than 3. `total` starts at 0 so there is a number to add to. The printed total is 13.


```python
# Keep a running total of the scores that are greater than 3. Scores that are not greater are skipped, and they do not change the total.
# Start from nothing, before any score is added.
total = 0
# Look at 2, then 5, then 8.
for score in [2, 5, 8]:
    # Add the score only when it is greater than 3.
    if score > 3:
        # Add this score onto the running total.
        total = total + score
# The total of the scores that qualified.
print(total)
```

## A while loop that stops

`while` repeats as long as its condition is true. This one stops when `n` passes 3, because the last line adds 1 each time.

On this site the kernel runs on the browser's main thread. A `while` that never becomes false can freeze the tab. The condition here has a clear end.


```python
# Count from 1 up to 3, then stop. The loop keeps going while n is still 3 or less, and each pass adds 1.
# Start the count at 1.
n = 1
# Keep going while n has not passed 3.
while n <= 3:
    # Show the current count.
    print(n)
    # Move the count up by 1, or the loop would never end.
    n = n + 1
```
