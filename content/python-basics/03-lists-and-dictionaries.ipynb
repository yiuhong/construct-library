# Lists and dictionaries

How can I keep values in a list, or look one up by a key?

A list keeps values in order. The first index is 0, not 1. A negative index counts from the end. Run the next cell.


```python
# The first item, the last item, and how many items there are. Counting starts at 0, and a negative count steps back from the end.
# Three scores in one list, in this order.
scores = [2, 4, 7]
# The first item. The first index is 0.
print(scores[0])
# The last item. Counting back from the end starts at -1.
print(scores[-1])
# How many items are in the list.
print(len(scores))
```

## Adding a value

`append` adds one value at the end. The list stored in the kernel changes. `len` is the new length.


```python
# Add a score on the end. The list gets longer, and the earlier items stay in their places.
# Start with three scores.
scores = [2, 4, 7]
# Add 5 after the last item.
scores.append(5)
# The longer list, then how many items it now has.
print(scores)
print(len(scores))
```

## A dictionary

A dictionary pairs a key with a value. Look up the value by putting the key in square brackets.


```python
# Look a value up by its name, rather than by its position. The name is the key, and the stored item is the value.
# One reading, with a time and a true-or-false result.
reading = {"time": 12, "correct": True}
# The value stored under the key time.
print(reading["time"])
# The value stored under the key correct.
print(reading["correct"])
```
