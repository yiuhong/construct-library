# Python as a calculator

How can I use Python for arithmetic?

Python will do arithmetic for you. Click the next cell and press Shift+Enter. The result is printed under the cell.


```python
# Add, subtract, multiply, and divide. Each line prints one result under the cell.
# Add.
print(1 + 2)
# Subtract.
print(10 - 3)
# Multiply.
print(4 * 5)
# Divide. The result keeps the remainder as a decimal.
print(9 / 2)
```

## Order

Multiplication and division happen before addition and subtraction. Parentheses happen first. The two lines in the next cell are not the same expression.


```python
# Multiplication happens before addition, unless parentheses say otherwise. The two lines use the same numbers and differ only in the parentheses.
# Multiply 3 by 4 first, then add 2.
print(2 + 3 * 4)
# Add inside the parentheses first, then multiply by 4.
print((2 + 3) * 4)
```

## Powers and remainders

`**` raises a number to a power. `//` divides and keeps the whole number of times it fits. `%` is the remainder.


```python
# A power, how many times one whole number fits into another, and what is left over. The last two lines go together: the quotient and the remainder.
# 2 to the power 5.
print(2 ** 5)
# How many times 2 fits into 7, ignoring the remainder.
print(7 // 2)
# What is left over after that division.
print(7 % 2)
```
