# Types and variables

How can I store a value under a name, and tell what type it has?

A variable is a name for a value. Run the next cell. It stores a number under the name `score`, then prints that number.


```python
# Store a score under a name, then show it. The name is yours to choose; the value is what gets printed.
# Store 7 under the name score.
score = 7
# Show the stored value.
print(score)
```

## Types

A value has a type. `7` is an int, a whole number. `3.5` is a float, a number with a decimal point. `"trial"` is a str, text. `True` and `False` are bool values.

`type` reports the type. The printed name is the part to read.


```python
# Four kinds of value: a whole number, a decimal, text, and a true-or-false value. The printed word is the name of that kind.
# A whole number.
count = 3
# A decimal.
height = 3.5
# Text. The quotes mark it as text, and they are not part of the value.
label = "trial"
# A true-or-false value. True and False are written with a capital letter.
correct = True
# The kind of each value, one print per value.
print(type(count).__name__)
print(type(height).__name__)
print(type(label).__name__)
print(type(correct).__name__)
```

## A name can change

Run the next cell. `score` was 7. This cell adds 1 and stores the new value under the same name. The old value is gone.

Run the first code cell first if `score` is not defined yet.


```python
# The kernel still holds the score from the first cell. This cell adds 1 to that stored value and shows the result. Run the first cell before this one.
# Replace the stored score with one more than it was.
score = score + 1
# The new score.
print(score)
```
