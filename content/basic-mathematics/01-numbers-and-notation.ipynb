# Numbers and notation

A number can be a whole number, a negative, or a decimal. Click the next cell and press Shift+Enter.


```python
print(-2)
print(0)
print(3)
print(0.5)
```

## Fractions and powers

A fraction is division. One half is `1 / 2`. A power is repeated multiplication. `10 ** 2` is 10 squared, written $10^2$. `10 ** 3` is 10 cubed.


```python
print(1 / 2)
print(10 ** 2)
print(10 ** 3)
```

## Absolute value

The absolute value of a number is its distance from zero. `abs` computes it. `-5` and `5` are the same distance from zero.

On a number line, order runs from left to right. `-2` is left of `0`, and `0` is left of `3`.


```python
print(abs(-5))
print(abs(5))
print(-2 < 0 < 3)
```
