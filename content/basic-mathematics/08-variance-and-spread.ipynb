# Variance and spread

How can I measure how far values sit from their mean?

The mean says where the values sit. The variance says how far they sit from that mean. Square each distance, then average the squares. Squaring keeps every distance positive, and a far value counts more than a near one.

$$
\mathrm{Var}(x) = \frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2
$$

This divides by $n$, the length of the list. It describes this list. The standard deviation is the square root of the variance, in the same units as the values.

For `[1, 3]` the mean is 2 and the variance is 1. For `[2, 2]` the variance is 0.


```python
# Variance is the average of the squared distances from the mean. Squaring stops a distance above the mean and a distance below it from cancelling.
# A rule for the variance of one list.
def variance(values):
    # The mean the distances are measured from.
    mean = sum(values) / len(values)
    # A running total of the squared distances.
    total = 0
    # One value at a time.
    for value in values:
        # Add this value's squared distance from the mean.
        total = total + (value - mean) ** 2
    # Divide by how many values there were. That average is the variance.
    return total / len(values)

# A list with some spread, then a list whose two values match.
print(variance([1, 3]))
print(variance([2, 2]))
# The square root puts that spread back into the same units as the data.
print(variance([1, 3]) ** 0.5)
```

## Same mean, different spread

Both lists below have mean 5. One does not spread. The other does. The figure shows the values. The first plot in a session can take a moment while the plotting library downloads.


```python
# Two lists with the same centre. One never moves, and the other is spread out. The prints give the variances, and the figure shows the lists.
import matplotlib.pyplot as plt

# The same variance rule as the cell above.
def variance(values):
    # The mean the distances are measured from.
    mean = sum(values) / len(values)
    # A running total of the squared distances.
    total = 0
    # One value at a time.
    for value in values:
        # Add this value's squared distance from the mean.
        total = total + (value - mean) ** 2
    # The average of those squared distances.
    return total / len(values)

# No spread, then a wide spread.
print(variance([5, 5, 5]))
print(variance([1, 5, 9]))

# The flat series stays at 5. The other runs from 1 through 5 to 9.
# A figure for the two series.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The series that stays at 5.
ax.plot([1, 2, 3], [5, 5, 5], marker="o", label="No spread")
# The series that moves away from the centre.
ax.plot([1, 2, 3], [1, 5, 9], marker="o", label="Spread out")
# The horizontal axis is the place in the list.
ax.set_xlabel("Position in the list")
ax.set_ylabel("Value")
# A key, so the two series can be told apart.
ax.legend()
# Show the figure under this cell.
plt.show()
```
