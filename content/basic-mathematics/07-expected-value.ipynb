# Expected value

The expected value is a weighted average. Multiply each outcome by its probability, then add those products.

$$
E[X] = \sum x \, P(x)
$$

It is the average you would see if the same chance were repeated many times. One run can land on any outcome that has some probability.


## A fair die

The faces 1 through 6 are equally likely, so each probability is $1/6$. The expected value is the ordinary mean of the six faces. You cannot roll 3.5. That number is the long-run average.


```python
faces = [1, 2, 3, 4, 5, 6]
expected = sum(faces) / len(faces)
print(expected)
```

## Unequal probabilities

A payoff of 10 with probability 0.5, and 0 otherwise, has expected value 5. A payoff of 4 with probability 0.25, and 0 otherwise, has expected value 1. The rare payoff is not dropped. It is weighted by how rare it is.


```python
print(0 * 0.5 + 10 * 0.5)
print(4 * 0.25 + 0 * 0.75)
```
