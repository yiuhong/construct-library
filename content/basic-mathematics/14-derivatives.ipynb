# Derivatives

How can I estimate a rate of change at one point?

A derivative is a rate of change. Over a step of size $h$, the average rate of $f$ is

$$
\frac{f(x + h) - f(x)}{h}
$$

The derivative $f'(x)$ is the limit of that rate as $h$ approaches 0. It is the slope of the curve at $x$.

For $f(x) = x^2$, the derivative is $f'(x) = 2x$. At $x = 3$ the slope is 6. The next cell computes the average rate with a small $h$, then the exact slope. The two numbers are close.


```python
# A short rise over a short run, set beside the exact slope of x squared at x = 3. A shorter step would sit even closer to that slope.
# The curve: x squared.
def f(x):
    return x ** 2

# Where the slope is measured.
x = 3
# A short step along x.
h = 0.0001
# Rise over that step, divided by the length of the step.
average = (f(x + h) - f(x)) / h
# That average slope, then the exact slope of this curve at x.
print(round(average, 4))
print(2 * x)
```

## A straight line

The line $y = 5x + 1$ has slope 5 at every $x$. Its derivative is the constant 5. Adding a constant moves the line up or down and does not change the slope. The derivative of a constant is 0.

The dashed line in the figure is the tangent to $x^2$ at $x = 3$. It touches the curve there and has slope 6. The first plot in a session can take a moment while the plotting library downloads.


```python
# The curve y = x squared, and the straight tangent that touches it at x = 3. The tangent uses the slope of the curve at that point.
import matplotlib.pyplot as plt

# x values from 2 to 4.
xs = []
# Height of the curve at each x.
curve = []
# Height of the tangent at each x.
tangent = []
# i divided by 10 runs from 2.0 to 4.0, in steps of a tenth.
for i in range(20, 41):
    x = i / 10
    xs.append(x)
    curve.append(x ** 2)
    tangent.append(9 + 6 * (x - 3))

# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The curve.
ax.plot(xs, curve)
# The tangent, dashed so it stays distinct from the curve.
ax.plot(xs, tangent, linestyle="--")
# Name the horizontal axis.
ax.set_xlabel("x")
# Name the vertical axis.
ax.set_ylabel("y")
# Show the figure under this cell.
plt.show()
```
