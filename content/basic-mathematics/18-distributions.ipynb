# Distributions

How can I match a named distribution to a kind of outcome?

A distribution says how probability is spread over outcomes. A model picks a distribution that matches the kind of outcome it describes. Three named ones cover a yes-or-no trial, a count of such trials, and a measurement spread around a mean.


## Bernoulli and binomial

A Bernoulli outcome is one yes or no. Success has probability $p$. Failure has probability $1 - p$. A model uses it for one choice, or one trial that is correct or not.

A binomial count is how many successes appear in $n$ independent Bernoulli trials with the same $p$. A model uses it for "how many correct out of $n$".

$$
P(K = k) = \binom{n}{k} p^k (1 - p)^{n - k}
$$

The binomial coefficient counts the ways to place $k$ successes in $n$ trials. `math.comb` computes it. Two successes in three trials, with $p = 0.5$, have probability 0.375.


```python
# Chance of two successes in three trials when each trial has probability 0.5.
import math

n = 3
k = 2
p = 0.5
probability = math.comb(n, k) * p ** k * (1 - p) ** (n - k)
print(probability)
```

## Normal

A normal distribution is a bell curve around a mean $\mu$, with spread $\sigma$. A model uses it for a measurement that can fall either side of the mean, when many small contributions add up. The curve below is the standard normal, $\mu = 0$ and $\sigma = 1$.

$$
\frac{1}{\sqrt{2\pi}} \exp\left(-\frac{x^2}{2}\right)
$$

The height is a density. The area under the whole curve is 1. The height at one point is not itself a probability.

The first plot in a session can take a moment while the plotting library downloads.


```python
# The standard normal density. It is highest at zero and falls off on both sides.
import math
import matplotlib.pyplot as plt

xs = []
ys = []
for i in range(-100, 101):
    x = i / 25
    xs.append(x)
    ys.append(math.exp(-0.5 * x * x) / math.sqrt(2 * math.pi))

# Draw the density.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(xs, ys)
ax.set_xlabel("x")
ax.set_ylabel("Density")
plt.show()
```

## Which one a model picks

One yes or no is Bernoulli. A count out of a fixed number of trials is binomial. A measurement spread around a mean is often normal. A normal curve is a poor description of a single coin flip, and a coin-flip probability is a poor description of a measurement that has a spread.
