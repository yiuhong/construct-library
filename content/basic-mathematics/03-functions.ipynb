# Functions

How can I read a function as a rule from an input to an output?

A function is a rule that turns an input into an output. The same input always gives the same output.

Write $f(x) = 2x + 1$. The letter is the input slot. $f(0) = 1$, $f(1) = 3$, and $f(2) = 5$. The `def` below is how this notebook computes that rule.


```python
# A rule that turns an input into 2 times that input, plus 1. Each line below calls that same rule with a different input.
# Name the rule f. It takes one input, x.
def f(x):
    return 2 * x + 1

# The input, then the rule's result, for 0, 1, and 2.
print(0, f(0))
print(1, f(1))
print(2, f(2))
```

## The letter is a slot

The name of the input does not change the rule. $g(t) = t^2$ is the square of whatever you pass in. $g(3) = 9$, and $g(3)$ again is still 9.


```python
# The same input gives the same output each time. Nothing inside the rule changes between the two calls.
# Square the input.
def g(t):
    return t ** 2

# Call the rule twice with the same input.
print(g(3))
print(g(3))
```

## In a model

A model often writes a response as a function of a condition, a time, or a parameter. Reading it means: name the input, apply the rule, get the output.
