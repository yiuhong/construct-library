# Integrals

How can I add a quantity over an interval, as an area?

An integral adds a quantity over an interval. When that quantity is the height of a curve, the integral is the area under the curve.

A constant height is a rectangle. Height 2 from $x = 0$ to $x = 3$ has area $2 \times 3 = 6$.

$$
\int_0^3 2 \, dx = 6
$$

Read that, in a model, as a rate of 2 per unit time held for 3 units of time. The total is 6.


```python
print(2 * 3)
```

## A triangle, then a sum of slices

The line $y = x$ from 0 to 2 makes a triangle with base 2 and height 2. The area is base times height, divided by 2.

$$
\int_0^2 x \, dx = 2
$$

Rectangles can approximate that area. The next cell uses four slices of width 0.5. The height of each slice is the value at its right edge. The loop stops once $x$ passes 2. The sum is a bit above 2. Thinner slices sit closer to the exact area.

This is the idea you need when a model accumulates a rate. It is a small set of pictures, not a full list of integral rules.


```python
width = 0.5
total = 0.0
x = width
while x <= 2:
    total = total + width * x
    x = x + width
print(total)
```
