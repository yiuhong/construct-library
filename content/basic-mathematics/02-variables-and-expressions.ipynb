# Variables and expressions

How can I let a letter stand for a number inside an expression?

A letter can stand for a number. Store that number under the letter's name. An expression combines numbers and names into a new value.

Run the next cell. `x` is 5. The expression `2 * x - 3` is computed from that.


```python
# Give x a value, then use that stored value in an expression. The name keeps the value until a later line changes it.
# Store 5 under the name x.
x = 5
# Multiply the stored x by 2, then subtract 3.
print(2 * x - 3)
```

## Substitution

To evaluate an expression, replace the letter with its number, then do the arithmetic. Parentheses still happen first. With `x` equal to 5, `3 * (x - 1)` is `3 * 4`.


```python
# Parentheses decide what happens first. Without them, multiplication happens before subtraction. The second line squares x.
# The value both lines use.
x = 5
# Subtract inside the parentheses first, then multiply by 3.
print(3 * (x - 1))
# x times itself.
print(x ** 2)
```

## A name keeps the value it was given

The next cell stores `y` from `x`, then changes `x`. `y` does not follow. It keeps the number it was given.


```python
# y keeps the value it was given at that moment. Changing x afterwards leaves y alone, because y does not look x up again.
# Start with x equal to 5.
x = 5
# y stores the result of this sum, not a live link back to x.
y = x + 1
# This changes x only.
x = 0
# The new value of x.
print(x)
# y still holds the value from before x changed.
print(y)
```
