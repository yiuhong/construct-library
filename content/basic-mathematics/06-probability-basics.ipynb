# Probability basics

A probability is a number from 0 to 1. 0 means the outcome does not happen. 1 means it does. A value in between is a chance.

A fair coin has two outcomes, and neither is more likely than the other. $P(\text{heads}) = 1/2$. The probability of "not heads" is $1 - 1/2$.


```python
print(1 / 2)
print(1 - 1 / 2)
```

## Add outcomes that do not overlap

A fair die has six faces, 1 through 6. Each face has probability $1/6$. The even faces are 2, 4, and 6. Those faces do not overlap, so the probabilities add:

$$
P(\text{even}) = 1/6 + 1/6 + 1/6 = 1/2
$$


```python
faces = [1, 2, 3, 4, 5, 6]
even = []
for face in faces:
    if face % 2 == 0:
        even.append(face)
print(even)
print(len(even) / len(faces))
```

## The complement

The probability that an outcome does not happen is 1 minus the probability that it does. If an outcome has probability 0.2, the rest has probability 0.8.
