# Linear equations

A linear equation has the unknown to the first power only. Solve $3x + 1 = 10$ by undoing each step. Subtract 1 from both sides: $3x = 9$. Divide both sides by 3: $x = 3$.

Check by putting 3 back in. $3 \times 3 + 1$ is 10, the same as the right-hand side.


```python
x = 3
print(3 * x + 1)
```

## The graph is a line

$y = 2x + 1$ is a line. The 2 is the slope: $y$ increases by 2 when $x$ increases by 1. The 1 is the intercept: $y$ is 1 when $x$ is 0.

The figure appears under the cell. The first plot in a session can take a moment while the plotting library downloads.


```python
import matplotlib.pyplot as plt

xs = [-1, 0, 1, 2]
ys = []
for x in xs:
    ys.append(2 * x + 1)

fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(xs, ys, marker="o")
ax.axhline(0, color="#cccccc", linewidth=1)
ax.axvline(0, color="#cccccc", linewidth=1)
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.show()
```
