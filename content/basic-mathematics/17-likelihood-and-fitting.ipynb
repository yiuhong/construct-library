# Likelihood and fitting

How can I choose a parameter by a rule stated for the data?

A model has a parameter. A fit picks the value of that parameter that matches the data under a rule you state. "Best fit" means the winner under that rule. Likelihood and least squares are two such rules.


## Likelihood

The likelihood is how probable the data are if the parameter takes a chosen value. Four coin flips gave three heads and one tail. If the probability of heads is $p$, and the flips are independent,

$$
L(p) = p^3 (1 - p)
$$

A constant that does not depend on $p$ would not change which $p$ wins, so it is left out. The next cell tries four values of $p$. For this count of heads, the likelihood is highest at the observed proportion, 3/4.


```python
# Three heads and one tail. Try a few values of the unknown proportion of heads. A proportion that makes this sequence more probable scores higher.
# How probable three heads and one tail are, if the proportion of heads is p.
def likelihood(p):
    return p ** 3 * (1 - p)

# Four guesses for that proportion.
for p in [0.25, 0.5, 0.75, 0.9]:
    # The guess, and how probable the sequence is, to four decimals.
    print(p, round(likelihood(p), 4))
```

## Least squares

Three points: (1, 2), (2, 4), and (3, 7). Fit a line through the origin, $y = ax$. The error at a point is $y - ax$. Least squares picks the slope $a$ that makes the sum of squared errors smallest.

The next cell tries three slopes, then computes the slope that wins for a line through the origin:

$$
a = \frac{\sum x y}{\sum x^2}
$$

That formula is the best fit under the least-squares rule. A different rule could pick a different line.


```python
# How far a line through the origin misses these points, for a few slopes. A smaller sum of squares is a closer line.
# The inputs.
xs = [1, 2, 3]
# The observed outcomes.
ys = [2, 4, 7]

# Sum of squared misses for a line of slope a through the origin.
def sse(a):
    # A running total of the squared misses.
    total = 0
    # One point at a time.
    for i in range(3):
        total = total + (ys[i] - a * xs[i]) ** 2
    return total

# Three slopes to compare.
for a in [2, 2.2, 2.5]:
    # The slope, and its sum of squares.
    print(a, round(sse(a), 3))

# The slope that makes the sum of squares as small as it can be, for a line that is required to pass through the origin.
# Totals used to find the best slope: each x times y, and each x squared.
total_xy = 0
total_xx = 0
# Add each point into those totals.
for i in range(3):
    total_xy = total_xy + xs[i] * ys[i]
    total_xx = total_xx + xs[i] ** 2
# The slope with the smallest sum of squares.
print(round(total_xy / total_xx, 3))
```
