# Limits

How can I see what value an expression approaches?

A limit is the value an expression approaches as the input approaches some number.

$$
\lim_{x \to 1} \frac{x^2 - 1}{x - 1} = 2
$$

At $x = 1$ the denominator is 0, so the expression is not defined there. For every other $x$, $x^2 - 1 = (x - 1)(x + 1)$, so the fraction equals $x + 1$. As $x$ approaches 1, $x + 1$ approaches 2.

The limit can exist when the expression has no value at the point. The next cell prints values from both sides of 1. It does not substitute 1.


```python
# Values from both sides of 1. The fraction is not defined at 1 itself.
def f(x):
    return (x ** 2 - 1) / (x - 1)

for x in [0.9, 0.99, 0.999, 1.001, 1.01, 1.1]:
    print(x, round(f(x), 4))
```
