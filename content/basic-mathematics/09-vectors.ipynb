# Vectors

A vector, here, is a list of numbers kept together. Two scores for one person can be a vector of length 2. NumPy stores that list as an array.

Adding two vectors adds each place. The two arrays have to be the same length. A number times a vector multiplies every place.

The first time a cell imports numpy, that library may take a moment to download.


```python
import numpy as np

a = np.array([1.0, 2.0])
b = np.array([3.0, 4.0])
print(a + b)
print(2 * a)
```

## Dot product and length

The dot product multiplies matching places and adds those products. It is one number. For these two vectors it is $1 \times 3 + 2 \times 4 = 11$.

The length is the square root of the sum of squares. It is the straight-line distance from the origin to the point the vector names. For `[1, 2]` that length is the square root of 5.


```python
import numpy as np

a = np.array([1.0, 2.0])
b = np.array([3.0, 4.0])
print(np.dot(a, b))
print(np.linalg.norm(a))
```
