# An interval around the estimate

How can I place an interval around a sample mean?

A point estimate is one number. An interval places a lower end and an upper end around it. A rough interval in this lesson reaches about two standard errors either side of the sample mean.

$$
\bar{x} \pm 2 \times \mathrm{SE}
$$

The sample mean is 10 and the standard error is 1. Two standard errors are 2, so the lower end is 8 and the upper end is 12. This is a rough interval for reading the estimate. It is not a derivation of a critical value from a named table.


## Two standard errors either side

The cell subtracts and adds the same half-width.


```python
# A rough interval: two standard errors below the mean, and two above it.
# The sample mean and its standard error.
mean = 10
standard_error = 1
# About two standard errors, as the half-width of this lesson's interval.
half_width = 2 * standard_error
# The two ends.
lower = mean - half_width
upper = mean + half_width
print(lower)
print(upper)
```
