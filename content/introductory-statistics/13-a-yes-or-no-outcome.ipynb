# A yes-or-no outcome

How can a line state a probability instead of a mean on the outcome's own scale?

A yes-or-no outcome is coded 1 or 0. Its mean is a probability, and a probability has to stay between 0 and 1. A straight line on that probability can leave the interval. A generalised linear model keeps the line, and uses a link so that the line is allowed to describe the mean.

For a yes-or-no outcome the link used here is the logit. The line is on the logit scale, and the probability is read back from it:

$$
p = \frac{1}{1 + e^{-\eta}}
$$

$\eta$ is the linear predictor, the line. When $\eta = 0$, the probability is 0.5. When $\eta = \log 3$, the odds are 3 and the probability is 0.75. The counts in a binomial outcome use the same link. This lesson uses the probability form.

The cell does not fit a line. It reads a probability from a linear predictor that is already stated.

Teaching note. Nelder, J. A., & Wedderburn, R. W. M. (1972). Generalized linear models. *Journal of the Royal Statistical Society. Series A (General), 135*(3), 370-384. https://doi.org/10.2307/2344614


## From the line to a probability

`probability` undoes the logit. The print shows the two linear predictors from the text. A short curve shows the same map for a few values of the line. The first plot in a session can take a moment while the plotting library downloads.


```python
# Turn a linear predictor into a probability. The line stays unrestricted. The probability stays between 0 and 1.
import math
import matplotlib.pyplot as plt

def probability(linear_predictor):
    # e to the power of minus the line.
    expo = math.exp(-linear_predictor)
    # The logistic map, from the line back to a probability.
    return 1 / (1 + expo)

print(probability(0))
print(probability(math.log(3)))

# A few points along the line, so the curve can be seen.
etas = [-2, -1, 0, 1, 2]
probs = []
for eta in etas:
    probs.append(probability(eta))
# The probability read from each linear predictor.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(etas, probs, marker="o")
# The horizontal axis is the line, on the logit scale.
ax.set_xlabel("Linear predictor")
# The vertical axis is the probability.
ax.set_ylabel("Probability")
# Show the figure under this cell.
plt.show()
```
