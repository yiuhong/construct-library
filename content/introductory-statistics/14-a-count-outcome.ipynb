# A count outcome

How can the same kind of model state a mean count?

A count is 0, 1, 2, and so on. Its mean has to stay positive. The generalised linear model from the previous notebook still uses a line. For a count, the link used here is the log, so the line is the log of the mean:

$$
\text{mean} = e^{\eta}
$$

When $\eta = 0$, the mean count is 1. When $\eta = \log 4$, the mean count is 4. A Poisson distribution is a simple distribution for counts in this pattern. Its mean is the number the log link produces. This lesson only reads that mean. It does not fit the line.


## Exponentiate the line

`predicted_count` applies the exp that undoes the log link.


```python
# The log link keeps the mean count positive. Exponentiating the line reads the mean back.
import math

def predicted_count(linear_predictor):
    # Undo the log. The result is the mean count.
    return math.exp(linear_predictor)

print(predicted_count(0))
print(predicted_count(math.log(4)))
```
