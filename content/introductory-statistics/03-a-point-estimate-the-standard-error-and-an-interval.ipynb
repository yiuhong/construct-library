# A point estimate, the standard error, and an interval

How does one sample become an estimate, a standard error, and an interval?

You will take the mean of two coursework scores as a point estimate of the cohort mean, build the standard error of that mean, and only then place a rough interval two standard errors either side. By the end you should be able to do those steps in that order.

A point estimate is one number. The standard error is the width that comes from the sample's own spread and its size. The interval is that width laid either side of the estimate, so it cannot come first.

The previous lesson showed that two samples can have different means. This lesson keeps one sample of two scores from the coursework cohort and measures it against the cohort mean you already know.


## The point estimate

The sample is S4 and S6, scores 4 and 10. Both scores are in the cohort. The point estimate is the sample mean, $(4 + 10) / 2 = 7$. The cohort mean is 6. The error is $7 - 6 = 1$, and the absolute error is 1. The sign says the estimate is above the cohort mean. The absolute error drops the sign.


```python
# The sample mean is the point estimate. Then measure how far it sits from the cohort mean.
# S4 and S6, two scores from the coursework cohort.
sample = [4, 10]
# The cohort mean, from all six scores. The sample does not replace that list.
cohort_mean = 6
# The point estimate.
estimate = sum(sample) / len(sample)
# Signed error: estimate minus the cohort mean.
error = estimate - cohort_mean
# Drop the sign.
absolute_error = abs(error)
print(estimate)
print(absolute_error)
```

## The standard error

The sample standard deviation describes spread in the sample. Square each distance from the sample mean, divide by $n - 1$, and take the square root. The standard error of the mean divides that standard deviation by $\sqrt{n}$.

$$
\mathrm{SE} = \frac{s}{\sqrt{n}}
$$

The distances from 7 are $-3$ and $3$. The squared distances sum to 18. Divide by $2 - 1 = 1$, so the sample variance is 18 and $s = \sqrt{18}$. Then

$$
\mathrm{SE} = \frac{\sqrt{18}}{\sqrt{2}} = \sqrt{9} = 3.
$$

Two students are enough to see this arithmetic and far too few for a published standard error. A full analysis would use more of the cohort. It would still need a stated rule for how the ends of the interval are chosen.


```python
# Sample variance, using n - 1, then the standard error of the mean.
import math

# S4 and S6 again.
sample = [4, 10]
n = len(sample)
# The sample mean.
mean = sum(sample) / n
# A running total of the squared distances.
total = 0
for value in sample:
    # Square the distance from the sample mean.
    total = total + (value - mean) ** 2
# Divide by n - 1 for this sample variance.
variance = total / (n - 1)
# The standard error is sqrt(variance / n), which is s / sqrt(n).
# Here that is sqrt(18 / 2) = sqrt(9) = 3.
standard_error = math.sqrt(variance / n)
print(total)
print(variance)
print(standard_error)
```

## An interval around the estimate

A rough interval in this lesson reaches about two standard errors either side of the sample mean.

$$
\bar{x} \pm 2 \times \mathrm{SE}
$$

The sample mean is 7 and the standard error is 3. Two standard errors are 6, so the lower end is $7 - 6 = 1$ and the upper end is $7 + 6 = 13$. The cohort mean, 6, sits inside that interval. This is a rough interval for reading the estimate. It is not a critical value from a named table. With two students the interval is very wide, which is the sample size showing up in the arithmetic.


```python
# A rough interval: two standard errors below the mean, and two above it.
# The sample mean and the standard error just obtained.
mean = 7
standard_error = 3
# About two standard errors, as the half-width of this lesson's interval.
half_width = 2 * standard_error
# The two ends.
lower = mean - half_width
upper = mean + half_width
print(lower)
print(upper)
```
