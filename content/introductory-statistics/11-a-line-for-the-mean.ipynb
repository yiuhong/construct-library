# A line for the mean

How does a straight line state the mean of an outcome at each value of $x$?

This notebook is in Introductory statistics, in the module Linear models and GLM. The line is a statement about the mean, not only a picture of points.

$$
\text{mean} = b_0 + b_1 x
$$

For $x$ of `[0, 1, 2]` and outcomes `[1, 3, 5]`, the line $\text{mean} = 1 + 2x$ matches every point. The intercept $b_0 = 1$ is the mean at $x = 0$. The slope $b_1 = 2$ is how much that mean changes when $x$ increases by 1. At $x = 2$ the mean is 5.


## Read the mean off the line

The cell evaluates the line at each $x$. The figure shows the points and the line on top of one another, because this small example was chosen to fit exactly. The first plot in a session can take a moment while the plotting library downloads.


```python
# The line states a mean at each x. These points were chosen so the line hits them.
import matplotlib.pyplot as plt

xs = [0, 1, 2]
ys = [1, 3, 5]
# The intercept is the mean at x = 0.
intercept = 1
# The slope is the change in the mean when x increases by 1.
slope = 2
# A predicted mean for each x.
predicted = []
for x in xs:
    predicted.append(intercept + slope * x)
print(intercept)
print(predicted)

# The points, and the line that states their mean.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The observed outcomes.
ax.scatter(xs, ys, label="Observed")
# The mean the line states at each x.
ax.plot(xs, predicted, label="Mean from the line")
# The horizontal axis is x.
ax.set_xlabel("x")
# The vertical axis is the outcome.
ax.set_ylabel("Outcome")
# A key for the points and the line.
ax.legend()
# Show the figure under this cell.
plt.show()
```
