# A group difference as a line, and the residuals

How large is the gap between rest and trained, once it is written as a line?

You will subtract the two group means, write that gap as the slope of a line with a 0/1 group code, and then find the residual for each score. By the end you should be able to read the group difference as a slope, and the sum of squared residuals as what the line did not account for.

A comparison of two means is a line, not a separate kind of arithmetic. The residuals are the part of each score that the group line does not claim.

The previous lesson put an interval around one sample mean, as if the cohort were one pool of scores. Rest and trained are two groups in that same cohort, and the gap between their means is the comparison.


## The two group means

Rest is S1, S2, and S3, with scores 2, 5, and 8, so the rest mean is 5. Trained is S4, S5, and S6, with scores 4, 7, and 10, so the trained mean is 7. Trained minus rest is 2. The dots are the six coursework scores. The diamonds are the two means those scores were reduced to.


```python
# Two group means from the coursework scores, and the gap between them.
import matplotlib.pyplot as plt

# Rest, then trained. These are the two halves of the score column.
rest = [2, 5, 8]
trained = [4, 7, 10]
# Each mean divides the group's sum by the group's length.
rest_mean = sum(rest) / len(rest)
trained_mean = sum(trained) / len(trained)
# The comparison this lesson reports.
difference = trained_mean - rest_mean
print(rest_mean)
print(trained_mean)
print(difference)

# One dot per score, and a diamond at each group mean.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter([0.9, 1.0, 1.1], rest, label="Rest")
ax.scatter([1.9, 2.0, 2.1], trained, label="Trained")
ax.scatter([1, 2], [rest_mean, trained_mean], marker="D", color="black", label="Group mean")
ax.set_xticks([1, 2], ["Rest", "Trained"])
ax.set_xlabel("Group")
ax.set_ylabel("Score")
ax.legend()
plt.show()
```

## The same gap as a line

Code rest as 0 and trained as 1. The mean of the score is then a line in that code:

$$
\text{mean} = b_0 + b_1 \times \text{group}
$$

$b_0$ is the mean when the group code is 0, so it is the rest mean, 5. $b_1$ is the change when the code goes from 0 to 1, so it is the group difference, 2. The prediction for trained is $5 + 2 = 7$, the trained mean.


```python
# The group gap is the slope of a line that uses a 0/1 group code.
import matplotlib.pyplot as plt

# Rest is coded 0. Trained is coded 1.
rest_mean = 5
trained_mean = 7
# The intercept is the mean at group code 0.
intercept = rest_mean
# The slope is the change from code 0 to code 1.
slope = trained_mean - rest_mean
# The line at group code 1.
prediction = intercept + slope * 1
print(intercept)
print(slope)
print(prediction)

# The two group means sit on the line.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot([0, 1], [intercept, prediction], marker="o")
# Show the group names instead of the codes 0 and 1.
ax.set_xticks([0, 1], ["Rest", "Trained"])
# The horizontal axis is the group code.
ax.set_xlabel("Group code")
# The vertical axis is the group mean.
ax.set_ylabel("Mean score")
# Show the figure under this cell.
plt.show()
```

## Residuals from the group line

The line predicts 5 for every rest score and 7 for every trained score. A residual is the score minus that prediction. The rest residuals are $-3$, $0$, and $3$. The trained residuals are $-3$, $0$, and $3$. Square them and add: $9 + 0 + 9 + 9 + 0 + 9 = 36$. The line accounts for the group means. The sum of squared residuals is what the line did not account for.


```python
# Observed score minus the group prediction, then the sum of those squared residuals.
# Rest rows are coded 0. Trained rows are coded 1. The score follows the same order.
group = [0, 0, 0, 1, 1, 1]
score = [2, 5, 8, 4, 7, 10]
# The line from the group means.
intercept = 5
slope = 2

# The residual for S1, computed on its own so it can be printed.
first_prediction = intercept + slope * group[0]
first_residual = score[0] - first_prediction

# A running total of the squared residuals.
total = 0
for i in range(len(score)):
    # The line's prediction for this row.
    prediction = intercept + slope * group[i]
    # Observed minus predicted.
    residual = score[i] - prediction
    total = total + residual ** 2
print(first_residual)
print(total)
```
