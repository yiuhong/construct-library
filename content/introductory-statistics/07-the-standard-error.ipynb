# The standard error

How does the spread of a sample become a standard error for the mean?

The sample standard deviation describes spread. For an estimate of a population spread, divide the sum of squared distances by $n - 1$, not by $n$, then take the square root. The standard error of the mean divides that standard deviation by $\sqrt{n}$.

$$
\mathrm{SE} = \frac{s}{\sqrt{n}}
$$

The sample `[0, 0, 0, 4]` has mean 1. The squared distances are 1, 1, 1, and 9, and they sum to 12. Divide by $4 - 1 = 3$ to get 4, so $s = 2$. Then $\mathrm{SE} = 2 / \sqrt{4} = 1$. A larger sample, with the same spread, would make this standard error smaller.


## From squared distances to the standard error

The cell totals the squared distances, divides by $n - 1$, takes the square root, and then divides by $\sqrt{n}$.


```python
# Sample standard deviation, using n - 1, then the standard error of the mean.
import math

# Four sample values.
sample = [0, 0, 0, 4]
n = len(sample)
# The sample mean.
mean = sum(sample) / n
# A running total of the squared distances.
total = 0
for value in sample:
    # Square the distance from the sample mean.
    total = total + (value - mean) ** 2
# Divide by n - 1 for this sample standard deviation.
variance = total / (n - 1)
standard_deviation = math.sqrt(variance)
# The standard error divides that spread by the square root of the sample size.
standard_error = standard_deviation / math.sqrt(n)
print(standard_deviation)
print(standard_error)
```
