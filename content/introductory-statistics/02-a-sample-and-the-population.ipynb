# A sample and the population

What changes when I describe a sample instead of every value?

The population here is the whole list, `[1, 2, 3, 4, 5, 6]`. Its mean is $21/6 = 3.5$. A sample is only some of those values. The even values `[2, 4, 6]` are one sample, and their mean is 4.

The sample mean describes the sample. It does not have to equal the population mean. Later lessons ask how far an estimate from a sample can move.


## Two means

The cell computes the mean of the whole list and the mean of the sample.


```python
# The mean of every value, and the mean of a sample taken from that list.
# The whole list.
population = [1, 2, 3, 4, 5, 6]
# One sample: the even values.
sample = [2, 4, 6]
# Each mean is a sum divided by its own length.
population_mean = sum(population) / len(population)
sample_mean = sum(sample) / len(sample)
print(population_mean)
print(sample_mean)
```
