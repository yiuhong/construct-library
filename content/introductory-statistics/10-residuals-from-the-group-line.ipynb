# Residuals from the group line

What is left over after the group line has made its prediction?

The line from the previous notebook predicts 4 for every rest value and 7 for every trained value. A residual is the observed value minus that prediction.

Rest is `[2, 4, 6]`. The residuals are $-2$, $0$, and $2$. Trained is `[5, 7, 9]`. The residuals are $-2$, $0$, and $2$. Square them and add: $4 + 0 + 4 + 4 + 0 + 4 = 16$. The line accounts for the group means. The sum of squared residuals is what the line did not account for.


## One residual, then the sum of squares

The cell uses the group code to choose the prediction, then accumulates the squared residuals.


```python
# Observed minus the group prediction, then the sum of those squared residuals.
# Rest rows are coded 0. Trained rows are coded 1. The outcome follows the same order.
group = [0, 0, 0, 1, 1, 1]
outcome = [2, 4, 6, 5, 7, 9]
# The line from the previous notebook.
intercept = 4
slope = 3

# The residual for the first rest value, computed on its own so it can be printed.
first_prediction = intercept + slope * group[0]
first_residual = outcome[0] - first_prediction

# A running total of the squared residuals.
total = 0
for i in range(len(outcome)):
    # The line's prediction for this row.
    prediction = intercept + slope * group[i]
    # Observed minus predicted.
    residual = outcome[i] - prediction
    total = total + residual ** 2
print(first_residual)
print(total)
```
