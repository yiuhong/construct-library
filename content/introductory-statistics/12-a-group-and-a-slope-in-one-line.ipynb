# A group and a slope in one line

How can one line carry a group difference and a slope together?

The group comparison and the straight-line mean are the same kind of term. Put both in one equation:

$$
\text{mean} = b_0 + b_{\text{group}} \times \text{group} + b_x x
$$

These four rows fit that line with nothing left over.

| group | $x$ | outcome |
| --- | --- | --- |
| 0 | 0 | 1 |
| 0 | 1 | 3 |
| 1 | 0 | 2 |
| 1 | 1 | 4 |

$b_0 = 1$, $b_{\text{group}} = 1$, and $b_x = 2$. At group 1 and $x = 1$ the mean is $1 + 1 + 2 = 4$. The group term shifts the mean by 1. The slope still adds 2 when $x$ increases by 1. ANOVA-style group comparisons and regression sit in this one linear model.


## Check each row against the line

The loop builds the prediction from the three coefficients. Each prediction matches the outcome, so every residual is 0.


```python
# One line with a group term and a slope. These rows were chosen so the line fits exactly.
group = [0, 0, 1, 1]
x = [0, 1, 0, 1]
outcome = [1, 3, 2, 4]
intercept = 1
group_coefficient = 1
slope = 2

predicted = []
for i in range(len(outcome)):
    # Shared intercept, plus the group shift, plus the slope times x.
    mean = intercept + group_coefficient * group[i] + slope * x[i]
    predicted.append(mean)
print(group_coefficient)
print(predicted)
```
