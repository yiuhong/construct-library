# A point estimate

Which single number from the sample do I use as an estimate of the population mean?

This notebook is in Introductory statistics, in the module Estimation. A point estimate is one number. For the centre of a numerical measurement, this lesson uses the sample mean.

The sample is `[3, 5, 7]`. The estimate is $(3 + 5 + 7) / 3 = 5$. If the population mean were 6, the error would be $5 - 6 = -1$, and the absolute error would be 1. The sign says the estimate is below the population mean. The absolute error says how far, without the sign.


## Estimate, then absolute error

The cell computes the sample mean and the absolute distance from a stated population mean.


```python
# The sample mean is the point estimate. Then measure how far it sits from a stated population mean.
# The sample.
sample = [3, 5, 7]
# A population mean used only to measure the error. The sample does not contain this number.
population_mean = 6
# The point estimate.
estimate = sum(sample) / len(sample)
# Signed error: estimate minus the population mean.
error = estimate - population_mean
# Drop the sign.
absolute_error = abs(error)
print(estimate)
print(absolute_error)
```
