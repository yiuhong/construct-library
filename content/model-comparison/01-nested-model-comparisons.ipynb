# Nested model comparisons

How can I compare a model with a larger one that can copy it?

One model is nested in another when the larger model can copy it by fixing an extra parameter. The constant model predicts every outcome with one number, the mean. A line $y = b + ax$ copies that constant when the slope $a$ is fixed at 0. The line is the larger model. The constant is nested inside it.

On the way to its best fit, the larger model can do anything the smaller one can do. Its best sum of squared errors is therefore never higher, if the search finds that best fit. A lower loss is not yet a reason to keep the extra parameter. It is only the first comparison. The next lessons add a penalty and a check on new points.

The points are $x = 1, 2, 3$ and $y = 2, 2, 5$.


```python
# Error if every outcome is predicted by the mean, then error if a line is allowed to tilt. The drop is how much the tilt buys.
# The inputs.
x = [1, 2, 3]
# The outcomes.
y = [2, 2, 5]
# How many points.
n = 3
# Totals the line formula needs.
sum_x = sum(x)
sum_y = sum(y)
sum_xx = sum(v * v for v in x)
sum_xy = sum(a * b for a, b in zip(x, y))

# The constant prediction: the mean of the outcomes.
mean = sum_y / n
# Squared error of predicting every outcome by that mean.
sse_constant = sum((v - mean) ** 2 for v in y)
# The mean, and that error.
print(mean, sse_constant)

# The shared denominator of the slope and the intercept.
det = n * sum_xx - sum_x * sum_x
# The intercept.
b = (sum_y * sum_xx - sum_x * sum_xy) / det
# The slope.
a = (n * sum_xy - sum_x * sum_y) / det
# Squared error of the tilting line.
sse_line = sum((y[i] - (b + a * x[i])) ** 2 for i in range(n))
# Intercept, slope, and the line's error.
print(round(b, 3), round(a, 3), round(sse_line, 3))
# How much the error fell when the line was allowed to tilt.
print(round(sse_constant - sse_line, 3))
```

## Reading the drop

The constant model has a sum of squared errors of 6. The line has 1.5. The drop is 4.5. The best intercept on these points is 0, so the winning line happens to go through the origin. The constant model is still nested in the line, because the line could have used a slope of 0. It just preferred not to.

If the outcomes were all equal, the constant model would already have a loss of 0. The line could not improve on that.
