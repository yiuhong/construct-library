# Evidence versus prediction

How can I tell a training fit apart from an error on a new point?

Evidence, here, means the fit on the points already used to choose the model. Prediction means the error on a point that was not used.

Training points: $x = 1, 2, 3$ and $y = 1, 2, 3$. The constant model predicts 2, and its sum of squared errors is 2. The line $y = x$ fits these three points exactly, so its sum of squared errors is 0. On the training points, the line wins.

New points: $(4, 2)$ and $(5, 2)$. The constant model still predicts 2, so both squared errors are 0. The line predicts 4 and 5. It wins on the training points and loses on the new ones. A better fit to the sample is not the same claim as a better prediction.


```python
# On the training points the line fits exactly and a constant prediction does not.
train_y = [1, 2, 3]
constant = sum(train_y) / len(train_y)
sse_constant = sum((v - constant) ** 2 for v in train_y)
sse_line = sum((train_y[i] - (i + 1)) ** 2 for i in range(3))
print(constant, sse_constant, sse_line)

# On new points, the constant prediction can land closer than the line that fit the training set.
new = [(4, 2), (5, 2)]
constant_errors = [(y - constant) ** 2 for x, y in new]
line_errors = [(y - x) ** 2 for x, y in new]
print(sum(constant_errors) / len(new))
print(sum(line_errors) / len(new))
```

## What the comparison is for

The training scores were 2 for the constant and 0 for the line. On the two new points the mean squared errors are 0 for the constant and 6.5 for the line. The order reverses.

Neither number replaces the other. The training loss says how the model sat on the points it was given. The held-out error says how a prediction did. A report that quotes only the training loss has not yet said anything about prediction.
