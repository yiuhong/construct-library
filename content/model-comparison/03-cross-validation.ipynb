# Cross-validation

How can I score a point that did not help to choose the prediction?

AIC scores the same points that were used to fit the model. Cross-validation holds a point out, fits on the rest, and scores the held-out point. The score is about prediction.

Leave-one-out, for a mean: each value is predicted by the mean of the others. For $[2, 4, 6]$:

- hold out 2, predict $(4 + 6) / 2 = 5$, squared error 9
- hold out 4, predict $(2 + 6) / 2 = 4$, squared error 0
- hold out 6, predict $(2 + 4) / 2 = 3$, squared error 9

The mean of those squared errors is 6.


```python
data = [2, 4, 6]
errors = []
for i in range(3):
    others = [data[j] for j in range(3) if j != i]
    prediction = sum(others) / len(others)
    errors.append((data[i] - prediction) ** 2)
    print(data[i], prediction, errors[-1])
print(sum(errors) / len(errors))

mean = sum(data) / len(data)
insample = sum((v - mean) ** 2 for v in data) / len(data)
print(round(insample, 3))
```

## Why the in-sample number is smaller

The last print is the average squared distance from the mean of all three points, about 2.667. That is smaller than 6. Those three points chose the prediction, and then they were scored as well. Leave-one-out scores a point that did not help to choose the prediction. That is the harder test, and it matches the question: what would this mean say about a value it did not see?
