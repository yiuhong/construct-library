# AIC and BIC

A lower loss is not free when the model spends an extra parameter. Akaike's information criterion (Akaike, 1974) and Schwarz's criterion, usually called the Bayesian information criterion or BIC (Schwarz, 1978), add a penalty so the extra parameter has to earn its place.

Akaike, H. (1974). A new look at the statistical model identification. *IEEE Transactions on Automatic Control, 19*(6), 716–723.

Schwarz, G. (1978). Estimating the dimension of a model. *The Annals of Statistics, 6*(2), 461–464.

For a least-squares fit with $n$ points, residual sum of squares $S$, and $k$ parameters in the mean, this notebook uses

$$
\mathrm{AIC} = n \ln(S / n) + 2k
$$

$$
\mathrm{BIC} = n \ln(S / n) + k \ln(n)
$$

Constants that do not change which model wins are left out. $k$ counts parameters in the mean, not a separate noise parameter. Lower is better. BIC's penalty grows with $n$, so it asks more of an extra parameter than AIC does.


```python
import math

def aic(n, S, k):
    return n * math.log(S / n) + 2 * k

def bic(n, S, k):
    return n * math.log(S / n) + k * math.log(n)

n = 4
S = 4
for k in [1, 2]:
    print(k, round(aic(n, S, k), 3), round(bic(n, S, k), 3))
```

## The same loss, then a real drop

With $S = 4$ and $n = 4$, the loss term is 0 because $S / n = 1$ and $\ln 1 = 0$. Both criteria then prefer $k = 1$. The second parameter did not reduce $S$, so the penalty is the whole difference.

The next cell is a pair where the larger model does reduce the loss. Model A has $k = 1$ and $S = 8$. Model B has $k = 2$ and $S = 2$. The same $n = 4$.


```python
import math

def aic(n, S, k):
    return n * math.log(S / n) + 2 * k

def bic(n, S, k):
    return n * math.log(S / n) + k * math.log(n)

n = 4
print(round(aic(n, 8, 1), 3), round(bic(n, 8, 1), 3))
print(round(aic(n, 2, 2), 3), round(bic(n, 2, 2), 3))
```
