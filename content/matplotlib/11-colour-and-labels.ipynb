# Colour and labels

What must be written on a figure for it to be read away from this notebook?

This notebook is in Matplotlib, in the module Polish and export. The scatter in Scatterplots used colour and size. A reader who has only the figure still needs the units, a title that names the measurement, and a legend that names the groups.

Colour is a weak way to separate groups on its own. The two seminars here also use different markers: a circle for morning and a square for afternoon. The colours differ in lightness as well, a near-black and a mid grey, so the groups are still distinct if the figure is read in grey. The argument that sets the colour is spelled `color`. That is Matplotlib's spelling.


## A figure that can leave the notebook

The same two seminars as in Scatterplots. The size is still the session count. The labels say so in the caption of the axes, not only in this paragraph. The first plot in a session can take a moment while the plotting library downloads.


```python
# Label the scatter so the groups, the units, and the size can be read from the figure.
import matplotlib.pyplot as plt

# Morning: revision hours, test score, practice sessions.
morning_hours = [2, 3, 4, 5, 6]
morning_score = [4, 5, 6, 7, 8]
morning_sessions = [1, 2, 2, 3, 4]
# Afternoon, in the same three columns.
afternoon_hours = [2, 4, 5, 7, 8]
afternoon_score = [3, 5, 6, 6, 9]
afternoon_sessions = [1, 1, 3, 2, 4]

fig, ax = plt.subplots(figsize=(6.6, 4.0), layout="constrained")
# A circle, near-black. s is the marker area: 40 times the session count.
ax.scatter(
    morning_hours,
    morning_score,
    s=[40 * count for count in morning_sessions],
    color="#1c1c1c",
    marker="o",
    label="Morning",
)
# A square, mid grey, so the group is not carried by colour alone.
ax.scatter(
    afternoon_hours,
    afternoon_score,
    s=[40 * count for count in afternoon_sessions],
    color="#8a8a8a",
    marker="s",
    label="Afternoon",
)
# The horizontal label names the measurement.
ax.set_xlabel("Revision hours")
ax.set_ylabel("Test score")
# The title says what the points are, including the size.
ax.set_title("Test score against revision. Marker area is practice sessions.")
ax.legend(loc="upper left", title="Seminar")
plt.show()
```
