# Box plots

What does the box show that the median alone does not?

You will draw the Monday and Thursday quizzes with the recorded 18 held out, put that score back, and compute the fence that leaves it beyond the whisker. By the end you should be able to read the median, the middle half, and a point that sits outside the whisker.

The line in the box is the median. The box runs from the lower quartile to the upper quartile, the middle half of the scores, using Matplotlib's default quartile rule. The whiskers run out to the last score that still sits within 1.5 box-heights of either end of the box. That 1.5 rule is Matplotlib's default. A score beyond the whisker is drawn as its own point.

The previous lesson showed the pile of all twelve scores. A box plot reduces each seminar to that middle half. These boxes are a sketch of two short lists. A published box plot would name the quartile rule and the sample size. A different rule can move the ends of the box.


## The same median, the 18 held out

Monday scores are 4, 5, 6, 6, 7, and 8. Thursday, with the recorded 18 held out, is 4, 5, 6, 7, and 8. Both medians are 6. The boxes are not the same height.

The cell prints the lower quartile, the median, and the upper quartile for each list. Those numbers come from NumPy's default percentile, which is the rule Matplotlib's box plot uses here. They describe these lists. They are not a finding about a seminar.

The first plot in a session can take a moment while the plotting library downloads.


```python
# Two box plots, with the recorded 18 held out. Both medians are 6. The boxes differ.
import matplotlib.pyplot as plt
import numpy as np

# Six Monday scores, already in order.
monday = [4, 5, 6, 6, 7, 8]
# Five Thursday scores. The recorded 18 is held out. The median matches Monday.
thursday = [4, 5, 6, 7, 8]

# Lower quartile, median, and upper quartile, with NumPy's default percentile.
print([float(value) for value in np.percentile(monday, [25, 50, 75])])
print([float(value) for value in np.percentile(thursday, [25, 50, 75])])

fig, ax = plt.subplots(figsize=(6.2, 3.8), layout="constrained")
# tick_labels names the two boxes. The default whisker is 1.5 box-heights.
ax.boxplot([monday, thursday], tick_labels=["Monday", "Thursday"])
ax.set_ylabel("Weekly quiz")
ax.set_title("Same median, the recorded 18 held out")
plt.show()
```

## The 18 put back

Monday still prints `[5.25, 6.0, 6.75]`. The box runs from 5.25 to 6.75, and the line is at 6. Thursday without the 18 prints `[5.0, 6.0, 7.0]`. The box runs from 5 to 7, and the line is again at 6. The Thursday middle half is wider. The medians do not show that.

Put the recorded 18 back and Thursday is 4, 5, 6, 7, 8, and 18. The median is now 6.5, the average of the two middle scores, 6 and 7. The cell prints the new quartiles and draws the boxes again. The 18 is a separate point. The whisker does not stretch out to it. This cell uses `plt` from the cell above.


```python
# Thursday with the recorded 18 put back. The median moves, and 18 is drawn on its own.
monday = [4, 5, 6, 6, 7, 8]
thursday = [4, 5, 6, 7, 8, 18]

# The Thursday quartiles now include the recorded 18.
print([float(value) for value in np.percentile(thursday, [25, 50, 75])])

fig, ax = plt.subplots(figsize=(6.2, 3.8), layout="constrained")
ax.boxplot([monday, thursday], tick_labels=["Monday", "Thursday"])
ax.set_ylabel("Weekly quiz")
ax.set_title("The recorded 18 put back")
plt.show()
```

## The fence past the upper quartile

The box height is the upper quartile minus the lower quartile. Matplotlib's default whisker stops at the last score within 1.5 box-heights of the box. For Thursday with the 18 included, the quartiles are 5.25, 6.5, and 7.75. The box height is `7.75 - 5.25`, which is 2.5. The upper fence is `7.75 + 1.5 * 2.5`, which is 11.5.

The highest Thursday score that is still at or below 11.5 is 8, so the whisker stops at 8. The recorded 18 sits above the fence, so it is drawn as its own point. This cell prints the box height and the upper fence. It uses `np` from the first code cell.


```python
# The upper fence for Thursday once the recorded 18 is included.
thursday = [4, 5, 6, 7, 8, 18]
# Lower quartile, median, and upper quartile.
q1, med, q3 = [float(value) for value in np.percentile(thursday, [25, 50, 75])]
# The box height is the middle half, from the lower quartile to the upper quartile.
box_height = q3 - q1
# 1.5 box-heights past the upper quartile. Scores above this fence are drawn alone.
upper_fence = q3 + 1.5 * box_height
print(box_height)
print(upper_fence)
```
