# Heat maps

How does a table of scores become a grid of colour?

This notebook is in Matplotlib, in the module Common plots. Five students, A to E, each answered four questions. `ax.imshow` paints one cell per score. Darker is higher only because this colour map, `Greys`, was chosen that way. Another map could reverse it. The colour bar is the legend for the shading.

`imshow` puts the first row of the array at the top of the picture. Student A is that first row, so A is the top row. If you expected the first row at the bottom, the picture would look flipped. The numbers written in the cells are the scores, so the shading is not the only way to read the table.


## Five students, four questions

The rows are A, B, C, D, E. The columns are Q1, Q2, Q3, Q4. The first plot in a session can take a moment while the plotting library downloads.


```python
# Paint a small score table. The first row of the array is the top row of the picture.
import matplotlib.pyplot as plt
import numpy as np

# One row per student, A to E. One column per question, Q1 to Q4.
scores = np.array([
    [4, 5, 6, 5],
    [6, 6, 7, 8],
    [3, 4, 4, 5],
    [7, 8, 8, 9],
    [5, 5, 6, 6],
])

fig, ax = plt.subplots(figsize=(6.2, 4.4), layout="constrained")
# Greys: a higher score is a darker cell.
image = ax.imshow(scores, cmap="Greys")
# Name the columns.
ax.set_xticks([0, 1, 2, 3])
ax.set_xticklabels(["Q1", "Q2", "Q3", "Q4"])
# Name the rows. Row 0 is the top row, student A.
ax.set_yticks([0, 1, 2, 3, 4])
ax.set_yticklabels(["A", "B", "C", "D", "E"])
ax.set_xlabel("Question")
ax.set_ylabel("Student")
ax.set_title("Short test")
# The colour bar says what a dark cell means.
fig.colorbar(image, ax=ax, label="Score")
# Write the score in the cell. Light ink on a dark cell, dark ink on a light cell.
for row in range(scores.shape[0]):
    for column in range(scores.shape[1]):
        value = int(scores[row, column])
        ink = "#ffffff" if value >= 7 else "#1c1c1c"
        ax.text(column, row, str(value), ha="center", va="center", color=ink)
plt.show()
```
