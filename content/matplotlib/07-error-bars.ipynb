# Error bars

What does the line through a mean add?

You will draw the Monday and Thursday means with a standard error, hold the recorded 18 out of the Thursday mean, and then put the six Thursday scores on the same axes as that line. By the end you should be able to say what the line measures.

The point is the mean. The line is the standard error: the sample standard deviation, divided by the square root of the sample size. The sample standard deviation uses `n - 1` on the squared distances from the mean. The line is one summary of how far the mean moved inside this small list. It is a standard error, which is a different object from the range of the scores, and a different object from a confidence interval. Six scores, or five when one row is held out, is too small to report as a result about a seminar.

The previous lesson showed the middle half of each seminar. The line here is about the mean, not about that middle half.


## Two seminar means

Monday is 4, 5, 6, 6, 7, and 8. The total is 36, so the mean is 6. Thursday is 4, 5, 6, 7, 8, and 18. The total is 48, so the mean is 8. Monday's standard error rounds to 0.577. Thursday's rounds to 2.082. The Thursday line is longer because the recorded 18 sits far from 8.

The first plot in a session can take a moment while the plotting library downloads.


```python
# A point for each seminar mean, and a line for the standard error.
import matplotlib.pyplot as plt

def standard_error(values):
    # The sample size.
    n = len(values)
    # The mean of this seminar.
    mean = sum(values) / n
    # The sample variance: squared deviations, divided by n - 1.
    spread = sum((value - mean) ** 2 for value in values) / (n - 1)
    # The standard deviation, then divided by the square root of n.
    return (spread ** 0.5) / (n ** 0.5)

# Six Monday scores, and six Thursday scores including the recorded 18.
monday = [4, 5, 6, 6, 7, 8]
thursday = [4, 5, 6, 7, 8, 18]
seminars = [monday, thursday]
names = ["Monday", "Thursday"]

# One mean and one standard error per seminar.
means = [sum(values) / len(values) for values in seminars]
errors = [standard_error(values) for values in seminars]
# The means, then the standard errors rounded to three decimal places.
print(means)
print([round(value, 3) for value in errors])

fig, ax = plt.subplots(figsize=(6.4, 3.8), layout="constrained")
# fmt="o" draws the mean as a point. yerr draws the line above and below it.
ax.errorbar(names, means, yerr=errors, fmt="o", color="#1c1c1c", capsize=4)
ax.set_ylabel("Mean weekly quiz")
ax.set_title("Six scores per seminar")
# Start at zero so the means keep their height. Leave room for the Thursday line.
ax.set_ylim(0, 12)
plt.show()
```

## The same means, with 18 held out

Hold out the recorded 18 and Thursday is 4, 5, 6, 7, and 8. The total is 30, so the mean is 6, the same as Monday. Five scores remain, so the Thursday standard error divides by the square root of 5, not of 6. It rounds to 0.707. Monday is still six scores, and its standard error is still 0.577. The two lines are not from the same sample size. This cell uses `plt` and `standard_error` from the first code cell.


```python
# The same two means, with the recorded 18 held out of Thursday.
monday = [4, 5, 6, 6, 7, 8]
# Five Thursday scores. The sample size is no longer 6.
thursday = [4, 5, 6, 7, 8]
seminars = [monday, thursday]
names = ["Monday", "Thursday"]

means = [sum(values) / len(values) for values in seminars]
errors = [standard_error(values) for values in seminars]
print(means)
print([round(value, 3) for value in errors])
# The two sample sizes, so the different divisors stay visible.
print([len(values) for values in seminars])

fig, ax = plt.subplots(figsize=(6.4, 3.8), layout="constrained")
ax.errorbar(names, means, yerr=errors, fmt="o", color="#1c1c1c", capsize=4)
ax.set_ylabel("Mean weekly quiz")
ax.set_title("The recorded 18 held out")
ax.set_ylim(0, 10)
plt.show()
```

## The line is not the range

The six Thursday scores are drawn as points, with the mean and its standard error on the same axes. The range is the largest score minus the smallest, `18 - 4`, which is 14. The standard error rounds to 2.082. The line covers that standard error on either side of the mean. It does not run out to 18, and it is not a confidence interval.

The cell prints the range and the standard error. This cell uses `plt` and `standard_error` from the first code cell.


```python
# The six Thursday scores as points, and the standard error of their mean as a line.
thursday = [4, 5, 6, 7, 8, 18]
# The mean is 8. The range uses only the lowest and highest scores.
thursday_mean = sum(thursday) / len(thursday)
thursday_range = max(thursday) - min(thursday)
thursday_error = standard_error(thursday)
print(thursday_range)
print(round(thursday_error, 3))

fig, ax = plt.subplots(figsize=(6.2, 3.8), layout="constrained")
# One point per Thursday score, stacked on a single horizontal position.
ax.scatter([1] * len(thursday), thursday, color="#8a8a8a", zorder=2)
# The mean, with the standard error above and below it.
ax.errorbar(
    [1],
    [thursday_mean],
    yerr=[thursday_error],
    fmt="o",
    color="#1c1c1c",
    capsize=4,
    zorder=3,
)
ax.set_xlim(0, 2)
ax.set_xticks([1])
ax.set_xticklabels(["Thursday"])
ax.set_ylabel("Weekly quiz")
ax.set_title("The scores, the mean, and the standard error")
ax.set_ylim(0, 20)
plt.show()
```
