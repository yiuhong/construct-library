# Colour, labels, annotations, and saving

What has to be on a figure that leaves this notebook, and what sets the pixel width of the file?

You will label a scatter so the seminar, the units, and the marker area can be read from the figure, point at the score recorded above 10, and save that figure at two values of `dpi`. By the end you should be able to name the seminars from the legend, and to get the pixel width from the inches and the dpi.

Colour is a weak way to separate the seminars on its own. The Monday points use a circle and a near-black. The Thursday points use a square and a mid grey, so the groups still differ if the figure is read in grey. The argument that sets the colour is spelled `color`. That is Matplotlib's spelling.

The previous lesson painted a grid and added a colour bar. This lesson stays with the scatter of revision hours and the weekly quiz. The page is 6 inches wide and 4 inches tall. Without `bbox_inches="tight"`, the pixel width is the width in inches times `dpi`, and the pixel height is the height in inches times `dpi`.


## Labels that can leave the notebook

A reader who has only the figure still needs the units, a title that names the measurement, and a legend that names the seminars. The size is still the practice count, and the title says so. Marker area is 40 times the session count.

Monday revision hours are 1, 2, 3, 3, 4, and 5, with quizzes 4, 5, 6, 6, 7, and 8, and practice counts 1, 1, 2, 2, 3, and 3. Thursday revision hours are 2, 2, 3, 3, 4, and 4, with quizzes 4, 5, 6, 7, 8, and 18, and practice counts 1, 2, 2, 2, 3, and 4.

The first plot in a session can take a moment while the plotting library downloads.


```python
# Label the scatter so the seminars, the units, and the size can be read from the figure.
import matplotlib.pyplot as plt

# Monday: revision hours, weekly quiz, practice sessions.
monday_hours = [1, 2, 3, 3, 4, 5]
monday_quiz = [4, 5, 6, 6, 7, 8]
monday_sessions = [1, 1, 2, 2, 3, 3]
# Thursday, in the same three columns. The last quiz was recorded as 18.
thursday_hours = [2, 2, 3, 3, 4, 4]
thursday_quiz = [4, 5, 6, 7, 8, 18]
thursday_sessions = [1, 2, 2, 2, 3, 4]

# Six inches wide and four inches tall.
fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")
# A circle, near-black. s is the marker area: 40 times the session count.
ax.scatter(
    monday_hours,
    monday_quiz,
    s=[40 * count for count in monday_sessions],
    color="#1c1c1c",
    marker="o",
    label="Monday",
)
# A square, mid grey, so the seminar is not carried by colour alone.
ax.scatter(
    thursday_hours,
    thursday_quiz,
    s=[40 * count for count in thursday_sessions],
    color="#8a8a8a",
    marker="s",
    label="Thursday",
)
# The horizontal label names the measurement.
ax.set_xlabel("Revision hours")
ax.set_ylabel("Weekly quiz")
# The title says what the points are, including the size.
ax.set_title("Weekly quiz against revision. Marker area is practice sessions.")
ax.set_ylim(0, 20)
ax.legend(loc="upper left", title="Seminar")
plt.show()
```

## A note on the score above 10

The weekly quiz is marked out of 10. One Thursday score was recorded as 18, at 4 revision hours. The loop stops at the first score above 10.

`ax.axhline` draws the marked maximum across the axes. It is a reference line, not a fitted line. `ax.annotate` puts the words at `xytext` and draws an arrow to `xy`. The words sit off the point, so they do not cover the marker. Two Thursday students revised for 4 hours. The arrow points at the score of 18, not at the score of 8. This cell uses `plt` from the cell above.


```python
# Mark the quiz maximum, and point at the first score recorded above it.
revision_hours = [1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, 4]
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
sessions = [1, 1, 2, 2, 3, 3, 1, 2, 2, 2, 3, 4]
# Monday is the first six rows. Thursday is the last six.
seminar = ["Monday"] * 6 + ["Thursday"] * 6
# The marked maximum.
maximum = 10

# Walk the pairs until a score is above the maximum, then stop.
found_hours = None
found_score = None
for hours, score in zip(revision_hours, weekly_quiz):
    if score > maximum:
        found_hours = hours
        found_score = score
        break
print(found_score)

fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")
# Draw each seminar with its own marker, so the note sits on the labelled figure.
for name, marker, colour in (("Monday", "o", "#1c1c1c"), ("Thursday", "s", "#8a8a8a")):
    xs = [hours for hours, group in zip(revision_hours, seminar) if group == name]
    ys = [score for score, group in zip(weekly_quiz, seminar) if group == name]
    areas = [40 * count for count, group in zip(sessions, seminar) if group == name]
    ax.scatter(xs, ys, s=areas, color=colour, marker=marker, label=name)
# A dashed reference line at the marked maximum. It is not a fit.
ax.axhline(maximum, color="#b5b5b5", linestyle="--")
# xy is the point. xytext is where the words sit, in the same hours-and-quiz units.
ax.annotate(
    "Recorded above the maximum",
    xy=(found_hours, found_score),
    xytext=(1.0, 16),
    arrowprops={"arrowstyle": "->", "color": "#1c1c1c"},
)
ax.set_xlabel("Revision hours")
ax.set_ylabel("Weekly quiz")
ax.set_title("A note on the score recorded as 18")
ax.set_xlim(0, 6)
ax.set_ylim(0, 20)
ax.legend(loc="upper left", title="Seminar")
plt.show()
```

## Pixel width follows the inches and the dpi

`fig.savefig` writes the figure to a file. This figure is 6 inches wide and 4 inches tall. At `dpi=100` the PNG is 600 pixels wide. At `dpi=150` a second PNG of the same figure is 900 pixels wide. The inches do not change. The pixel count does. The height in pixels is 4 times the dpi. The checks ask for the width.

`bbox_inches="tight"` would crop the spare margin, and the pixel count would then no longer be just inches times `dpi`. These two files do not use that argument.

The files are written in the notebook's folder in this browser. Open the file browser and look in `matplotlib` for `weekly-quiz-scatter.png` and `weekly-quiz-scatter-fine.png`. Download from there if you want a copy. Nothing is uploaded. Saving does not replace `plt.show`: the cell still draws the figure under the cell, and it also writes the files. This cell uses `plt` from the cell above.


```python
# Save the labelled, annotated figure twice. The inches stay put. The dpi changes the pixel count.
revision_hours = [1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, 4]
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
sessions = [1, 1, 2, 2, 3, 3, 1, 2, 2, 2, 3, 4]
seminar = ["Monday"] * 6 + ["Thursday"] * 6
maximum = 10

found_hours = None
found_score = None
for hours, score in zip(revision_hours, weekly_quiz):
    if score > maximum:
        found_hours = hours
        found_score = score
        break

# Six inches wide and four inches tall.
fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")
for name, marker, colour in (("Monday", "o", "#1c1c1c"), ("Thursday", "s", "#8a8a8a")):
    xs = [hours for hours, group in zip(revision_hours, seminar) if group == name]
    ys = [score for score, group in zip(weekly_quiz, seminar) if group == name]
    areas = [40 * count for count, group in zip(sessions, seminar) if group == name]
    ax.scatter(xs, ys, s=areas, color=colour, marker=marker, label=name)
ax.axhline(maximum, color="#b5b5b5", linestyle="--")
ax.annotate(
    "Recorded above the maximum",
    xy=(found_hours, found_score),
    xytext=(1.0, 16),
    arrowprops={"arrowstyle": "->", "color": "#1c1c1c"},
)
ax.set_xlabel("Revision hours")
ax.set_ylabel("Weekly quiz")
ax.set_title("Weekly quiz against revision. Marker area is practice sessions.")
ax.set_xlim(0, 6)
ax.set_ylim(0, 20)
ax.legend(loc="upper left", title="Seminar")

# 6 inches times 100 dpi is 600 pixels wide. 4 inches times 100 is 400 pixels tall.
fig.savefig("weekly-quiz-scatter.png", dpi=100)
# The same inches at a finer dpi: 6 times 150 is 900 pixels wide.
fig.savefig("weekly-quiz-scatter-fine.png", dpi=150)
# Show the figure under this cell as well as saving it.
plt.show()
```
