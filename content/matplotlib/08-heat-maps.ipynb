# Heat maps

How does a table of scores become a grid of colour?

You will paint the six Monday rows, then all twelve rows, and then a small grid of questionnaire counts. By the end you should be able to say which row of the array is the top row of the picture, and when the columns do not share a unit.

`imshow` paints one cell per number. Darker is higher only because this colour map, `Greys`, was chosen that way. Another map could reverse it. The colour bar is the legend for the shading. `imshow` puts the first row of the array at the top of the picture. The numbers written in the cells repeat the table, so the shading is not the only way to read it.

The previous lesson drew means and a standard error. Those summaries drop the rows. A heat map keeps a grid.


## Six Monday rows

Each row is one Monday student, in weekly quiz class order. The columns are the weekly quiz, revision hours, practice sessions, and the mini-test. The first row of the array is the student who scored 4, so that student is the top row of the picture.

The four columns are not the same unit. A dark cell in revision hours is a larger number of hours. It is not a higher quiz. The colour is comparable inside one column more fairly than across columns. The grid's largest number is 8, and light ink is used on the darker cells.

The first plot in a session can take a moment while the plotting library downloads.


```python
# Paint the six Monday rows. The first row of the array is the top row of the picture.
import matplotlib.pyplot as plt
import numpy as np

# One row per Monday student. Columns: weekly quiz, revision hours, practice, mini-test.
monday = np.array([
    [4, 1, 1, 1],
    [5, 2, 1, 2],
    [6, 3, 2, 2],
    [6, 3, 2, 3],
    [7, 4, 3, 3],
    [8, 5, 3, 4],
])

fig, ax = plt.subplots(figsize=(6.6, 4.4), layout="constrained")
# Greys: a higher number is a darker cell, within the range of this grid.
image = ax.imshow(monday, cmap="Greys")
ax.set_xticks([0, 1, 2, 3])
ax.set_xticklabels(["Quiz", "Revision", "Practice", "Mini-test"])
# Row 0 is the top row, the first Monday student.
ax.set_yticks([0, 1, 2, 3, 4, 5])
ax.set_yticklabels(["1", "2", "3", "4", "5", "6"])
ax.set_xlabel("Column")
ax.set_ylabel("Monday row")
ax.set_title("Monday rows")
# The colour bar says what a dark cell means. The columns do not share a unit.
fig.colorbar(image, ax=ax, label="Value")
# Light ink on a dark cell, dark ink on a light cell.
peak = float(monday.max())
for row in range(monday.shape[0]):
    for column in range(monday.shape[1]):
        value = int(monday[row, column])
        ink = "#ffffff" if value >= peak * 0.7 else "#1c1c1c"
        ax.text(column, row, str(value), ha="center", va="center", color=ink)
plt.show()
```

## All twelve rows

The same four columns, for every student. Rows 1 to 6 are Monday. Rows 7 to 12 are Thursday. The last row's quiz cell is the recorded 18.

That one cell stretches the colour scale. A quiz of 8, which was among the darkest cells in the Monday grid, is much lighter here, because the scale now runs up to 18. The numbers in the cells are unchanged. This cell uses `plt` and `np` from the first code cell.


```python
# All twelve rows. The recorded 18 stretches the colour scale.
# Columns stay in the same order: weekly quiz, revision hours, practice, mini-test.
scores = np.array([
    [4, 1, 1, 1],
    [5, 2, 1, 2],
    [6, 3, 2, 2],
    [6, 3, 2, 3],
    [7, 4, 3, 3],
    [8, 5, 3, 4],
    [4, 2, 1, 2],
    [5, 2, 2, 2],
    [6, 3, 2, 3],
    [7, 3, 2, 3],
    [8, 4, 3, 3],
    [18, 4, 4, 4],
])

fig, ax = plt.subplots(figsize=(6.6, 6.4), layout="constrained")
image = ax.imshow(scores, cmap="Greys")
ax.set_xticks([0, 1, 2, 3])
ax.set_xticklabels(["Quiz", "Revision", "Practice", "Mini-test"])
ax.set_yticks(range(12))
ax.set_yticklabels([str(number) for number in range(1, 13)])
ax.set_xlabel("Column")
ax.set_ylabel("Row in the weekly quiz class")
ax.set_title("All twelve rows")
fig.colorbar(image, ax=ax, label="Value")
peak = float(scores.max())
for row in range(scores.shape[0]):
    for column in range(scores.shape[1]):
        value = int(scores[row, column])
        ink = "#ffffff" if value >= peak * 0.7 else "#1c1c1c"
        ax.text(column, row, str(value), ha="center", va="center", color=ink, fontsize=8)
plt.show()
```

## A grid of counts

Every cell here is a count of students, so the colour is one unit. The rows are the seminars. The columns are the questionnaire item `confident`, yes and then no. Monday is 4 and 2. Thursday is 5 and 1. The first row of the array is Monday, and Monday is the top row of the picture. This cell uses `plt` and `np` from the first code cell.


```python
# Seminar by the questionnaire item. Every cell is a count of students.
# Rows are Monday, then Thursday. Columns are yes, then no.
counts = np.array([
    [4, 2],
    [5, 1],
])

fig, ax = plt.subplots(figsize=(5.4, 3.6), layout="constrained")
image = ax.imshow(counts, cmap="Greys")
ax.set_xticks([0, 1])
ax.set_xticklabels(["Yes", "No"])
ax.set_yticks([0, 1])
ax.set_yticklabels(["Monday", "Thursday"])
ax.set_xlabel("Confident")
ax.set_ylabel("Seminar")
ax.set_title("Counts of the questionnaire item")
fig.colorbar(image, ax=ax, label="Students")
peak = float(counts.max())
for row in range(counts.shape[0]):
    for column in range(counts.shape[1]):
        value = int(counts[row, column])
        ink = "#ffffff" if value >= peak * 0.7 else "#1c1c1c"
        ax.text(column, row, str(value), ha="center", va="center", color=ink)
plt.show()
```
