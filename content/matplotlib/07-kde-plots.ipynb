# KDE plots

What is simplified when a smooth curve is drawn over ten scores?

This notebook is in Matplotlib, in the module Common plots. A kernel density estimate puts a bump on every score and adds the bumps. `scipy.stats.gaussian_kde` does that with normal bumps. The width of each bump is a bandwidth. This notebook keeps the library's default. It does not search for a better one.

For one measurement, that default multiplies the sample standard deviation by a factor. The factor is the sample size raised to the power `-1/5`. For 10 scores, `10 ** (-1/5)` is about 0.631. The cell prints `kde.factor` so you can see that it matches.

The scores are the same ten as in the histogram lesson: `[4, 5, 5, 6, 6, 6, 7, 7, 8, 9]`.


## What this curve is not

This is a picture of ten chosen scores. It is not a density you would report for a class.

What is simplified:

- Ten scores. The curve looks smoother than a list this short can support.
- The bandwidth is the default. It was not chosen by comparing several curves, and it was not checked against the histogram before being kept.
- The bumps are normal, so the curve can spill below 0 or above 10 even though a score on this test cannot.
- The histogram beside the curve uses `density=True`. The bar areas then add to 1, so the height is no longer the count from the histogram lesson. That rescaling lets the bars and the curve share a vertical axis. It is there for the picture, not because the counts stopped mattering.

A fuller density estimate would still need more scores, a bandwidth you chose and checked, the histogram beside the curve, and care at the edges when a score cannot fall outside a range. JupyterLite can run `gaussian_kde`. The gap is the sample and the unchecked default, not a missing command.

The first time this session imports scipy, that library downloads. That can take a moment. The first plot can take a moment as well.


```python
# A normal bump on each of ten scores, added up, using the default bandwidth.
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gaussian_kde

# The same ten scores as the histogram lesson. One row of one measurement.
scores = np.array([4, 5, 5, 6, 6, 6, 7, 7, 8, 9], dtype=float)
# The default estimate. For one measurement, factor is n to the power -1/5.
kde = gaussian_kde(scores)
# The factor the library chose, and the power it should match.
print(kde.factor)
print(10 ** (-1 / 5))

# A grid to draw the curve on. The estimate itself is not limited to these points.
xs = np.linspace(2, 11, 200)
ys = kde(xs)

fig, ax = plt.subplots(figsize=(6.6, 3.6), layout="constrained")
# density=True rescales the bars so their area is about 1. The height is not a count.
ax.hist(scores, bins=[4, 6, 8, 10], density=True, color="#d0d0d0", edgecolor="white")
# The sum of the bumps, drawn on the grid.
ax.plot(xs, ys, color="#1c1c1c")
ax.set_xlabel("Test score")
ax.set_ylabel("Density")
ax.set_title("A smooth curve on ten scores")
plt.show()
```
