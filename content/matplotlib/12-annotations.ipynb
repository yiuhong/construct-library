# Annotations

How do you point at one score without covering it?

This notebook is in Matplotlib, in the module Polish and export. A weekly quiz is `[5, 6, 7, 8, 9]` in weeks `[1, 2, 3, 4, 5]`. The pass mark is 7. Week 3 is the first week at or above that mark.

`ax.axhline` draws the pass mark across the axes. It is a reference line, not a fitted line. `ax.annotate` puts words at `xytext` and can draw an arrow to `xy`. The words sit off the point, so they do not cover the marker.


## The first week at the pass mark

The loop stops at the first score that is at least 7. The arrow points at that week. The first plot in a session can take a moment while the plotting library downloads.


```python
# Mark the pass line, and point at the first week that reaches it.
import matplotlib.pyplot as plt

# The week of the quiz.
weeks = [1, 2, 3, 4, 5]
# The score that week.
scores = [5, 6, 7, 8, 9]
# The pass mark.
mark = 7

# Walk the pairs until a score reaches the mark, then stop.
found = None
for week, score in zip(weeks, scores):
    if score >= mark:
        found = week
        break
print(found)

fig, ax = plt.subplots(figsize=(6.6, 3.8), layout="constrained")
ax.plot(weeks, scores, marker="o", color="#1c1c1c")
# A dashed reference line at the pass mark. It is not a fit.
ax.axhline(mark, color="#b5b5b5", linestyle="--")
# xy is the point. xytext is where the words sit, in the same week-and-score units.
ax.annotate(
    "First week at the pass mark",
    xy=(found, mark),
    xytext=(3.6, 5.5),
    arrowprops={"arrowstyle": "->", "color": "#1c1c1c"},
)
ax.set_xlabel("Week")
ax.set_ylabel("Quiz score")
ax.set_title("A note on the first pass")
ax.set_xlim(0.5, 5.5)
ax.set_ylim(4, 10)
plt.show()
```
