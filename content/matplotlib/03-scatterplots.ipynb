# Scatterplots

How can a scatter show the seminar and a third measurement?

You will draw one point per student, colour it by seminar, and set the marker area from the practice count, first with the score recorded as 18 and then with that row held out. By the end you should be able to read the height as the weekly quiz and the area as the practice sessions.

A scatter keeps two numbers on one panel. Colour can name the seminar. Size can carry the practice count, if you remember that the size is an area and that the legend does not explain it.

The previous lesson put the quiz and the hours on two panels, because those units differ. Here revision hours and the weekly quiz are the two axes, so one panel holds both, and practice is a third column on the same rows.


## Two seminars, colour, and size

Each point is one student in the weekly quiz class. Horizontal position is revision hours. Vertical position is the weekly quiz. The colour is the seminar. The size is the number of practice sessions.

Monday revision hours are 1, 2, 3, 3, 4, and 5, with quizzes 4, 5, 6, 6, 7, and 8, and practice counts 1, 1, 2, 2, 3, and 3. Two Monday students share 3 hours, a quiz of 6, and 2 sessions, so one mark hides the other.

Thursday revision hours are 2, 2, 3, 3, 4, and 4, with quizzes 4, 5, 6, 7, 8, and 18, and practice counts 1, 2, 2, 2, 3, and 4. The vertical axis runs to 20 so the recorded 18 is on the page.

`s` is the marker area, not the diameter. The cell multiplies the session count by 40 so the marks are large enough to see. The legend names the seminar. It does not explain the size. The argument that sets the colour is spelled `color`. That is Matplotlib's spelling. The lesson text uses colour.

The first plot in a session can take a moment while the plotting library downloads.


```python
# Two seminars on one scatter. Colour is the seminar. Size is the practice count.
import matplotlib.pyplot as plt

# Monday: revision hours, weekly quiz, and practice sessions, in class order.
monday_hours = [1, 2, 3, 3, 4, 5]
monday_quiz = [4, 5, 6, 6, 7, 8]
monday_sessions = [1, 1, 2, 2, 3, 3]

# Thursday, in the same three columns. The last quiz was recorded as 18.
thursday_hours = [2, 2, 3, 3, 4, 4]
thursday_quiz = [4, 5, 6, 7, 8, 18]
thursday_sessions = [1, 2, 2, 2, 3, 4]

# One axes for both seminars. The vertical limit leaves room for the recorded 18.
fig, ax = plt.subplots(figsize=(6.6, 4.2), layout="constrained")
# Monday points. s is the marker area: 40 times the session count.
# The argument is spelled color, which is Matplotlib's spelling.
ax.scatter(
    monday_hours,
    monday_quiz,
    s=[40 * count for count in monday_sessions],
    color="#1c1c1c",
    label="Monday",
)
# Thursday points, in a lighter grey, with the same size rule.
ax.scatter(
    thursday_hours,
    thursday_quiz,
    s=[40 * count for count in thursday_sessions],
    color="#8a8a8a",
    label="Thursday",
)
ax.set_xlabel("Revision hours")
ax.set_ylabel("Weekly quiz")
ax.set_ylim(0, 20)
ax.set_title("Revision, quiz, and practice sessions")
# The legend names the two seminars. It does not show what the sizes mean.
ax.legend(loc="upper left")
plt.show()
```

## The recorded 18, held out

The same scatter, without the Thursday row whose quiz was recorded as 18. That student revised for 4 hours and attended 4 practice sessions. Holding the row out lets the other eleven scores use the quiz's own scale, from 0 to 10.

The remaining Thursday revision hours are 2, 2, 3, 3, and 4, with quizzes 4, 5, 6, 7, and 8, and practice counts 1, 2, 2, 2, and 3. Monday is unchanged. Twelve students is a sketch of the class. It is not a result about revision. This cell uses `plt` from the cell above.


```python
# The same scatter with the recorded 18 held out, so the other scores use a 0 to 10 scale.
monday_hours = [1, 2, 3, 3, 4, 5]
monday_quiz = [4, 5, 6, 6, 7, 8]
monday_sessions = [1, 1, 2, 2, 3, 3]

# Thursday without the last row.
thursday_hours = [2, 2, 3, 3, 4]
thursday_quiz = [4, 5, 6, 7, 8]
thursday_sessions = [1, 2, 2, 2, 3]

fig, ax = plt.subplots(figsize=(6.6, 4.2), layout="constrained")
ax.scatter(
    monday_hours,
    monday_quiz,
    s=[40 * count for count in monday_sessions],
    color="#1c1c1c",
    label="Monday",
)
ax.scatter(
    thursday_hours,
    thursday_quiz,
    s=[40 * count for count in thursday_sessions],
    color="#8a8a8a",
    label="Thursday",
)
ax.set_xlabel("Revision hours")
ax.set_ylabel("Weekly quiz")
ax.set_ylim(0, 10)
ax.set_title("The recorded 18 held out")
ax.legend(loc="upper left")
plt.show()
```

## One point is not its area

The Monday student with 5 revision hours scored 8 and attended 3 practice sessions, so that marker's area is `40 * 3`, which is 120. The Thursday student recorded as 18 revised for 4 hours and attended 4 sessions, so that marker's area is `40 * 4`, which is 160.

A larger mark is more sessions. It is a higher quiz only when the point sits higher. The quiz is the height. This cell uses no new figure. It prints the two areas.


```python
# Marker area is 40 times the practice count. The weekly quiz is the height, not the area.
# Monday row with 5 revision hours: quiz 8, practice 3.
monday_area = 40 * 3
# Thursday row recorded as 18: practice 4.
thursday_area = 40 * 4
# 120, then 160. Neither number is the quiz score.
print(monday_area)
print(thursday_area)
```
