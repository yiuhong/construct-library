# Figure size and layout

What does the figure size change, and what keeps the labels on the page?

This notebook is in Matplotlib, in the module Figures and axes. The size is a property of the figure, in inches. The axes sits inside that page. A long label needs room. `layout="constrained"` reserves that room when the figure is drawn.


## A wide page

Six item means from a short questionnaire. The figure is 8 inches wide and 3 inches tall. The cell prints that pair. The first plot in a session can take a moment while the plotting library downloads.


```python
# A wide, short page. figsize is the figure, in inches: width, then height.
import matplotlib.pyplot as plt

# The six questionnaire items, in the order they were asked.
items = ["Interest", "Clarity", "Pace", "Examples", "Feedback", "Overall"]
# The mean rating for that item, from 1 to 5.
means = [4.2, 3.8, 3.1, 4.5, 3.6, 4.0]

# Eight inches wide and three inches tall, with room reserved for the labels.
fig, ax = plt.subplots(figsize=(8, 3), layout="constrained")
# One point per item, joined so the order of the items is easy to follow.
ax.plot(items, means, marker="o", color="#1c1c1c")
# Turn the item names so they do not collide.
for label in ax.get_xticklabels():
    label.set_rotation(30)
    label.set_ha("right")
# Name what the horizontal axis lists.
ax.set_xlabel("Questionnaire item")
# Name the rating, including the scale.
ax.set_ylabel("Mean rating (1 to 5)")
ax.set_title("A wide figure")
# The width and the height that were requested, in inches.
print([float(value) for value in fig.get_size_inches()])
# Show the figure under this cell.
plt.show()
```

## The same ratings on a small page

A square page of 3.4 inches leaves less room for the item names. The first cell of this pair does not reserve space for the labels. The second uses `layout="constrained"`. The ratings do not change. This cell uses `plt` from the cell above.


```python
# A small page, without a layout pass. Long names can sit tight against the edge.
# The same items.
items = ["Interest", "Clarity", "Pace", "Examples", "Feedback", "Overall"]
# The same means.
means = [4.2, 3.8, 3.1, 4.5, 3.6, 4.0]

# A nearly square page. No layout argument, so the labels use the default margins.
fig, ax = plt.subplots(figsize=(3.4, 3.2))
ax.plot(items, means, marker="o", color="#1c1c1c")
for label in ax.get_xticklabels():
    label.set_rotation(30)
    label.set_ha("right")
ax.set_xlabel("Questionnaire item")
ax.set_ylabel("Mean rating (1 to 5)")
ax.set_title("No layout pass")
plt.show()
```

```python
# The same small page, with constrained layout reserving room for the labels.
items = ["Interest", "Clarity", "Pace", "Examples", "Feedback", "Overall"]
means = [4.2, 3.8, 3.1, 4.5, 3.6, 4.0]

# layout="constrained" moves the axes inwards so the text stays on the page.
fig, ax = plt.subplots(figsize=(3.4, 3.2), layout="constrained")
ax.plot(items, means, marker="o", color="#1c1c1c")
for label in ax.get_xticklabels():
    label.set_rotation(30)
    label.set_ha("right")
ax.set_xlabel("Questionnaire item")
ax.set_ylabel("Mean rating (1 to 5)")
ax.set_title("Room for the labels")
plt.show()
```
