# Scatterplots

How can a scatter show a group and a third measurement?

This notebook is in Matplotlib, in the module Common plots. Plotting and packages scattered one list, in one colour and one size. This notebook keeps the scatter and adds three things: a second group, a colour for each group, and a size for a third measurement.

Each point is a student. Horizontal position is revision hours. Vertical position is the test score. The colour is the seminar, morning or afternoon. The size is the number of practice sessions. Size here is the marker area, not the diameter: `s` is that area, and the cell multiplies the session count by 40 so the marks are large enough to see. The legend names the seminar. It does not explain the size.


## Two groups, colour, and size

The argument that sets the colour is spelled `color`. That is Matplotlib's spelling. The lesson text uses colour.

The first plot in a session can take a moment while the plotting library downloads.


```python
# Two seminars on one scatter. Colour is the group. Size is the session count.
import matplotlib.pyplot as plt

# Morning seminar: revision hours, test score, and practice sessions.
morning_hours = [2, 3, 4, 5, 6]
morning_score = [4, 5, 6, 7, 8]
morning_sessions = [1, 2, 2, 3, 4]

# Afternoon seminar, in the same three columns.
afternoon_hours = [2, 4, 5, 7, 8]
afternoon_score = [3, 5, 6, 6, 9]
afternoon_sessions = [1, 1, 3, 2, 4]

# One axes for both groups.
fig, ax = plt.subplots(figsize=(6.6, 4.0), layout="constrained")
# Morning points. s is the marker area. 40 times the session count.
# The argument is spelled color, which is Matplotlib's spelling.
ax.scatter(
    morning_hours,
    morning_score,
    s=[40 * count for count in morning_sessions],
    color="#1c1c1c",
    label="Morning",
)
# Afternoon points, in a lighter grey, with the same size rule.
ax.scatter(
    afternoon_hours,
    afternoon_score,
    s=[40 * count for count in afternoon_sessions],
    color="#8a8a8a",
    label="Afternoon",
)
ax.set_xlabel("Revision hours")
ax.set_ylabel("Test score")
ax.set_title("Revision, score, and practice sessions")
# The legend names the two seminars. It does not show what the sizes mean.
ax.legend(loc="upper left")
plt.show()
```

## Reading one point

The morning student with 6 revision hours scored 8, and that point uses 4 sessions, so its area is `40 * 4`. The afternoon student with 8 revision hours scored 9, and that point also uses 4 sessions. A larger mark is more sessions. It is not a higher score. The score is the height.

Five students in each seminar is a sketch of the table. It is not a result about a course.
