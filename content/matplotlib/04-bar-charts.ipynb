# Bar charts

When does a bar show a count honestly?

You will count the mini-test totals, draw those counts from zero and with the axis cut off, and then split the questionnaire item by seminar. By the end you should be able to say what a bar's length is, and what a vertical axis that starts above zero does to that length.

A bar is for a count, or for another amount that starts at zero. The length is that amount. Cutting the axis off changes how large a gap looks. It does not change the count.

The previous lesson used one point per student. A count collapses those rows. The mini-test is the number correct on a four-question mini-test, so the totals run from 0 to 4. In this class they run from 1 to 4.


## From zero, and cut off

The twelve mini-test totals, in weekly quiz class order, are 1, 2, 2, 3, 3, 4, 2, 2, 3, 3, 3, and 4. One student scored 1, four scored 2, five scored 3, and two scored 4.

The left panel starts at zero. The right panel starts at 1. Both panels use the same four counts. The bar for 1 correct has height 1, so a view that starts at 1 leaves that bar with no visible length. The counts have not changed.

The first plot in a session can take a moment while the plotting library downloads.


```python
# The same four counts, drawn from zero and then with the axis cut off below 1.
import matplotlib.pyplot as plt

# Mini-test totals from 1 correct to 4 correct.
labels = ["1", "2", "3", "4"]
# How many students recorded that total. One, four, five, and two.
counts = [1, 4, 5, 2]

# Two panels, so the cut-off axis can sit beside the axis that includes zero.
fig, axes = plt.subplots(1, 2, figsize=(8, 3.4), layout="constrained")

# The left panel includes zero. The bar height is the count.
axes[0].bar(labels, counts, color="#4a4a4a")
axes[0].set_ylim(0, 6)
axes[0].set_xlabel("Mini-test, number correct")
axes[0].set_ylabel("Count")
axes[0].set_title("From zero")

# The right panel starts at 1. The bar of height 1 has no visible length.
axes[1].bar(labels, counts, color="#4a4a4a")
axes[1].set_ylim(1, 6)
axes[1].set_xlabel("Mini-test, number correct")
axes[1].set_ylabel("Count")
axes[1].set_title("Cut off below 1")

# The total is the sum of the four counts, which is the class size.
print(sum(counts))
plt.show()
```

## Longer labels

`barh` draws the bars from a vertical list, which leaves room for a longer label. These are still the same four counts, and the horizontal axis still starts at zero. Twelve students answered. That is this class, not a survey result. This cell uses `plt` from the cell above.


```python
# Horizontal bars for the same counts. The length is still the count.
labels = ["1 correct", "2 correct", "3 correct", "4 correct"]
counts = [1, 4, 5, 2]

fig, ax = plt.subplots(figsize=(6.4, 3.2), layout="constrained")
# barh takes the labels on the vertical axis and the counts as lengths.
ax.barh(labels, counts, color="#4a4a4a")
# Zero stays at the left, so the length is the count.
ax.set_xlim(0, 6)
ax.set_xlabel("Count")
ax.set_ylabel("Mini-test")
ax.set_title("Number correct on the mini-test")
plt.show()
```

## The same item, split by seminar

`confident` is one questionnaire item, yes or no. Monday has 4 yes and 2 no. Thursday has 5 yes and 1 no. Grouped bars still start at zero. The length is a count of students. A gap between Monday and Thursday is a gap in those counts, readable because both bars share a baseline at zero. This cell uses `plt` from the cell above.


```python
# Yes and No, split by seminar. The length is still a count of students, drawn from zero.
labels = ["Yes", "No"]
# Monday: four yes, two no.
monday = [4, 2]
# Thursday: five yes, one no.
thursday = [5, 1]
positions = [0, 1]

fig, ax = plt.subplots(figsize=(6.4, 3.4), layout="constrained")
# Monday bars sit just left of each tick. Thursday bars sit just right.
ax.bar([p - 0.18 for p in positions], monday, width=0.36, color="#1c1c1c", label="Monday")
ax.bar([p + 0.18 for p in positions], thursday, width=0.36, color="#8a8a8a", label="Thursday")
ax.set_xticks(positions)
ax.set_xticklabels(labels)
ax.set_ylim(0, 6)
ax.set_xlabel("Response")
ax.set_ylabel("Count")
ax.set_title("Confident, by seminar")
ax.legend(title="Seminar")
plt.show()
```
