# Subplots

How do two related plots share one figure?

You will put the Monday quiz and the Monday revision hours side by side, stack those same two series, and then build a grid of four panels for both seminars. By the end you should be able to pick a panel from the axes array, and say when that index needs two numbers.

The quiz and the hours are different units. Two regions keep those units apart. A shared vertical scale would make an hour look like a quiz point.

The previous lesson had one axes on the page. One region is enough for one series. These panels are for two series from the Monday rows, and then for both seminars in the weekly quiz class.


## Two panels, side by side

`plt.subplots(1, 2)` builds one page and two axes. The call is one row and two columns. Matplotlib returns a flat array of length 2 when there is only one row, so `axes[0]` is the left panel and `axes[1]` is the right panel. `sharex=True` uses the same six row numbers on both panels. The vertical scales stay separate.

Monday weekly quiz: 4, 5, 6, 6, 7, 8. Monday revision hours: 1, 2, 3, 3, 4, 5. The cell prints the shape of the axes array.

The first plot in a session can take a moment while the plotting library downloads.


```python
# One figure, two axes. The quiz and the revision hours stay on their own scales.
import matplotlib.pyplot as plt

# Monday rows, in weekly quiz class order.
rows = [1, 2, 3, 4, 5, 6]
# The weekly quiz on those rows.
weekly_quiz = [4, 5, 6, 6, 7, 8]
# Revision hours on the same rows. Not the same unit as the quiz.
revision_hours = [1, 2, 3, 3, 4, 5]

# One row, two columns. sharex lines the row numbers up. The vertical scales stay separate.
fig, axes = plt.subplots(1, 2, figsize=(8, 3.2), layout="constrained", sharex=True)
# The shape is (2,): one row is returned as a flat list of axes.
print(axes.shape)

# The left axes. The score axis starts at zero.
axes[0].plot(rows, weekly_quiz, marker="o", color="#1c1c1c")
axes[0].set_ylabel("Weekly quiz")
axes[0].set_title("Monday quiz")
axes[0].set_ylim(0, 10)

# The right axes. Its vertical label is hours, not score.
axes[1].plot(rows, revision_hours, marker="o", color="#1c1c1c")
axes[1].set_ylabel("Revision hours")
axes[1].set_title("Monday revision")
axes[1].set_ylim(0, 6)
# The shared horizontal label sits on the figure, once, for both panels.
fig.supxlabel("Monday row")

# Show both panels under this cell.
plt.show()
```

## The same two series, stacked

`plt.subplots(2, 1)` stacks the panels. Again there is only one column, so the array is flat: `axes[0]` is the top panel and `axes[1]` is the bottom panel. The units are still separate. This cell uses `plt` from the cell above.


```python
# The same two Monday series, stacked. Top is the quiz. Bottom is the revision hours.
rows = [1, 2, 3, 4, 5, 6]
weekly_quiz = [4, 5, 6, 6, 7, 8]
revision_hours = [1, 2, 3, 3, 4, 5]

# Two rows, one column. sharex keeps one set of row numbers.
fig, axes = plt.subplots(2, 1, figsize=(6.2, 4.6), layout="constrained", sharex=True)
# Still a flat list, because there is only one column.
print(axes.shape)

axes[0].plot(rows, weekly_quiz, marker="o", color="#1c1c1c")
axes[0].set_ylabel("Weekly quiz")
axes[0].set_title("Monday quiz")
axes[0].set_ylim(0, 10)

axes[1].plot(rows, revision_hours, marker="o", color="#1c1c1c")
axes[1].set_xlabel("Monday row")
axes[1].set_ylabel("Revision hours")
axes[1].set_title("Monday revision")
axes[1].set_ylim(0, 6)

plt.show()
```

## A grid of four

`plt.subplots(2, 2)` returns a two-dimensional array. The shape is `(2, 2)`. `axes[0, 0]` is the top left, `axes[0, 1]` the top right, `axes[1, 0]` the bottom left, and `axes[1, 1]` the bottom right. A single index is no longer enough.

The top row is Monday. The bottom row is Thursday. The left column is the weekly quiz. The right column is revision hours. Thursday's quiz includes the score recorded as 18, so that panel's vertical scale runs higher than Monday's. The hours panels still do not share a scale with the quiz panels. This cell uses `plt` from the cell above.


```python
# Four panels. Row one of the array is Monday. Row two is Thursday.
# A two-by-two call returns a grid, so each panel needs a row index and a column index.
rows = [1, 2, 3, 4, 5, 6]
monday_quiz = [4, 5, 6, 6, 7, 8]
monday_hours = [1, 2, 3, 3, 4, 5]
# Thursday, in weekly quiz class order. The last quiz was recorded as 18.
thursday_quiz = [4, 5, 6, 7, 8, 18]
thursday_hours = [2, 2, 3, 3, 4, 4]

fig, axes = plt.subplots(2, 2, figsize=(8, 6.2), layout="constrained")
# The shape is (2, 2), not a flat list.
print(axes.shape)

# Top left: Monday quiz.
axes[0, 0].plot(rows, monday_quiz, marker="o", color="#1c1c1c")
axes[0, 0].set_ylabel("Weekly quiz")
axes[0, 0].set_title("Monday quiz")
axes[0, 0].set_ylim(0, 10)

# Top right: Monday revision hours. A different unit, so its own vertical scale.
axes[0, 1].plot(rows, monday_hours, marker="o", color="#1c1c1c")
axes[0, 1].set_ylabel("Revision hours")
axes[0, 1].set_title("Monday revision")
axes[0, 1].set_ylim(0, 6)

# Bottom left: Thursday quiz, including the recorded 18, so the scale runs to 20.
axes[1, 0].plot(rows, thursday_quiz, marker="o", color="#1c1c1c")
axes[1, 0].set_xlabel("Row in the seminar")
axes[1, 0].set_ylabel("Weekly quiz")
axes[1, 0].set_title("Thursday quiz")
axes[1, 0].set_ylim(0, 20)

# Bottom right: Thursday revision hours.
axes[1, 1].plot(rows, thursday_hours, marker="o", color="#1c1c1c")
axes[1, 1].set_xlabel("Row in the seminar")
axes[1, 1].set_ylabel("Revision hours")
axes[1, 1].set_title("Thursday revision")
axes[1, 1].set_ylim(0, 6)

plt.show()
```
