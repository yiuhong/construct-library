# A straight-line fit

How can I fit a straight line through a few points?

scipy can fit a straight line through a few points. The line has a slope and an intercept. The slope is how much y changes when x increases by 1. The intercept is the line's y when x is 0.

The first time this session imports scipy, that library downloads. That can take a moment.


## Slope and intercept

`linregress` fits the line. `fit.slope` and `fit.intercept` are the two numbers. On the points below, the slope is 2 and the intercept is -1. At x = 1 the line is 2 * 1 - 1, which is 1, the first y value.


```python
# The slope and the intercept of the straight line that fits these points. The slope is the rise per step in x, and the intercept is the height at x = 0.
# SciPy's stats tools include a straight-line fit.
from scipy import stats

# The inputs.
x = [1, 2, 3, 4]
# The outcomes, in the same order.
y = [1, 3, 5, 7]
# Fit one straight line. The result holds the slope and the intercept.
fit = stats.linregress(x, y)
# The fitted slope.
print(fit.slope)
# The fitted intercept.
print(fit.intercept)
```

## The line on the points

The points are drawn with `scatter`. The fitted line is drawn with `plot`. For each x, the line's y is the slope times x, plus the intercept.


```python
# The points, and the line's prediction at each of them. The prediction is the slope times x, plus the intercept.
import matplotlib.pyplot as plt
from scipy import stats

# The same inputs as the cell above.
x = [1, 2, 3, 4]
# The same outcomes.
y = [1, 3, 5, 7]
# Fit the line again, so this cell stands on its own.
fit = stats.linregress(x, y)
# One predicted height per input, filled by the loop.
y_line = []
# Take each input in turn.
for value in x:
    # The line's height at this input.
    y_line.append(fit.slope * value + fit.intercept)

# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The observed points, not joined.
ax.scatter(x, y)
# The fitted line through those inputs.
ax.plot(x, y_line)
# Name the horizontal axis.
ax.set_xlabel("x")
# Name the vertical axis.
ax.set_ylabel("y")
# A title.
ax.set_title("Fitted line")
# Show the figure under this cell.
plt.show()
```
