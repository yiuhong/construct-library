# A pandas table and a plot

How can I store columns under names, and plot two of them?

You will build a table of the six Monday rows, sum and average the weekly quiz, and plot revision hours against that quiz. By the end you should be able to select a column by its name and use two columns as the axes.

A table stores columns under names. The columns have the same length. pandas calls this table a DataFrame. Naming the columns means the plot can ask for revision hours and the weekly quiz instead of remembering which list was which.

The previous lesson kept those two measurements as two lists, and joined them. A table is the same Monday pairs, with a name on each column, plus the seminar name so this slice stays visible.


## Build a table

Pass a dictionary to `DataFrame`. Each key is a column name. The six rows are the Monday seminar, in weekly quiz class order. Revision hours are 1, 2, 3, 3, 4, and 5. The weekly quiz is 4, 5, 6, 6, 7, and 8. The seminar column is the same word six times. It is there so the table says which slice this is. The plot in the last cell does not use it.

The numbers on the left of the printed table are row labels. They are not a column you added. The first time this session imports pandas, that library downloads. That can take a moment.


```python
# One row per Monday student. The lists become columns of one table.
# pandas builds the table.
import pandas as pd

# Six Monday rows. The seminar column names the slice. The other two columns are the pairs from the previous lesson.
table = pd.DataFrame({
    "seminar": ["Monday", "Monday", "Monday", "Monday", "Monday", "Monday"],
    "revision_hours": [1, 2, 3, 3, 4, 5],
    "weekly_quiz": [4, 5, 6, 6, 7, 8],
})
# How many students are in the table.
print(len(table))
# Print the table, row by row.
print(table)
```

## Sum and mean of one column

`table["weekly_quiz"]` selects the quiz column. The sum is every score added. The six scores are 4, 5, 6, 6, 7, and 8, which add to 36. The mean is the sum divided by how many scores there are, so `36 / 6` is 6. Run the table cell above first if you have not.


```python
# The total of the weekly quiz, and the mean of that column. The table comes from the cell above, so run that cell first.
# Add the weekly quiz column. The six scores add to 36.
print(table["weekly_quiz"].sum())
# The mean of that same column. 36 divided by 6 is 6.
print(table["weekly_quiz"].mean())
```

## Plot two columns

Revision hours are the horizontal axis. The weekly quiz is the vertical axis. The seminar column stays in the table and off the axes. The vertical axis starts at zero, so the height is the score. Run the table cell above first if you have not.


```python
# Weekly quiz against revision hours, using the table from the cell above. Each row becomes one point.
# Matplotlib draws the line.
import matplotlib.pyplot as plt

# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# Revision hours on the horizontal axis, weekly quiz on the vertical, one marker per row.
ax.plot(table["revision_hours"], table["weekly_quiz"], marker="o")
# Name the horizontal axis.
ax.set_xlabel("Revision hours")
# Name the vertical axis.
ax.set_ylabel("Weekly quiz")
ax.set_ylim(0, 10)
# A title that names the Monday slice.
ax.set_title("Monday revision and quiz")
# Show the figure under this cell.
plt.show()
```
