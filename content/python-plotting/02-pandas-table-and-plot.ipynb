# A pandas table and a plot

How can I store columns under names, and plot two of them?

A table stores columns under names. The columns have the same length. pandas calls this table a DataFrame. This lesson builds a small one and plots two of its columns.


## Build a table

Pass a dictionary to `DataFrame`. Each key is a column name. The first time this session imports pandas, that library downloads. That can take a moment.


```python
# One row per session, with that session's score beside it. The two lists become two columns of one table.
# pandas builds the table.
import pandas as pd

# A table with a session column and a score column.
table = pd.DataFrame({
    "session": [1, 2, 3, 4],
    "score": [2, 5, 5, 8],
})
# Print the table, row by row.
print(table)
```

## Sum and mean of one column

The numbers on the left of the printed table are row labels. They are not a column you added.

`table["score"]` selects the score column. The sum is every score added. The mean is the sum divided by how many scores there are. The sum below is 20. The mean is 5.


```python
# The total of the score column, and the mean of that column. The table comes from the cell above, so run that cell first.
# Add the score column.
print(table["score"].sum())
# The mean of that same column.
print(table["score"].mean())
```

## Plot two columns

The session column is the horizontal axis. The score column is the vertical axis. Run the table cell above first if you have not.


```python
# Score against session, using the table from the cell above. Each row of the table becomes one point.
# Matplotlib draws the line.
import matplotlib.pyplot as plt

# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# Session on the horizontal axis, score on the vertical, one marker per row.
ax.plot(table["session"], table["score"], marker="o")
# Name the horizontal axis.
ax.set_xlabel("Session")
# Name the vertical axis.
ax.set_ylabel("Score")
# Tick marks only at the four sessions.
ax.set_xticks([1, 2, 3, 4])
# Show the figure under this cell.
plt.show()
```
