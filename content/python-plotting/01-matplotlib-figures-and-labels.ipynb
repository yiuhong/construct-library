# Matplotlib figures and labels

How can I draw points on labelled axes, and join them?

You will scatter revision hours against the weekly quiz for six Monday rows, join those points, and then grow the line one point at a time. By the end you should be able to make a labelled scatter and a line through the same pairs.

A figure is the page. An axes is the region you draw on. `plt.subplots` returns both. `fig` is the figure. `ax` is the axes. The labels belong to the axes. Without them, the page does not say what the numbers are.

Imports and plots drew one line of scores. These six rows are the Monday seminar of the weekly quiz class, in class order. Each row is one student. Revision hours are 1, 2, 3, 3, 4, and 5. The weekly quiz on those rows is 4, 5, 6, 6, 7, and 8. Later lessons in this short introduction keep this Monday slice.


## Points, with labels

The two lists are paired. The first revision time goes with the first quiz score. Two Monday students share 3 hours and a quiz of 6, so those two points sit on top of each other. `scatter` marks each pair. `set_xlabel`, `set_ylabel`, and `set_title` write the text on the axes. The vertical axis starts at zero, so the height is the score.

The first plot in a session can take a moment while the plotting library downloads.


```python
# One point per Monday student, with a title and names on the axes. The points are not joined.
# Matplotlib draws the figure.
import matplotlib.pyplot as plt

# Revision hours for the six Monday rows, in weekly quiz class order.
revision_hours = [1, 2, 3, 3, 4, 5]
# The weekly quiz on those rows, in the same order.
weekly_quiz = [4, 5, 6, 6, 7, 8]

# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# One point per pair. A scatter plot does not join the points.
ax.scatter(revision_hours, weekly_quiz)
# Name the horizontal axis, including what was measured.
ax.set_xlabel("Revision hours")
# Name the vertical axis. The quiz is marked out of 10.
ax.set_ylabel("Weekly quiz")
# Start at zero, so the height is the score.
ax.set_ylim(0, 10)
# A title for the figure.
ax.set_title("Monday revision and quiz")
# Show the figure under this cell.
plt.show()
```

## Join the points

`plot` joins the pairs in list order. The points and the line are the same pairs. From the second student to the third, revision hours stay at 3 and the quiz stays at 6, so that step of the line does not move. This cell uses `plt` from the cell above.


```python
# The same points, joined in Monday row order. The scatter marks each point, and the line connects them.
# The same revision hours as the cell above.
revision_hours = [1, 2, 3, 3, 4, 5]
# The same weekly quiz scores.
weekly_quiz = [4, 5, 6, 6, 7, 8]

# A new figure.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The points.
ax.scatter(revision_hours, weekly_quiz)
# A line through those points, in list order.
ax.plot(revision_hours, weekly_quiz)
# Name the horizontal axis.
ax.set_xlabel("Revision hours")
# Name the vertical axis.
ax.set_ylabel("Weekly quiz")
ax.set_ylim(0, 10)
# A title.
ax.set_title("Monday revision and quiz")
# Show the figure under this cell.
plt.show()
```

## A line that grows

Matplotlib can redraw the same axes, one frame at a time. Each frame adds one Monday row and extends the line. The axes stay fixed, from 0 to 6 hours and from 0 to 10 on the quiz, so the frame does not jump when a point appears.

Run the cell, then press play under the figure. The slider moves through the six frames. The player is HTML in the page. The step that repeats 3 hours and a quiz of 6 adds the second of the two students who share that point.

Redrawing the frames prints a repeated warning from this plotting build. The cell hides that warning so the player stays easy to see.


```python
# Reveal one Monday row at a time. Each frame of the player adds the next student.
import warnings
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML

# Hide a warning from this version of Matplotlib, so the player can start cleanly. The warning is about an old way of passing coordinates, and it does not change the figure.
warnings.filterwarnings("ignore", message="The x parameter as float was deprecated")
warnings.filterwarnings("ignore", message="The y parameter as float was deprecated")

# Revision hours, in Monday row order.
revision_hours = [1, 2, 3, 3, 4, 5]
# The matching weekly quiz scores.
weekly_quiz = [4, 5, 6, 6, 7, 8]

# The figure the player will draw into.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# Fix the horizontal range, so the axes do not jump as points appear.
ax.set_xlim(0, 6)
# Fix the vertical range as well. The quiz is marked out of 10.
ax.set_ylim(0, 10)
# Name the horizontal axis.
ax.set_xlabel("Revision hours")
# Name the vertical axis.
ax.set_ylabel("Weekly quiz")
# A title.
ax.set_title("One Monday row at a time")
# An empty line. The frames fill it in.
drawn = ax.plot([], [], marker="o")
# plot gives back a list. The line is the first item.
line = drawn[0]

# One frame of the animation. frame starts at 0.
def grow(frame):
    # How many rows to show, from the first up to this frame.
    n = frame + 1
    # Keep only the first n revision times and scores.
    line.set_data(revision_hours[:n], weekly_quiz[:n])

# Build a player under the figure. Press play, or drag the slider, to step through the frames.
player = animation.FuncAnimation(
    fig, grow, frames=len(revision_hours), interval=500, blit=False
)
page = player.to_jshtml()
# Close the static figure so it is not also shown above the player.
plt.close(fig)
# The player itself, which the notebook shows under the cell.
HTML(page)
```
