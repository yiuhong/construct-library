# Fair-coin expectation

A fair coin has two faces, heads and tails, each with probability 1/2.

Let the payoff be 1 for heads and 0 for tails. The expected payoff is the probability-weighted sum of those payoffs.


```python
p = 0.5
E = p * 1 + (1 - p) * 0
# E is the expected value of the payoff
print(E)
```
