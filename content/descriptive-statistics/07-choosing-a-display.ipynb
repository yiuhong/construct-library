# Choosing a display

Which display fits the measurements I have?

Match the display to the measurements, not the other way round.

- One category, such as yes or no: a bar of counts.
- One number, such as a score: a histogram of bins, or dots along a line.
- Two numbers recorded on the same rows: a scatter.

`display_for` returns that choice for three short names: `"category"`, `"number"`, and `"two numbers"`.


## A small rule, then one example of each

The function is a lookup. The figure draws a toy example of each choice so the three displays can be compared. The numbers in the figure are only there to show the layout. The first plot in a session can take a moment while the plotting library downloads.


```python
# Map a kind of measurement to a display, then show a small example of each display.
import matplotlib.pyplot as plt

def display_for(kind):
    # A category is drawn as a bar of counts.
    if kind == "category":
        return "bar"
    # A single number is drawn as a histogram.
    if kind == "number":
        return "histogram"
    # Two numbers on the same rows are drawn as a scatter.
    if kind == "two numbers":
        return "scatter"
    return "none"

print(display_for("category"))
print(display_for("number"))
print(display_for("two numbers"))

# Three panels, one display each.
fig, axes = plt.subplots(1, 3, figsize=(6.4, 3.2))
# Counts of two labels.
axes[0].bar(["yes", "no"], [3, 2])
# Name this panel.
axes[0].set_title("Bar")
# One number, already counted into bins.
axes[1].bar(["1 to 4", "4 to 7"], [4, 3])
# Name this panel.
axes[1].set_title("Histogram")
# Two numbers, one point per row.
axes[2].scatter([1, 2, 3, 4], [2, 3, 5, 6])
# Name this panel.
axes[2].set_title("Scatter")
# Give the three panels room.
fig.tight_layout()
# Show the figure under this cell.
plt.show()
```
