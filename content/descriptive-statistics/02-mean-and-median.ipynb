# Mean and median

When do the mean and the median tell different stories about the same list?

The mean is the sum divided by how many values there are. The median is the middle value once the list is sorted. For an odd-length list, that middle value is one of the measurements.

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

The list `[2, 3, 4, 5, 20]` sums to 34. Its mean is $34/5 = 6.8$. Sorted, the middle value is 4, so the median is 4. The 20 pulls the mean up. It does not change which value sits in the middle.


## One list, two centres

The cell prints both centres and draws the five values. A horizontal line marks each centre. The first plot in a session can take a moment while the plotting library downloads.


```python
# The mean uses every value. The median uses the middle value after sorting. The figure shows both centres.
import matplotlib.pyplot as plt

# Five measurements, one of them far from the others.
values = [2, 3, 4, 5, 20]
# How many measurements.
n = len(values)
# The mean is the sum divided by that count.
mean = sum(values) / n
# Sort a copy so the middle value can be read off.
ordered = sorted(values)
# The middle place in a list of five is place 2, counting from 0.
median = ordered[n // 2]
print(mean)
print(median)

# Place each value along the list, and mark the two centres.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# One point per measurement.
ax.plot([1, 2, 3, 4, 5], values, marker="o", linestyle="none", label="Values")
# A dashed line at the mean.
ax.axhline(mean, color="black", linestyle="--", label="Mean")
# A dotted line at the median.
ax.axhline(median, color="grey", linestyle=":", label="Median")
# The horizontal axis is the place in the list.
ax.set_xlabel("Place in the list")
# The vertical axis is the measurement.
ax.set_ylabel("Value")
# A key, so the points and the two centres can be told apart.
ax.legend()
# Show the figure under this cell.
plt.show()
```
