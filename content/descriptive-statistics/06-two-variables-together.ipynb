# Two variables together

How can I describe two measurements that were recorded on the same rows?

Each row has hours of practice and a score: hours `[1, 2, 3, 4]` and scores `[2, 3, 5, 6]`. A scatter puts one measurement on each axis and one point per row. Points that rise together are a positive association.

A single number for that pattern is the sum of the paired deviations. Subtract the mean of hours from each hour, subtract the mean of the scores from each score, multiply those two deviations, and add the products. The hours mean is 2.5 and the score mean is 4. The four products are 3, 0.5, 0.5, and 3, and they sum to 7. The sum is positive, which matches the rising scatter.

Two category columns would be a table of counts instead of a scatter. These two columns are both numbers, so the scatter is the display.


## Paired deviations, and the scatter

The loop multiplies one pair of deviations at a time. The figure uses the same four rows. The first plot in a session can take a moment while the plotting library downloads.


```python
# Sum the products of paired deviations, then plot one point per row.
import matplotlib.pyplot as plt

# Hours and scores, in the same row order.
hours = [1, 2, 3, 4]
scores = [2, 3, 5, 6]
# Each list has its own mean.
mean_hours = sum(hours) / len(hours)
mean_scores = sum(scores) / len(scores)

# A running total of the products of paired deviations.
total = 0
# One row at a time. The index picks the matching hour and score.
for i in range(len(hours)):
    # How far this hour sits from the hours mean.
    hour_distance = hours[i] - mean_hours
    # How far the score on the same row sits from the score mean.
    score_distance = scores[i] - mean_scores
    # Add the product. A positive product means the two deviations have the same sign.
    total = total + hour_distance * score_distance
print(mean_scores)
print(total)

# One point per row. Hours run along the horizontal axis.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter(hours, scores)
# Hours run along the horizontal axis.
ax.set_xlabel("Hours")
# Scores run along the vertical axis.
ax.set_ylabel("Score")
# Show the figure under this cell.
plt.show()
```
