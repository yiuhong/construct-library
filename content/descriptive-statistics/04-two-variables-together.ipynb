# Two variables together

How do two numbers on the same students sit together?

You will plot revision hours against the weekly quiz, sum the products of the paired deviations, and contrast that with a count of two categories. By the end you should be able to say when a scatter is the display, and when a table of counts is.

A scatter can show that two columns rise together. A histogram of either column on its own cannot.

The previous lesson displayed one variable, and then that variable split by seminar. Seminar is a category. Revision hours are a second number, paired with the quiz on the same rows.


## One point per student

The revision hours, in weekly quiz class row order, are 1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, and 4. The quiz scores on those same rows are 4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, and 18. A scatter puts hours on the horizontal axis and the quiz on the vertical axis, one point per student. The points tend to rise. The student recorded as 18 revised for 4 hours, so that point sits high rather than far to the right.

A scatter is the display when both columns are numbers on the same rows.


```python
# One point per student: revision hours along the horizontal axis, weekly quiz on the vertical.
import matplotlib.pyplot as plt

# The two columns, in the same row order as the weekly quiz class.
revision_hours = [1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, 4]
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]

# One point per row.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter(revision_hours, weekly_quiz)
# Hours run along the horizontal axis.
ax.set_xlabel("Revision hours")
# The quiz runs along the vertical axis.
ax.set_ylabel("Weekly quiz")
# Show the figure under this cell.
plt.show()
```

## Paired deviations

A single number for that rising pattern is the sum of the paired deviations. Subtract the mean of the hours from each hour, subtract the mean of the quiz from each score, multiply within the row, and add the products. The hours mean is 3 and the quiz mean is 7. The twelve products are 6, 2, 0, 0, 0, 2, 3, 2, 0, 0, 1, and 11, and they sum to 27. The sum is positive, which matches the points rising together. The last product, 11, belongs to the recorded 18. That one row is a large part of the sum, not the whole of it.


```python
# Sum the products of paired deviations, one student at a time.
# Hours and the quiz, in the same row order.
revision_hours = [1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, 4]
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
# Each column has its own mean.
mean_hours = sum(revision_hours) / len(revision_hours)
mean_quiz = sum(weekly_quiz) / len(weekly_quiz)

# A running total of the products of paired deviations.
total = 0
# One row at a time. The index picks the matching hour and score.
for i in range(len(revision_hours)):
    # How far this hour sits from the hours mean.
    hour_distance = revision_hours[i] - mean_hours
    # How far the quiz on the same row sits from the quiz mean.
    quiz_distance = weekly_quiz[i] - mean_quiz
    # Add the product. A positive product means the two deviations have the same sign.
    total = total + hour_distance * quiz_distance
print(mean_hours)
print(mean_quiz)
print(total)
```

## When a scatter is the wrong display

Seminar and `confident` are both categories. Crossing them makes four counts, not a scatter of codes. In row order, Monday's answers are no, no, yes, yes, yes, yes: two no and four yes. Thursday's answers are no, yes, yes, yes, yes, yes: one no and five yes. Those counts are the display for two categories. The scatter was the right display for hours and the quiz because both columns were numbers.


```python
# Two categories become counts, not a scatter. Seminar crossed with confident.
seminar = [
    "Monday", "Monday", "Monday", "Monday", "Monday", "Monday",
    "Thursday", "Thursday", "Thursday", "Thursday", "Thursday", "Thursday",
]
confident = [
    "no", "no", "yes", "yes", "yes", "yes",
    "no", "yes", "yes", "yes", "yes", "yes",
]

# One counter for each pair of labels.
monday_yes = 0
monday_no = 0
thursday_yes = 0
thursday_no = 0
# One student at a time.
for i in range(len(seminar)):
    # Monday, split by the questionnaire answer.
    if seminar[i] == "Monday" and confident[i] == "yes":
        monday_yes = monday_yes + 1
    elif seminar[i] == "Monday" and confident[i] == "no":
        monday_no = monday_no + 1
    # Thursday, split the same way.
    elif seminar[i] == "Thursday" and confident[i] == "yes":
        thursday_yes = thursday_yes + 1
    else:
        thursday_no = thursday_no + 1
print(monday_yes)
print(monday_no)
print(thursday_yes)
print(thursday_no)
```
