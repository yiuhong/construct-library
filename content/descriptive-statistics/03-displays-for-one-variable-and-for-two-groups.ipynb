# Displays for one variable and for two groups

Which display fits one measurement, and the same measurement in two seminars?

You will draw a bar for a category, a histogram for the weekly quiz, and one dot per student in Monday and in Thursday. By the end you should be able to pick among those pictures, and to say what a wider bin hides.

A bar counts labels. A histogram counts a number after it has been cut into bins. Dots keep every score when the class is split.

The previous lesson reduced the quiz to a centre and a spread. Those summaries do not show the shape of the scores, or that the two seminars are not the same list.


## A bar for a category

`confident` is yes or no, the same column as in the class table. Nine students answered yes and three answered no. A bar is the display for that kind of column: one bar per answer, and the height is the count. A histogram would be the wrong picture, because yes and no are not positions on a number line.


```python
# A bar of the confident counts. A category is drawn as bars, not as a histogram.
import matplotlib.pyplot as plt

# The confident column, in weekly quiz class row order.
confident = [
    "no", "no", "yes", "yes", "yes", "yes",
    "no", "yes", "yes", "yes", "yes", "yes",
]
# One counter per answer.
yes_count = 0
no_count = 0
for answer in confident:
    # A yes adds to the yes count.
    if answer == "yes":
        yes_count = yes_count + 1
    else:
        no_count = no_count + 1
print(yes_count)
print(no_count)

# One bar per answer. The height is the count.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.bar(["yes", "no"], [yes_count, no_count])
# The horizontal axis names the answers.
ax.set_xlabel("Confident")
# The vertical axis is the count.
ax.set_ylabel("Count")
# Show the figure under this cell.
plt.show()
```

## A histogram for one number

The weekly quiz is one number, so a histogram is the display. The bins have width 2 and start at 4. A score belongs to a bin from the left edge up to, but not including, the right edge.

- 4 up to but not including 6 holds both 4s and both 5s: 4 scores.
- 6 up to but not including 8 holds the three 6s and both 7s: 5 scores.
- 8 up to but not including 10 holds both 8s: 2 scores.
- 10 up to but not including 20 holds the recorded 18: 1 score.

The height is a count. The bin width is a choice about how to group the same twelve scores.


```python
# Place each weekly quiz score in a bin of width 2, then draw one bar per bin.
import matplotlib.pyplot as plt

# The weekly_quiz column, in weekly quiz class row order.
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
# One counter per bin: [4, 6), [6, 8), [8, 10), and [10, 20).
counts = [0, 0, 0, 0]
# One score at a time.
for score in weekly_quiz:
    # The first bin stops before 6.
    if score < 6:
        counts[0] = counts[0] + 1
    # The second bin stops before 8.
    elif score < 8:
        counts[1] = counts[1] + 1
    # The third bin stops before 10.
    elif score < 10:
        counts[2] = counts[2] + 1
    # What remains is the recorded 18.
    else:
        counts[3] = counts[3] + 1
print(counts)

# The labels name the bins. The heights are the counts just computed.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.bar(["4 to 6", "6 to 8", "8 to 10", "10 to 20"], counts)
# The horizontal axis names the bins.
ax.set_xlabel("Bin")
# The vertical axis is how many scores fell in the bin.
ax.set_ylabel("Count")
# Show the figure under this cell.
plt.show()
```

## What a wider bin hides

Cut the same scores once, at 10. Eleven scores fall below 10, and the recorded 18 is the only score at 10 or above. That picture is still a histogram, and it hides the pile between 4 and 8 that the narrower bins showed. The bin width changed. The quiz did not.


```python
# The same weekly quiz, in two wide bins: below 10, and 10 or above.
import matplotlib.pyplot as plt

# The weekly_quiz column again.
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
# One counter for below 10, and one for the rest.
below = 0
above = 0
for score in weekly_quiz:
    # The wide bin stops before 10.
    if score < 10:
        below = below + 1
    else:
        above = above + 1
print(below)
print(above)

# Two bars. The first one swallows the pile the narrower bins had separated.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.bar(["Below 10", "10 and above"], [below, above])
ax.set_xlabel("Bin")
ax.set_ylabel("Count")
plt.show()
```

## The two seminars

Monday's six scores are the first six in the column: 4, 5, 6, 6, 7, and 8. Their mean is 6. Thursday's six scores are 4, 5, 6, 7, 8, and 18. Their mean is 8. Thursday minus Monday is 2. The dots keep every score, so the gap is visible as the recorded 18 rather than as a shift of the whole Thursday seminar. The horizontal position is the seminar. The small offsets are only there so dots in one seminar do not sit on top of one another. They are not a second measurement.


```python
# A mean for each seminar, then one dot per weekly quiz score.
import matplotlib.pyplot as plt

# Monday, then Thursday. These are the two halves of the weekly_quiz column.
monday = [4, 5, 6, 6, 7, 8]
thursday = [4, 5, 6, 7, 8, 18]
# Each mean is the seminar's sum divided by how many scores it has.
monday_mean = sum(monday) / len(monday)
thursday_mean = sum(thursday) / len(thursday)
# The gap, Thursday minus Monday.
gap = thursday_mean - monday_mean
print(monday_mean)
print(thursday_mean)
print(gap)

# Monday sits near 1. Thursday sits near 2. The small offsets stop the dots overlapping.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# Monday dots, slightly offset so they do not sit on one point.
ax.scatter([0.85, 0.95, 1.05, 1.15, 1.25, 1.35], monday, label="Monday")
# Thursday dots, offset in the same way.
ax.scatter([1.85, 1.95, 2.05, 2.15, 2.25, 2.35], thursday, label="Thursday")
# A diamond at each seminar mean.
ax.scatter([1, 2], [monday_mean, thursday_mean], marker="D", color="black", label="Seminar mean")
# Show seminar names instead of 1 and 2.
ax.set_xticks([1, 2], ["Monday", "Thursday"])
# The horizontal axis is the seminar.
ax.set_xlabel("Seminar")
# The vertical axis is the quiz.
ax.set_ylabel("Weekly quiz")
# A key for the dots and the means.
ax.legend()
# Show the figure under this cell.
plt.show()
```
