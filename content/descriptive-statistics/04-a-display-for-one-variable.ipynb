# A display for one variable

What does a bin count show that the raw list does not?

A histogram groups a numerical measurement into bins and counts how many values fall in each bin. The height is a count. The width of the bin is a choice. A wide bin can hide a pile that a narrower bin would show.

The scores are `[1, 2, 2, 3, 5, 5, 5, 8]`. Split them into bins of width 3, starting at 1: from 1 up to but not including 4, from 4 up to but not including 7, and from 7 up to but not including 10. The first bin holds 1, 2, 2, and 3, so its count is 4. The second holds the three 5s, so its count is 3. The last holds 8, so its count is 1.


## Count the bins, then draw them

The loop places each score in one bin. The bar chart is the histogram: one bar per bin, and the bar height is the count. The first plot in a session can take a moment while the plotting library downloads.


```python
# Place each score in a bin of width 3, then draw one bar per bin.
import matplotlib.pyplot as plt

# Eight scores on one measurement.
scores = [1, 2, 2, 3, 5, 5, 5, 8]
# One counter per bin: [1, 4), [4, 7), and [7, 10).
counts = [0, 0, 0]
# One score at a time.
for score in scores:
    # The first bin stops before 4.
    if score < 4:
        counts[0] = counts[0] + 1
    # The second bin stops before 7.
    elif score < 7:
        counts[1] = counts[1] + 1
    # What remains is the last bin.
    else:
        counts[2] = counts[2] + 1
# Print the three bin counts.
print(counts)

# The labels name the bins. The heights are the counts just computed.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.bar(["1 to 4", "4 to 7", "7 to 10"], counts)
# The horizontal axis names the bins.
ax.set_xlabel("Bin")
# The vertical axis is how many scores fell in the bin.
ax.set_ylabel("Count")
# Show the figure under this cell.
plt.show()
```
