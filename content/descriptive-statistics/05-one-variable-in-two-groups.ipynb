# One variable in two groups

How can I compare one measurement across two groups?

Each group has its own list and its own mean. Quiet is `[3, 5, 7]`, so its mean is 5. Noisy is `[6, 8, 10]`, so its mean is 8. The gap is the noisy mean minus the quiet mean, which is 3.

A useful display keeps the group on the horizontal axis and the measurement on the vertical axis, with one dot per person. The dots show the values the means were calculated from.


## Two means, and the dots they came from

The offsets on the horizontal axis are only so that dots in the same group do not sit on top of one another. They are not a second measurement. The first plot in a session can take a moment while the plotting library downloads.


```python
# A mean for each group, then one dot per value, grouped along the horizontal axis.
import matplotlib.pyplot as plt

# The quiet group, then the noisy group.
quiet = [3, 5, 7]
noisy = [6, 8, 10]
# Each mean is the group's sum divided by how many values it has.
quiet_mean = sum(quiet) / len(quiet)
noisy_mean = sum(noisy) / len(noisy)
# The gap, noisy minus quiet.
gap = noisy_mean - quiet_mean
print(quiet_mean)
print(noisy_mean)
print(gap)

# Quiet sits near 1. Noisy sits near 2. The small offsets stop the dots overlapping.
# A figure for the two groups.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# Quiet dots, slightly offset so they do not sit on one point.
ax.scatter([0.9, 1.0, 1.1], quiet, label="Quiet")
# Noisy dots, offset in the same way.
ax.scatter([1.9, 2.0, 2.1], noisy, label="Noisy")
# A diamond at each group mean.
ax.scatter([1, 2], [quiet_mean, noisy_mean], marker="D", color="black", label="Group mean")
# Show group names instead of 1 and 2.
ax.set_xticks([1, 2], ["Quiet", "Noisy"])
# The horizontal axis is the group.
ax.set_xlabel("Group")
# The vertical axis is the measurement.
ax.set_ylabel("Measurement")
# A key for the dots and the means.
ax.legend()
# Show the figure under this cell.
plt.show()
```
