# A table of measurements

How do I turn one class into counts and a proportion?

You will put twelve students in one table, count a yes or no, and draw those counts as a bar. By the end you should be able to read a count and the share of the class it represents.

A mean or a plot later in the course is only as clear as the rows it came from. A proportion says how common an answer was, not only how many times it appeared.

Earlier lessons worked from short lists that were not a class. This table is the weekly quiz class, and the same rows are the ones the next lessons summarise.


## The weekly quiz class

Twelve students sit in two seminars, Monday and Thursday, six in each. Each row is one student. The weekly quiz is marked out of 10, and one Thursday score was recorded as 18, above that maximum. Later lessons keep that score in the column. `revision_hours` is hours spent revising. `practice` is how many practice sessions the student attended. `confident` is one questionnaire item, yes or no. `mini_test` is the number correct on a four-question mini-test, so it runs from 0 to 4. The numbers were chosen so the arithmetic in this course stays small.

The first time this session imports pandas or matplotlib, that library downloads. That can take a moment.


```python
# Build the weekly quiz class: one row per student, the same columns throughout this course.
import pandas as pd

# Twelve students. Row order is fixed, so later lessons can take a slice of the same rows.
weekly_quiz_class = pd.DataFrame({
    "seminar": [
        "Monday", "Monday", "Monday", "Monday", "Monday", "Monday",
        "Thursday", "Thursday", "Thursday", "Thursday", "Thursday", "Thursday",
    ],
    "weekly_quiz": [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18],
    "revision_hours": [1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, 4],
    "practice": [1, 1, 2, 2, 3, 3, 1, 2, 2, 2, 3, 4],
    "confident": [
        "no", "no", "yes", "yes", "yes", "yes",
        "no", "yes", "yes", "yes", "yes", "yes",
    ],
    "mini_test": [1, 2, 2, 3, 3, 4, 2, 2, 3, 3, 3, 4],
})
# How many students were recorded.
print(len(weekly_quiz_class))
# The table itself.
print(weekly_quiz_class)
```

## Count one answer

`confident` is a category. A count is how many times an answer appears, and a proportion is that count divided by how many students there are. Nine students answered yes, so the proportion of yes is $9/12 = 0.75$.


```python
# Count yes and no, then divide each count by the number of students.
confident = [
    "no", "no", "yes", "yes", "yes", "yes",
    "no", "yes", "yes", "yes", "yes", "yes",
]

# Start both counts at zero. The loop adds one for each answer.
yes_count = 0
no_count = 0
# One student at a time.
for answer in confident:
    # A yes adds to the yes count. The other answer in this column is no.
    if answer == "yes":
        yes_count = yes_count + 1
    else:
        no_count = no_count + 1

# How many students were recorded.
n = len(confident)
# A proportion is the count divided by that length.
yes_proportion = yes_count / n
no_proportion = no_count / n
print(yes_count)
print(yes_proportion)
print(no_count)
print(no_proportion)
```

## A bar for those counts

A bar chart is a direct picture of a category. The height of a bar is the count. Yes and no are labels, not positions on the quiz's number line, so a bar fits this column and a histogram does not.


```python
# Draw one bar per answer. The height is the count of that answer.
import matplotlib.pyplot as plt

# The two answers and their counts.
answers = ["yes", "no"]
counts = [9, 3]

# The bar height is the count. Categories sit on the horizontal axis.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# One bar per answer.
ax.bar(answers, counts)
# The horizontal axis names the answers.
ax.set_xlabel("Confident")
# The vertical axis is the count.
ax.set_ylabel("Count")
# Show the figure under this cell.
plt.show()
```
