# Predicting a category

How does a rule write a category for a new row?

This notebook is in Introduction to machine learning, in the module Supervised learning. The label is no longer a number. It is a word: pass or revise. The feature is a quiz score.

| quiz | label |
| --- | --- |
| 2 | revise |
| 3 | revise |
| 6 | pass |
| 8 | pass |

A cutoff of 5 writes pass when the quiz is at least 5, and revise otherwise. A new quiz of 5 is pass. A new quiz of 4 is revise. On these four rows the cutoff agrees with every label, because the rows were chosen to fall cleanly on either side.

Introductory statistics already used a logit to talk about the probability of a yes-or-no outcome. This notebook does not derive that link again. A cutoff keeps the category readable on four rows. A logistic fit on rows that separate this cleanly can push the slope very large, and that numerical detail is left aside.

Four rows, one feature, and a cutoff chosen by looking at these labels are not a placement rule. The cutoff was set with the labels in view, so it is not a test of a rule chosen in advance.


## Apply the cutoff

The cell writes a label for each training quiz, then for a quiz of 5 and a quiz of 4.


```python
# A quiz on one side of the cutoff or the other. The label is a word, not a score.
quizzes = [2, 3, 6, 8]
labels = ["revise", "revise", "pass", "pass"]
cutoff = 5

def label_for(quiz):
    # Pass if the quiz reaches the cutoff. Otherwise revise.
    if quiz >= cutoff:
        return "pass"
    return "revise"

# The label written for each training quiz.
written = []
for quiz in quizzes:
    written.append(label_for(quiz))
print(written)
# A new quiz on the cutoff, then a new quiz just below it.
print(label_for(5))
print(label_for(4))
```
