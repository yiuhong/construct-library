# Overfitting and generalisation

When can a model look perfect on the rows it trained on and still miss a new row?

This notebook is in Introduction to machine learning, in the module Learning from data. A training error uses the rows that set the model. Generalisation is about a row those rows did not include. A training error of zero does not answer that second question.

Three training scores, and one new row:

| hours | score | role |
| --- | --- | --- |
| 1 | 2 | training |
| 2 | 9 | training |
| 3 | 4 | training |
| 4 | 8 | new row |

A straight line cannot pass through all three training scores. A quadratic, hours and hours squared, has enough room with three points to pass through each of them. Its training error is then zero. The new row was not one of those three points.

Three training rows, one feature, and one new row are a sketch. The quadratic is not a claim about practice, and a prediction of a score far from the new row is not a result to report. A fuller check of generalisation uses more rows than parameters, and a split fixed before the new score is seen. Importing scikit-learn downloads it the first time. The first plot in a session can take a moment while the plotting library downloads.


## A line, and a curve that hits every training row

The cell fits both models on the three training rows and then names a score at 4 hours. The figure draws the bend only across the training hours. Its prediction at 4 hours is printed, because that number falls far outside the scores on the page.


```python
# Three training rows, and one new row that neither fit is allowed to see.
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures

train_hours = np.array([[1.0], [2.0], [3.0]])
train_scores = np.array([2.0, 9.0, 4.0])
new_hours = np.array([[4.0]])
new_score = 8.0

# A straight line. It will not pass through all three training scores.
line = LinearRegression()
line.fit(train_hours, train_scores)
line_train = line.predict(train_hours)
line_new = float(line.predict(new_hours)[0])
line_train_error = float(np.mean(np.abs(train_scores - line_train)))
line_new_error = abs(new_score - line_new)

# Hours and hours squared. With three points this can pass through every training score.
bend = make_pipeline(PolynomialFeatures(degree=2, include_bias=False), LinearRegression())
bend.fit(train_hours, train_scores)
bend_train = bend.predict(train_hours)
bend_new = float(bend.predict(new_hours)[0])
bend_train_error = float(np.mean(np.abs(train_scores - bend_train)))
bend_new_error = abs(new_score - bend_new)

print(round(line_train_error, 3))
print(round(line_new, 3))
print(round(line_new_error, 3))
print(round(bend_train_error, 3))
print(round(bend_new, 3))
print(round(bend_new_error, 3))

# Training rows, the line out to the new row, and the bend only where it was fitted.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter(train_hours.ravel(), train_scores, label="Training rows")
grid = np.linspace(1, 3, 40).reshape(-1, 1)
ax.plot(grid.ravel(), bend.predict(grid), label="Bend on the training rows")
ax.plot([1, 2, 3, 4], line.predict(np.array([[1.0], [2.0], [3.0], [4.0]])), label="Line")
ax.scatter([4], [new_score], marker="x", label="New row")
ax.set_xlabel("Hours")
ax.set_ylabel("Score")
ax.legend()
plt.show()
```

## Which error is which

The line's mean absolute error on the training rows is about 2.667. At 4 hours it names 7, and the new score is 8, so the absolute error on that row is 1.

The bend's training error is 0, because it passes through 2, 9, and 4. At 4 hours it names $-13$. The absolute error against 8 is 21. The training score looked better. The new row did not.

The figure stops the bend at 3 hours so the training rows stay readable. The printed $-13$ is the prediction that did not fit on that figure.
