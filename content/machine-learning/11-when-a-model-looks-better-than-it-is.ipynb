# When a model looks better than it is

How can a flattering number fail to be a test?

This notebook is in Introduction to machine learning, in the module Unsupervised learning and limits. Two small traps use the same idea. A score can look strong because it ignored the hard part, or because the row that was supposed to test the model was used to choose it.

Five labels, and no use of a quiz: pass, pass, pass, revise, pass. A rule that always writes pass is right on 4 of the 5 rows. That count never looked at a feature. On a table that is mostly pass, accuracy can look high while the rule is only repeating the common label.

A second trap uses one held-out score of 9, after a line fitted on hours 0, 1, and 2 with scores 1, 3, and 5. The line names 7, so the absolute error is 2. The mean of those three training scores is 3, so the absolute error of always predicting that mean is 6. The line looks better. Choosing it because the held-out error is smaller uses the held-out score to make the choice. The 2 is then not a clean test error. It is the error of a model that won on the same row being reported.

A third, shorter leak sits in the preparation. Scaling a column with a mean that includes the test row lets that row help set the scale. The features notebook did that on purpose, as a look at the preparation. Once a row is held out, the scale has to be set on the training rows only.

Five labels, three training rows, and one held-out score are illustrations of the trap. They are not a model comparison. Importing scikit-learn downloads it the first time.


## A common label, then a choice made on the test row

The cell counts how many labels an always-pass rule gets right, then fits the line and compares it with the training mean on the held-out score.


```python
# Two ways a number can look better than the test it claims to be.
import numpy as np
from sklearn.linear_model import LinearRegression

# Five labels. The rule ignores any quiz and always writes pass.
labels = ["pass", "pass", "pass", "revise", "pass"]
always_pass_correct = 0
for label in labels:
    if label == "pass":
        always_pass_correct = always_pass_correct + 1
print(always_pass_correct)
print(len(labels))

# Three training rows set a line and a mean. The held-out score is used only at the end.
train_hours = np.array([[0.0], [1.0], [2.0]])
train_scores = np.array([1.0, 3.0, 5.0])
held_out_score = 9.0
line = LinearRegression()
line.fit(train_hours, train_scores)
line_prediction = float(line.predict(np.array([[3.0]]))[0])
mean_prediction = float(np.mean(train_scores))
line_error = abs(held_out_score - line_prediction)
mean_error = abs(held_out_score - mean_prediction)
print(round(line_prediction, 3))
print(round(mean_prediction, 3))
print(round(line_error, 3))
print(round(mean_error, 3))
```

## What the smaller error is not

The line's absolute error is 2 and the mean's is 6. Reporting 2 because it is smaller spends the held-out row on the choice. A clean test would have chosen the line, or the mean, without that row, and only then scored the row that remained.
