# Prediction and explanation

What changes when the aim is a score for a new row, rather than an account of the rows already in hand?

This notebook opens Introduction to machine learning, in the module Learning from data. Introductory statistics already used a straight line to state a mean. The line here is the same kind of line. The question is different. The slope is not being read as an explanation. The line is being asked for a score it did not fit.

Three students sat a short test. Hours of practice and the score were chosen so that they sit on

$$
\text{score} = 1 + 2 \times \text{hours}
$$

| student | hours | score |
| --- | --- | --- |
| A | 0 | 1 |
| B | 1 | 3 |
| C | 2 | 5 |

A fourth student practised for 4 hours and is not in that table. The line names a score for that student. It was not asked to explain why the first three scores line up.

Three students and one new row are not a study of practice. One feature, hours, is doing all of the work. The three rows were built to sit on the line, so every residual in the table is zero. A residual of zero here is a choice of example, not a finding. A full prediction would use more rows, more than one feature, and a row the analyst had not already seen while drawing the line.


## A score the line did not fit

The cell evaluates the line at the three fitted hours, then at 4 hours. The figure shows the three students, the line, and the prediction for the new student. The first plot in a session can take a moment while the plotting library downloads.


```python
# The same line that can state a mean. Here it names a score for a student who was not used to draw it.
import matplotlib.pyplot as plt

# Three students. The scores were chosen to sit on score = 1 + 2 * hours.
hours = [0, 1, 2]
scores = [1, 3, 5]
intercept = 1
slope = 2

def predicted_score(hour):
    # The score the line names at this number of hours.
    return intercept + slope * hour

# A residual is the score minus the line. These three were built to sit on it.
residuals = []
for hour, score in zip(hours, scores):
    residuals.append(score - predicted_score(hour))
print(residuals)

# Four hours of practice. This student was not used to choose the intercept or the slope.
new_hours = 4
new_prediction = predicted_score(new_hours)
print(new_prediction)
print(slope)

# The fitted students, the line extended to the new student, and that prediction.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter(hours, scores, label="In the fit")
line_hours = [0, 1, 2, 3, 4]
ax.plot(line_hours, [predicted_score(hour) for hour in line_hours], label="Line")
ax.scatter([new_hours], [new_prediction], marker="x", label="New student")
ax.set_xlabel("Hours")
ax.set_ylabel("Score")
ax.legend()
plt.show()
```
