When do two test columns fail to add a second direction?

On this table the first component's share of the spread is 1, and the second share is 0. The two columns were typed as copies. A questionnaire would not do that. The share of 1 means the second column adds no new spread here. It does not mean a test is useless in general.

## The share of the spread

The cell fits two components on the copied columns and prints each component's share.


```python
# Two test columns that were typed as the same list. The second column adds no new spread.
import numpy as np
from sklearn.decomposition import PCA

vocabulary = [1.0, 2.0, 3.0, 4.0]
reading = [1.0, 2.0, 3.0, 4.0]
table = np.column_stack([vocabulary, reading])
components = PCA(n_components=2)
components.fit(table)
share = components.explained_variance_ratio_
# How many test columns went in.
print(table.shape[1])
# The share of spread on the first component, then the second.
print([round(float(value), 3) for value in share])
```
