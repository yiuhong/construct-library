# Predicting a number

How can two columns together name a score for a row that was held out?

This notebook is in Introduction to machine learning, in the module Supervised learning. The label is a number, a later test score. Supervised means the training rows already carry that label. The model is still a linear equation. Introductory statistics used one line to state a mean. Computational modelling used a line as a model with residuals. Here the equation has two inputs, and the number to read is the prediction for a row that did not set it. The slopes are printed so the arithmetic is visible. They are not an explanation of study time.

| hours | quiz | later score | role |
| --- | --- | --- | --- |
| 1 | 1 | 3 | training |
| 2 | 1 | 4 | training |
| 2 | 3 | 6 | training |
| 3 | 3 | 7 | training |
| 1 | 3 | 6 | held out |

The four training rows sit on $\text{later score} = 1 + \text{hours} + \text{quiz}$. The held-out row has 1 hour and a quiz of 3. The equation names 5. The later score was 6, so the absolute error is 1.

Four training rows were built to sit on a flat plane, with one held-out row and two features. A real later score would not sit on that plane. The held-out row is the last row, chosen by hand. Importing scikit-learn downloads it the first time.


## A plane fitted without the last row

The cell fits the plane on the four training rows and scores the held-out row.


```python
# Two features name a later score. The last row does not help to set the plane.
import numpy as np
from sklearn.linear_model import LinearRegression

train = np.array([
    [1.0, 1.0],
    [2.0, 1.0],
    [2.0, 3.0],
    [3.0, 3.0],
])
train_scores = np.array([3.0, 4.0, 6.0, 7.0])
held_out = np.array([[1.0, 3.0]])
held_out_score = 6.0

plane = LinearRegression()
plane.fit(train, train_scores)
# The intercept, then the weight on hours and the weight on the quiz.
print(round(float(plane.intercept_), 3))
print([round(float(value), 3) for value in plane.coef_])
prediction = float(plane.predict(held_out)[0])
absolute_error = abs(held_out_score - prediction)
print(round(prediction, 3))
print(round(absolute_error, 3))
```
