# Encoding and decoding

How can I turn a stimulus into a response, and estimate the stimulus from the response?

Encoding turns a stimulus into a response. In this illustration the response is a gain times the stimulus.

$$
\text{response} = \text{gain} \times \text{stimulus}
$$

Stimulus 3 and gain 2 give a response of 6. Decoding runs the same rule backwards. The estimate of the stimulus is the response divided by the gain. Response 6 and gain 2 give an estimate of 3. The estimate matches the stimulus because nothing was added.

These are numbers in a formula. They are not a recording.


```python
# Encoding scales the stimulus up by the gain. Decoding divides the response by that same gain, which would recover the stimulus if nothing else had been added.
# The stimulus presented.
stimulus = 3
# How strongly the response scales the stimulus.
gain = 2
# Encoding: response is gain times stimulus.
response = gain * stimulus
# Decoding: divide the response by the gain.
estimate = response / gain
# The response, then the decoded estimate.
print(response, estimate)
```

## A listed disturbance

Add a listed disturbance, one number per stimulus. The stimuli are 1, 2, and 3. The gain is 2. The disturbances are 0.5, $-0.5$, and 0.

The responses are 2.5, 3.5, and 6. Dividing by the gain estimates the stimuli as 1.25, 1.75, and 3. The absolute errors are 0.25, 0.25, and 0. Their mean is the average miss of the decoder on this list. The gain was known. A real decoder often has to use an estimated gain, and that is a further source of error. This cell does not estimate the gain.


```python
# A disturbance on the response comes back in the decoded stimulus, divided by the gain, so it comes back smaller than it was added.
# Three stimuli.
stimuli = [1, 2, 3]
# The same gain for each.
gain = 2
# What was added to each response.
disturbance = [0.5, -0.5, 0.0]
# Encode, then add the disturbance.
responses = [gain * stimuli[i] + disturbance[i] for i in range(3)]
# Decode each response by dividing by the gain.
estimates = [response / gain for response in responses]
# How far each decoded estimate sits from the stimulus.
errors = [abs(estimates[i] - stimuli[i]) for i in range(3)]
# The responses, the decoded estimates, and the average absolute error.
print(responses)
print(estimates)
print(round(sum(errors) / len(errors), 3))
```
