# A behavioural model and a neural summary

A behavioural model can produce a number, such as evidence for one response. A neural summary can be another number from the same moment. This notebook invents both, so the link is only an illustration of a comparison. It does not read a recording, and it does not carry out a neural analysis.

Evidence from the model: $[1, -2, 0.5]$. The simulated summary is half the evidence, plus a listed offset $[0.1, 0, -0.1]$.


```python
evidence = [1, -2, 0.5]
offset = [0.1, 0.0, -0.1]
summary = [0.5 * evidence[i] + offset[i] for i in range(3)]
print(summary)

matches = 0
for i in range(3):
    if evidence[i] * summary[i] > 0:
        matches = matches + 1
print(matches)
```

## What a matching sign shows

The products are positive, so the three pairs have the same sign, and the count is 3. The summary was built as a scaled copy of the evidence plus a small offset, so the agreement is no surprise. It shows how a comparison is counted. It does not show that a measurement tracked the behaviour.

A report of a comparison like this should say that the summary was simulated. A sign count is a weak summary even then: it ignores the size of either number. The mean of the products would keep the size. The count does not.
