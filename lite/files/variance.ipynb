# Variance of a fair coin

Use the same payoff as the expectation lesson: heads pays 1, tails pays 0, and the coin is fair, so p = 1/2.

For this payoff, Var(X) = p (1 − p).


```python
# Variance of a Bernoulli payoff. Heads pays 1, tails pays 0.
# Var(X) = p * (1 - p).
p = 0.5
variance = p * (1 - p)
print(f"Var(X) = {variance}")
```
