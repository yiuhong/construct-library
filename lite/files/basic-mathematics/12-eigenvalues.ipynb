# Eigenvalues

How can I find a direction that a square matrix only stretches?

For a square matrix $A$, an eigenvector is a vector $v$, not all zeros, that $A$ does not turn. $A$ only stretches or flips it. The amount of that stretch is a number $\lambda$, the eigenvalue.

$$
A v = \lambda v
$$

Run the next cell. $B$ times `[1, 1]` is `[3, 3]`, which is 3 times `[1, 1]`. So `[1, 1]` is an eigenvector of $B$, and 3 is its eigenvalue.


```python
# This matrix sends the vector to three times itself. The direction stays the same. Only the length changes.
# NumPy, for the matrix and the vector.
import numpy as np

# The matrix.
B = np.array([[2.0, 1.0], [1.0, 2.0]])
# A vector along a direction this matrix stretches.
v = np.array([1.0, 1.0])
# Where the matrix sends that vector.
print(B @ v)
# Three times the same vector, for comparison.
print(3 * v)
```

## Both eigenvalues

`eigvals` returns the eigenvalues. It does not return the vectors. For $B$ the eigenvalues are 3 and 1. The vector `[1, -1]` goes with 1: $B$ times that vector is the vector itself.

You will meet these when a model has a square matrix, for example a linear step from one state to the next. The matrix stretches some directions and shrinks others. Those directions are the eigenvectors. Recognising the equation above is the goal of this lesson.


```python
# A second direction, which this matrix leaves at the same length. Then both eigenvalues, sorted from smaller to larger.
# NumPy, for the matrix and the second direction.
import numpy as np

# The same matrix as the cell above.
B = np.array([[2.0, 1.0], [1.0, 2.0]])
# The other special direction.
w = np.array([1.0, -1.0])
# Where the matrix sends it.
print(B @ w)
# The eigenvalues as ordinary numbers. A tiny imaginary part from rounding is dropped.
vals = [float(value.real) for value in np.linalg.eigvals(B)]
# Smaller eigenvalue first.
print(sorted(vals))
```
