# Matrix operations

How can I add two tables, multiply them, and turn rows into columns?

Two matrices of the same shape add place by place. A single number multiplies every entry.


```python
# Add matching entries, or scale every entry by the same number. The shape of the result matches the shape you started with.
# NumPy holds each matrix.
import numpy as np

# The first matrix.
A = np.array([[1, 2], [3, 4]])
# The second matrix, the same shape as the first.
B = np.array([[5, 6], [7, 8]])
# Add entries that sit in the same place.
print(A + B)
# Double every entry of A.
print(2 * A)
```

## Multiply

Matrix multiplication is not place by place. Each entry of the product is one row of the first matrix, dotted with one column of the second.

The inner sizes have to match. A matrix of shape `(m, n)` can multiply a matrix of shape `(n, p)`. Both matrices below are 2 by 2. The top-left entry is $1 \times 1 + 2 \times 1 = 3$.


```python
# Each entry of the product comes from one row of the first matrix and one column of the second. Those two are multiplied place by place and then added.
# NumPy, for the two matrices.
import numpy as np

# The matrix whose rows are used.
A = np.array([[1, 2], [3, 4]])
# The matrix whose columns are used.
B = np.array([[1, 0], [1, 1]])
# The matrix product. This is not entry-by-entry multiplication.
print(A @ B)
```

## Transpose

The transpose writes each row as a column. A 2 by 3 table becomes 3 by 2. `.T` is the transpose.


```python
# Turn the rows into columns. The first row becomes the first column, and the shape swaps rows with columns.
# NumPy holds the table.
import numpy as np

# Two rows and three columns.
M = np.array([
    [1, 2, 3],
    [4, 5, 6],
])
# The transposed table: three rows and two columns.
print(M.T)
```
