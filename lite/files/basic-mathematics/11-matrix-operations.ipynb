# Matrix operations

How can I add two tables, multiply them, and turn rows into columns?

Two matrices of the same shape add place by place. A single number multiplies every entry.


```python
# Add matching entries, or scale every entry.
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(A + B)
print(2 * A)
```

## Multiply

Matrix multiplication is not place by place. Each entry of the product is one row of the first matrix, dotted with one column of the second.

The inner sizes have to match. A matrix of shape `(m, n)` can multiply a matrix of shape `(n, p)`. Both matrices below are 2 by 2. The top-left entry is $1 \times 1 + 2 \times 1 = 3$.


```python
# Each entry of the product comes from a row of the first matrix and a column of the second.
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[1, 0], [1, 1]])
print(A @ B)
```

## Transpose

The transpose writes each row as a column. A 2 by 3 table becomes 3 by 2. `.T` is the transpose.


```python
# Turn the rows into columns.
import numpy as np

M = np.array([
    [1, 2, 3],
    [4, 5, 6],
])
print(M.T)
```
