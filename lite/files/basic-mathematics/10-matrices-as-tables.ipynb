# Matrices as tables

How can I pick one entry out of a table of numbers?

A matrix is a table of numbers. Each row runs across. Each column runs down. This table has two rows and three columns.

|  | column 0 | column 1 | column 2 |
| --- | --- | --- | --- |
| row 0 | 1 | 2 | 3 |
| row 1 | 4 | 5 | 6 |

In writing, the first row is often called row 1. In this array the first row is index 0, the same indexing as a list. One entry is `table[row, column]`.


```python
# How many rows and columns, then two entries. Counting starts at 0, so the first row and the first column are both numbered 0.
# NumPy holds the table as an array.
import numpy as np

# Two rows and three columns.
table = np.array([
    [1, 2, 3],
    [4, 5, 6],
])
# The number of rows, then the number of columns.
print(table.shape)
# First row, third column.
print(table[0, 2])
# Second row, first column.
print(table[1, 0])
```

## Shape

`shape` is `(rows, columns)`. The table above has shape `(2, 3)`. The entry in row 0, column 2 is 3. The entry in row 1, column 0 is 4.

The next table is two people by two scores. Row 1, column 1 is the second score of the second person.


```python
# Two rows of scores, one row per person. The last line reads the second row and the second column.
# NumPy again, for a smaller table of scores.
import numpy as np

# Two scores in each of two rows.
scores = np.array([
    [5, 7],
    [6, 4],
])
# Rows, then columns.
print(scores.shape)
# Second row, second column.
print(scores[1, 1])
```
