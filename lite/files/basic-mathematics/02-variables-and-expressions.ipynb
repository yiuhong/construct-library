# Variables and expressions

A letter can stand for a number. Store that number under the letter's name. An expression combines numbers and names into a new value.

Run the next cell. `x` is 5. The expression `2 * x - 3` is computed from that.


```python
x = 5
print(2 * x - 3)
```

## Substitution

To evaluate an expression, replace the letter with its number, then do the arithmetic. Parentheses still happen first. With `x` equal to 5, `3 * (x - 1)` is `3 * 4`.


```python
x = 5
print(3 * (x - 1))
print(x ** 2)
```

## A name keeps the value it was given

The next cell stores `y` from `x`, then changes `x`. `y` does not follow. It keeps the number it was given.


```python
x = 5
y = x + 1
x = 0
print(x)
print(y)
```
