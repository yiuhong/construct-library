# Linear equations

How can I solve a linear equation and read the line it makes?

A linear equation has the unknown to the first power only. Solve $3x + 1 = 10$ by undoing each step. Subtract 1 from both sides: $3x = 9$. Divide both sides by 3: $x = 3$.

Check by putting 3 back in. $3 \times 3 + 1$ is 10, the same as the right-hand side.


```python
# Substitute 3 for x in the left-hand side, 3x + 1, and print that value.
# The value substituted for x.
x = 3
# Three times x, then add 1.
print(3 * x + 1)
```

## The graph is a line

$y = 2x + 1$ is a line. The 2 is the slope: $y$ increases by 2 when $x$ increases by 1. The 1 is the intercept: $y$ is 1 when $x$ is 0.

The figure appears under the cell. The first plot in a session can take a moment while the plotting library downloads.


```python
# One point for each chosen x on the line y = 2x + 1, then grey lines so the two axes are easy to see.
import matplotlib.pyplot as plt

# Four x values along the line.
xs = [-1, 0, 1, 2]
# The matching heights. The loop fills this list.
ys = []
# For each x, store the height 2x + 1.
for x in xs:
    ys.append(2 * x + 1)

# A figure and one set of axes, sized for the lesson page.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# Draw the points, joined in the order of the x values.
ax.plot(xs, ys, marker="o")
# A horizontal line where y is 0, and a vertical line where x is 0.
ax.axhline(0, color="#cccccc", linewidth=1)
ax.axvline(0, color="#cccccc", linewidth=1)
# Name the horizontal axis.
ax.set_xlabel("x")
# Name the vertical axis.
ax.set_ylabel("y")
# Show the figure under this cell.
plt.show()
```
