# Sums and averages

How can I add a list of values and find their mean?

The sum of a list is every value added. The mean is the sum divided by how many values there are. For values $x_1$ through $x_n$,

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

The bar over $x$ means the mean. The sigma means "add these". $n$ is the length of the list. The sum below is 20. The mean is 5.


```python
scores = [2, 4, 6, 8]
total = sum(scores)
n = len(scores)
mean = total / n
print(total)
print(mean)
```

## Distances from the mean

Each score minus the mean is a distance, negative when the score is below the mean. Those distances add to 0. That is what it means for 5 to be the balance point of this list. The mean is not "the middle item" unless the values happen to make it so. It is the sum divided by $n$.


```python
scores = [2, 4, 6, 8]
mean = sum(scores) / len(scores)
deviations = []
for score in scores:
    deviations.append(score - mean)
print(deviations)
print(sum(deviations))
```
