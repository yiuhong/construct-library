# Optimisation

How can I step toward an input that makes a function small?

To optimise a function, here, means to find an input that makes the output as small as possible.

Take $f(x) = (x - 2)^2$. A square is 0 at 0 and positive elsewhere, so the smallest value is at $x = 2$, and $f(2) = 0$.

The derivative is the slope, $f'(x) = 2(x - 2)$. Where the slope is positive, the function rises as $x$ increases. A step downhill goes the other way.

$$
x_{\text{new}} = x - \text{step} \cdot f'(x)
$$

The slope is the search direction. The step says how far to move. The next cell starts at 0 and takes 6 steps. The printed values move toward 2. The loop stops after those 6 steps.


```python
x = 0.0
step = 0.4
n = 0
while n < 6:
    slope = 2 * (x - 2)
    x = x - step * slope
    print(round(x, 3))
    n = n + 1
```

## More than one input

With several inputs, the gradient is the list of slopes, one per input. The search still steps the opposite way. This lesson uses one input, so the gradient is a single number. The next lesson uses the same kind of search when a fit looks for a best parameter.
