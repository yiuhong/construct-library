# Numbers and notation

How can I write a whole number, a fraction, a power, and a distance from zero?

A number can be a whole number, a negative, or a decimal. Click the next cell and press Shift+Enter.


```python
# A few of the kinds of number this lesson uses: a negative, zero, a whole number, and a decimal.
# A negative whole number.
print(-2)
# Zero, which is neither negative nor positive.
print(0)
# A positive whole number.
print(3)
# A decimal, halfway between 0 and 1.
print(0.5)
```

## Fractions and powers

A fraction is division. One half is `1 / 2`. A power is repeated multiplication. `10 ** 2` is 10 squared, written $10^2$. `10 ** 3` is 10 cubed.


```python
# A fraction is a division. A power multiplies a number by itself, once for each step in the exponent.
# One divided by two: the fraction one half.
print(1 / 2)
# Ten squared, which is 10 times 10.
print(10 ** 2)
# Ten cubed: one more factor of 10 than the line above.
print(10 ** 3)
```

## Absolute value

The absolute value of a number is its distance from zero. `abs` computes it. `-5` and `5` are the same distance from zero.

On a number line, order runs from left to right. `-2` is left of `0`, and `0` is left of `3`.


```python
# Distance from zero ignores the sign. The last line asks whether a number sits between two others.
# Distance of a negative number from zero.
print(abs(-5))
# The same distance for the matching positive number.
print(abs(5))
# True when the middle number sits between the other two.
print(-2 < 0 < 3)
```
