# Probability basics

How can I write a chance as a number from 0 to 1?

A probability is a number from 0 to 1. 0 means the outcome does not happen. 1 means it does. A value in between is a chance.

A fair coin has two outcomes, and neither is more likely than the other. $P(\text{heads}) = 1/2$. The probability of "not heads" is $1 - 1/2$.


```python
# A fair coin has two outcomes with equal probability. The second line is the probability of whichever outcome the first line did not take.
# One of the two faces.
print(1 / 2)
# Whatever probability is left, so the two probabilities add to 1.
print(1 - 1 / 2)
```

## Add outcomes that do not overlap

A fair die has six faces, 1 through 6. Each face has probability $1/6$. The even faces are 2, 4, and 6. Those faces do not overlap, so the probabilities add:

$$
P(\text{even}) = 1/6 + 1/6 + 1/6 = 1/2
$$


```python
# Keep the even faces of a die, then divide how many you kept by how many faces there are. That share is the probability of an even face.
# The six faces.
faces = [1, 2, 3, 4, 5, 6]
# The even faces, filled by the loop.
even = []
# Look at one face at a time.
for face in faces:
    # An even face leaves no remainder when divided by 2.
    if face % 2 == 0:
        # Keep that face.
        even.append(face)
# The faces that were kept, then their share of all six.
print(even)
print(len(even) / len(faces))
```

## The complement

The probability that an outcome does not happen is 1 minus the probability that it does. If an outcome has probability 0.2, the rest has probability 0.8.
