# Modules and help

A module is a file of names you can load. `import` loads it. The math module comes with this Python. After import, use its names as `math.sqrt` and `math.pi`.


```python
import math

print(math.sqrt(16))
print(round(math.pi, 2))
```

## A short description

A function's `__doc__` is a short text about what it does. `help(math.sqrt)` prints that kind of text too. Run the next cell and read the description of `sqrt`.


```python
import math

print(math.sqrt.__doc__)
```

## Import one name

`from math import sqrt` loads that one name. You can call `sqrt` without the `math.` prefix.


```python
from math import sqrt

print(sqrt(9))
```
