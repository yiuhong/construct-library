# Types and variables

A variable is a name for a value. Run the next cell. It stores a number under the name `score`, then prints that number.


```python
score = 7
print(score)
```

## Types

A value has a type. `7` is an int, a whole number. `3.5` is a float, a number with a decimal point. `"trial"` is a str, text. `True` and `False` are bool values.

`type` reports the type. The printed name is the part to read.


```python
count = 3
height = 3.5
label = "trial"
correct = True
print(type(count).__name__)
print(type(height).__name__)
print(type(label).__name__)
print(type(correct).__name__)
```

## A name can change

Run the next cell. `score` was 7. This cell adds 1 and stores the new value under the same name. The old value is gone.

Run the first code cell first if `score` is not defined yet.


```python
score = score + 1
print(score)
```
