# Functions

How can I name a block of code and get a result back?

A function is a name for some code. You pass it values. It returns a result. `def` starts the definition. `return` sends the result back. The block does not run until you call the function.

The printed number is the mean of 2, 4, and 7.


```python
# The mean of a list: the total divided by how many values there are. The same rule works for a list of any length.
# Name the rule mean. It takes one list.
def mean(values):
    # return sends the mean back to the caller.
    return sum(values) / len(values)

# Call the rule and round the result to three decimals.
print(round(mean([2, 4, 7]), 3))
```

## More than one input

A function can take more than one value. This one returns the larger of two numbers.


```python
# Return whichever input is greater. return is what sends that choice back to the caller.
# Two inputs, a and b.
def larger(a, b):
    # This branch runs only when a is the greater one.
    if a > b:
        # Send a back, and stop. The line after the branch does not run.
        return a
    # Otherwise send b back. This also covers a tie.
    return b

# Two calls, with a different input greater each time.
print(larger(4, 9))
print(larger(6, 2))
```

## If you forget return

If a function reaches the end without `return`, the result is `None`. The functions above return a number, so the caller can use it.
