# Lists and dictionaries

How can I keep values in a list, or look one up by a key?

A list keeps values in order. The first index is 0, not 1. A negative index counts from the end. Run the next cell.


```python
# The first item, the last item, and how many items there are.
scores = [2, 4, 7]
print(scores[0])
print(scores[-1])
print(len(scores))
```

## Adding a value

`append` adds one value at the end. The list stored in the kernel changes. `len` is the new length.


```python
# Add a score on the end. The list gets longer.
scores = [2, 4, 7]
scores.append(5)
print(scores)
print(len(scores))
```

## A dictionary

A dictionary pairs a key with a value. Look up the value by putting the key in square brackets.


```python
# Look a value up by its name, rather than by its position.
reading = {"time": 12, "correct": True}
print(reading["time"])
print(reading["correct"])
```
