# Samples and the mean of the sample means

Why can a sample mean miss the cohort mean, and what happens if every sample mean is averaged?

You will compare the cohort mean with one sample, then with a second sample, and then with the mean of every sample of two students. By the end you should be able to tell a sample mean from the cohort mean, and to see when the average of the sample means lands back on the cohort.

A sample describes the students in it. Two samples from the same six scores can sit far apart. That gap is sampling variation: the cohort did not change.

The previous lesson treated all six students as the population and wrote the probability of pass. The scores are the population here. A sample is only some of those scores.


## One sample and the cohort

The coursework scores, in row order from S1 to S6, are 2, 5, 8, 4, 7, and 10. They sum to 36, so the cohort mean is $36/6 = 6$. One sample is S1 and S4, with scores 2 and 4. Their mean is 3. That pair is a sample of two students. It is not the rest group, and it is not the trained group. The sample mean describes those two scores. It does not have to equal 6.


```python
# The mean of every coursework score, and the mean of one sample of two students.
# The score column, S1 through S6. This list is the population.
score = [2, 5, 8, 4, 7, 10]
# One sample: S1 and S4.
sample = [2, 4]
# Each mean is a sum divided by its own length.
cohort_mean = sum(score) / len(score)
sample_mean = sum(sample) / len(sample)
print(cohort_mean)
print(sample_mean)
```

## Two samples, two means

A second sample is S3 and S6, with scores 8 and 10. Their mean is 9. The cohort is still the same six scores. The gap between the sample means is $9 - 3 = 6$. A new pair can land on a new mean. The first plot in a session can take a moment while the plotting library downloads.


```python
# Two samples from the coursework scores, and the gap between their means.
import matplotlib.pyplot as plt

# The population both samples were drawn from.
score = [2, 5, 8, 4, 7, 10]
# S1 and S4, then S3 and S6.
sample_a = [2, 4]
sample_b = [8, 10]
# Each sample has its own mean.
mean_a = sum(sample_a) / len(sample_a)
mean_b = sum(sample_b) / len(sample_b)
# How far the second sample mean sits above the first.
gap = mean_b - mean_a
print(mean_a)
print(mean_b)
print(gap)

# One mark per sample mean.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter([1, 2], [mean_a, mean_b])
# Show the sample names instead of 1 and 2.
ax.set_xticks([1, 2], ["S1 and S4", "S3 and S6"])
# The horizontal axis is which sample.
ax.set_xlabel("Sample")
# The vertical axis is that sample's mean.
ax.set_ylabel("Sample mean")
# Show the figure under this cell.
plt.show()
```

## Every sample of size 2

Ignore order, and take every pair of students. S1 pairs with each of the other five, S2 with each of the four not yet paired, and so on: $5 + 4 + 3 + 2 + 1 = 15$ samples. The mean of those 15 sample means equals the cohort mean, 6. That is a fact about this complete list of pairs, not a promise about the one pair you happen to draw.


```python
# Every unordered pair of coursework scores, then the mean of those sample means.
# The score column, one entry per student, so two equal means would still be two students.
score = [2, 5, 8, 4, 7, 10]
# One mean per pair. The outer index stops before the last student.
sample_means = []
for i in range(len(score)):
    # The inner index starts after i, so each pair is kept once.
    for j in range(i + 1, len(score)):
        pair = [score[i], score[j]]
        sample_means.append(sum(pair) / len(pair))
# How many pairs were listed.
n_samples = len(sample_means)
# The mean of the sample means.
mean_of_means = sum(sample_means) / n_samples
print(n_samples)
print(mean_of_means)
```
