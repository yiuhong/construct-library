# Two samples, two means

Why can two samples from the same list have different means?

The population is `[2, 4, 6, 8]`. Sample A is `[2, 4]`, so its mean is 3. Sample B is `[6, 8]`, so its mean is 7. The list did not change. The samples did. The difference of the sample means is $7 - 3 = 4$.

That difference is sampling variation: a new sample can land on a new mean.


## The same list, two draws

The cell prints both sample means and the gap between them. The figure marks the two means. The first plot in a session can take a moment while the plotting library downloads.


```python
# Two samples from one list, and the gap between their means.
import matplotlib.pyplot as plt

# The list both samples were drawn from.
population = [2, 4, 6, 8]
# Two different samples.
sample_a = [2, 4]
sample_b = [6, 8]
# Each sample has its own mean.
mean_a = sum(sample_a) / len(sample_a)
mean_b = sum(sample_b) / len(sample_b)
# How far the second sample mean sits above the first.
gap = mean_b - mean_a
print(mean_a)
print(gap)

# One mark per sample mean.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter([1, 2], [mean_a, mean_b])
# Show the sample names instead of 1 and 2.
ax.set_xticks([1, 2], ["Sample A", "Sample B"])
# The horizontal axis is which sample.
ax.set_xlabel("Sample")
# The vertical axis is that sample's mean.
ax.set_ylabel("Sample mean")
# Show the figure under this cell.
plt.show()
```
