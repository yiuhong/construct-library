# The same difference as a line

How is that group gap the slope of a straight line?

Code rest as 0 and trained as 1. The mean of the outcome is then a line in that code:

$$
\text{mean} = b_0 + b_1 \times \text{group}
$$

$b_0$ is the mean when the group code is 0, so it is the rest mean, 4. $b_1$ is the change when the code goes from 0 to 1, so it is the group difference, 3. The prediction for trained is $4 + 3 = 7$, the trained mean.

A comparison of two means is a straight line with a 0/1 predictor. The later notebooks keep this line and then change the kind of outcome.


## Intercept, slope, and one prediction

The cell sets the two coefficients from the group means and predicts the trained mean from the line. The figure draws the two group means on that line. The first plot in a session can take a moment while the plotting library downloads.


```python
# The group gap is the slope of a line that uses a 0/1 group code.
import matplotlib.pyplot as plt

# Rest is coded 0. Trained is coded 1.
rest_mean = 4
trained_mean = 7
# The intercept is the mean at group code 0.
intercept = rest_mean
# The slope is the change from code 0 to code 1.
slope = trained_mean - rest_mean
# The line at group code 1.
prediction = intercept + slope * 1
print(intercept)
print(slope)
print(prediction)

# The two group means sit on the line.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot([0, 1], [intercept, prediction], marker="o")
# Show the group names instead of the codes 0 and 1.
ax.set_xticks([0, 1], ["Rest", "Trained"])
# The horizontal axis is the group code.
ax.set_xlabel("Group code")
# The vertical axis is the group mean.
ax.set_ylabel("Mean")
# Show the figure under this cell.
plt.show()
```
