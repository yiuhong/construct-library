# A difference of group means

How large is the gap between two group means?

This notebook is in Introductory statistics, in the module Comparing groups. Comparing groups starts with the two means. The next notebook writes that same gap as the slope of a line, which is the pattern the rest of this course keeps.

Rest is `[2, 4, 6]`, so its mean is 4. Trained is `[5, 7, 9]`, so its mean is 7. Trained minus rest is 3.


## Subtract the two centres

The cell prints each mean and the difference.


```python
# Two group means, and the gap between them.
# Rest, then trained.
rest = [2, 4, 6]
trained = [5, 7, 9]
# Each mean divides the group's sum by the group's length.
rest_mean = sum(rest) / len(rest)
trained_mean = sum(trained) / len(trained)
# The comparison this lesson reports.
difference = trained_mean - rest_mean
print(trained_mean)
print(difference)
```
