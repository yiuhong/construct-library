# An event and its complement

How do I write the probability of an event, and of its opposite?

This notebook opens Introductory statistics, in the module Probability and sampling. Probability here is the language for a draw from a list.

A bag holds 2 red chips and 3 blue chips, so 5 chips in all. The event "red" has probability $2/5 = 0.4$. The complement is the draw that is not red. Its probability is $1 - 0.4 = 0.6$. The two probabilities add to 1, because every chip is either red or not red.


## One subtraction

The cell divides the red count by the size of the bag, then subtracts that probability from 1.


```python
# The probability of red, then the probability of the complement.
# How many red chips, and how many chips in the bag.
red = 2
n = 5
# The probability is the red count divided by the size of the bag.
p_red = red / n
# The complement is everything else, so subtract from 1.
p_not_red = 1 - p_red
print(p_red)
print(p_not_red)
print(p_red + p_not_red)
```
