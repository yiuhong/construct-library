# An event and its complement

What is the probability of passing, and of not passing?

You will write the probability of pass in the coursework cohort, and the probability of fail, which is the complement. By the end you should be able to get the second probability by subtracting the first from 1.

Here a probability is a share of a class treated as the whole population. Every student is in the event or out of it, so the two shares use up the class.

The weekly quiz class was described as it stood. This course uses the coursework cohort, six students treated as the population, and starts from pass or fail.


## Pass in the cohort

The pass mark is 5. A score of 5 or more is recorded as pass, and a score below 5 is fail. Four of the six students passed.

| student | group | hours | score | passed | attempts |
| --- | --- | --- | --- | --- | --- |
| S1 | rest | 0 | 2 | fail | 1 |
| S2 | rest | 1 | 5 | pass | 2 |
| S3 | rest | 2 | 8 | pass | 2 |
| S4 | trained | 0 | 4 | fail | 1 |
| S5 | trained | 1 | 7 | pass | 2 |
| S6 | trained | 2 | 10 | pass | 4 |

The group is rest or trained. `hours` is hours of practice. `score` is the coursework score out of 10. `attempts` is how many times the student attempted the coursework. These values were chosen so the arithmetic in this course stays on whole numbers. Later lessons use the same six rows.

The probability of pass is the pass count divided by the size of the cohort, $4/6 = 2/3$.


```python
# The probability of pass in the coursework cohort.
# The passed column, in row order: S1 through S6.
passed = ["fail", "pass", "pass", "fail", "pass", "pass"]
# How many students passed.
pass_count = 0
for outcome in passed:
    # A pass adds one to the count.
    if outcome == "pass":
        pass_count = pass_count + 1
# The cohort is the whole population here.
n = len(passed)
# The probability is the pass count divided by the size of the cohort.
p_pass = pass_count / n
print(pass_count)
print(p_pass)
```

## The complement

Fail is the complement of pass: the students who did not pass. Its probability is $1 - 2/3 = 1/3$, which is also $2/6$. The two probabilities add to 1, because every student passed or failed.


```python
# The probability of pass, then the probability of the complement.
# Four of the six students passed.
pass_count = 4
n = 6
# The probability is the pass count divided by the size of the cohort.
p_pass = pass_count / n
# The complement is everyone who did not pass, so subtract from 1.
p_fail = 1 - p_pass
print(p_pass)
print(p_fail)
print(p_pass + p_fail)
```
