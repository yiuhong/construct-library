# A yes-or-no outcome and a count outcome

How can a line state a probability, or a mean count, and stay inside the values those outcomes allow?

You will read the cohort's pass rate through a logit link, and the attempts through a log link. By the end you should be able to turn a stated linear predictor into a probability and into a mean count.

Pass either happened or it did not, so its mean is a probability between 0 and 1. Attempts are a count, so their mean has to stay positive. A straight line on the outcome's own numbers can leave those bounds. A generalised linear model keeps the line and puts a link between the line and the mean.

The previous lesson's line stated a mean score directly. Pass and attempts are different outcomes on the same six students. The link is what changes.


## Pass, and the logit link

Four of the six students passed, so the pass proportion is $4/6 = 2/3$. A line drawn on that probability can fall below 0 or climb above 1. For a yes-or-no outcome the link used here is the logit. The line, written $\eta$, stays unrestricted, and the probability is read back from it:

$$
p = \frac{1}{1 + e^{-\eta}}
$$

When $\eta = 0$, the probability is 0.5. When $\eta = \log 2$, the odds are 2 and the probability is $2/3$, the cohort's pass rate. The cell does not fit $\eta$ from the rows. It reads a probability from a linear predictor that is already stated.

Teaching note. Nelder, J. A., & Wedderburn, R. W. M. (1972). Generalized linear models. *Journal of the Royal Statistical Society. Series A (General), 135*(3), 370-384. https://doi.org/10.2307/2344614


```python
# Turn a linear predictor into a probability. The line stays unrestricted. The probability stays between 0 and 1.
import math
import matplotlib.pyplot as plt

def probability(linear_predictor):
    # e to the power of minus the line.
    expo = math.exp(-linear_predictor)
    # The logistic map, from the line back to a probability.
    return 1 / (1 + expo)

# Zero on the logit scale, then the value that returns the cohort pass rate.
print(probability(0))
print(probability(math.log(2)))

# A few points along the line, so the curve can be seen.
etas = [-2, -1, 0, 1, 2]
probs = []
for eta in etas:
    probs.append(probability(eta))
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(etas, probs, marker="o")
# The horizontal axis is the line, on the logit scale.
ax.set_xlabel("Linear predictor")
# The vertical axis is the probability.
ax.set_ylabel("Probability")
plt.show()
```

## Attempts, and the log link

Attempts in the cohort, from S1 to S6, are 1, 2, 2, 1, 2, and 4. They sum to 12, so the mean count is $12/6 = 2$. A count's mean has to stay positive. The link used here is the log, so the line is the log of the mean count:

$$
\text{mean} = e^{\eta}
$$

When $\eta = 0$, the mean count is 1, which is the attempts value for S1 and for S4. When $\eta = \log 2$, the mean count is 2, the cohort mean. When $\eta = \log 4$, the mean count is 4, which is S6's attempts. A Poisson distribution is a simple distribution for counts in this pattern. Its mean is the number the log link produces. This cell only reads that mean. It does not fit the line.


```python
# The log link keeps the mean count positive. Exponentiating the line reads the mean back.
import math

def predicted_count(linear_predictor):
    # Undo the log. The result is the mean count.
    return math.exp(linear_predictor)

# The cohort mean of attempts is 2. The smallest attempts value is 1, and the largest is 4.
print(predicted_count(0))
print(predicted_count(math.log(2)))
print(predicted_count(math.log(4)))
```

## A line that leaves the probability scale

The same hours that sit on the coursework rows, 0, 1, and 2, show why the link is there. A straight line written on the probability itself, $0.1 + 0.5 \times \text{hours}$, gives 0.1, 0.6, and 1.1. The last of those is above 1, so it cannot be a probability. The same hours through a logit, with a stated line $\eta = -2 + 2 \times \text{hours}$, stay between 0 and 1.

Those coefficients are stated so the contrast is visible. They are not estimated from who passed or from the attempts. A full analysis would estimate a linear predictor for `passed` and, separately, one for `attempts`. It would use a binomial likelihood for the yes-or-no column and a count likelihood, such as the Poisson, for the attempts. Six rows are not enough to check either fit.


```python
# A line on the probability can leave 0 and 1. A logit of a line cannot.
import math

def probability(linear_predictor):
    # The same map as the logit cell.
    return 1 / (1 + math.exp(-linear_predictor))

# Hours from the coursework cohort. The coefficients here are stated, not fitted.
hours = [0, 1, 2]
naive = []
linked = []
for value in hours:
    # A straight line on the probability scale. At 2 hours it is above 1.
    naive.append(0.1 + 0.5 * value)
    # A stated line on the logit scale, read back as a probability.
    linked.append(probability(-2 + 2 * value))
print(naive)
print(linked)
```
