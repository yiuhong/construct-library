# A line for the mean, with a group and a slope

How can one line carry both a group shift and a slope for hours?

You will read a line for the rest students' hours and scores, then add a group term so the same hours slope describes the whole cohort. By the end you should be able to evaluate that line at a chosen group and a chosen number of hours.

A group comparison and a slope are the same kind of term. Each one shifts the mean. One equation can hold both.

The previous lesson gave every rest student the same prediction, because the line used only the group. Hours of practice are already on the rows. This lesson lets the mean change with hours, then puts the group shift back in.


## A line for the rest students

For S1, S2, and S3, hours are 0, 1, and 2 and scores are 2, 5, and 8. The line

$$
\text{mean} = 2 + 3 \times \text{hours}
$$

matches each of those scores. The intercept 2 is the mean at 0 hours. The slope 3 is how much that mean changes when hours increase by 1. At 2 hours the mean is 8.

These three rows were chosen so the line hits them exactly. A full analysis would estimate the intercept and the slope from scores that miss the line, and it would report how uncertain those two numbers are.


```python
# The line states a mean score at each number of hours, for the rest students only.
import matplotlib.pyplot as plt

# Rest rows from the coursework cohort: S1, S2, S3.
hours = [0, 1, 2]
score = [2, 5, 8]
# The intercept is the mean at 0 hours.
intercept = 2
# The slope is the change in the mean when hours increase by 1.
slope = 3
# A predicted mean for each hour.
predicted = []
for value in hours:
    predicted.append(intercept + slope * value)
print(intercept)
print(predicted)

# The scores, and the line that states their mean.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter(hours, score, label="Rest scores")
ax.plot(hours, predicted, label="Mean from the line")
ax.set_xlabel("Hours")
ax.set_ylabel("Score")
ax.legend()
plt.show()
```

## A group term in the same line

Trained students have the same hours, 0, 1, and 2, and scores 4, 7, and 10. That is the rest line plus 2 at every hour. One equation carries both:

$$
\text{mean} = 2 + 2 \times \text{group} + 3 \times \text{hours}
$$

Rest is coded 0 and trained is coded 1, as in the previous lesson. The group coefficient, 2, is the shift from rest to trained at the same hours. It matches the previous lesson's group difference because the two groups have the same hours. The slope on hours stays 3. At group 1 and 1 hour the mean is $2 + 2 + 3 = 7$, which is S5's score.


```python
# One line with a group term and a slope, on all six coursework rows.
# Rest is coded 0. Trained is coded 1. Hours and score follow S1 through S6.
group = [0, 0, 0, 1, 1, 1]
hours = [0, 1, 2, 0, 1, 2]
score = [2, 5, 8, 4, 7, 10]
intercept = 2
group_coefficient = 2
slope = 3

predicted = []
for i in range(len(score)):
    # Shared intercept, plus the group shift, plus the slope times hours.
    mean = intercept + group_coefficient * group[i] + slope * hours[i]
    predicted.append(mean)
print(group_coefficient)
print(predicted)
```

## What the exact fit leaves out

Every prediction above matches the score on that row, so every residual is 0 and the sum of squared residuals is 0. That is how these six rows were chosen, not a result to expect from a class. A full analysis would estimate the intercept, the group coefficient, and the hours slope, would report uncertainty around them, and would not treat a residual of zero as something the scores had to do.


```python
# Residual is score minus the line. On these rows each residual is 0.
group = [0, 0, 0, 1, 1, 1]
hours = [0, 1, 2, 0, 1, 2]
score = [2, 5, 8, 4, 7, 10]
intercept = 2
group_coefficient = 2
slope = 3

# A running total of the squared residuals.
total = 0
for i in range(len(score)):
    prediction = intercept + group_coefficient * group[i] + slope * hours[i]
    residual = score[i] - prediction
    total = total + residual ** 2
print(total)
```
