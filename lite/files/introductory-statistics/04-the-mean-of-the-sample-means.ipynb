# The mean of the sample means

What happens if I average every sample mean of a given size?

The population is `[0, 2, 4]`. Its mean is 2. Take every sample of size 2, without replacement, and ignore order. There are three samples: `[0, 2]` with mean 1, `[0, 4]` with mean 2, and `[2, 4]` with mean 3.

The mean of those three sample means is $(1 + 2 + 3) / 3 = 2$. For this balanced list of samples, the average of the sample means equals the population mean.


## All three samples

The cell stores each sample mean, then averages that short list.


```python
# Every sample of size 2, then the mean of those sample means.
# The three samples, written out in full.
samples = [[0, 2], [0, 4], [2, 4]]
# One mean per sample.
sample_means = []
for sample in samples:
    # The mean of this sample.
    sample_means.append(sum(sample) / len(sample))
# How many samples were listed.
n_samples = len(sample_means)
# The mean of the sample means.
mean_of_means = sum(sample_means) / n_samples
print(sample_means)
print(n_samples)
print(mean_of_means)
```
