# Comparing models

A smaller error means the predictions sit closer to these points. A higher likelihood means these data are more probable under the model. Neither score, on its own, says you should prefer a model that was given more parameters to adjust.

This lesson compares two models of the same three points, $(1, 3)$, $(2, 5)$, and $(3, 8)$.

The slope-only model predicts $\hat{y} = ax$. It adjusts one parameter, the slope.

The line model predicts $\hat{y} = b + ax$. It adjusts two parameters, the intercept and the slope.


## Three scores

Error here is the sum of squared residuals.

For the likelihood, take the residuals as normal with spread fixed at 1. That spread is not an adjusted parameter. For $n$ points,

$$
\log L = -\frac{n}{2}\log(2\pi) - \frac{1}{2}\sum (y - \hat{y})^2
$$

The sum of squares is the part that differs between the two models. The closer fit has the higher log likelihood.

Akaike's information criterion adds a penalty of 2 for each adjusted parameter. $k$ is that count. Smaller AIC is preferred in this lesson.

$$
\mathrm{AIC} = -2 \log L + 2k
$$

The line pays a larger penalty than the slope-only model, because it adjusts one more parameter. A small gain in likelihood can lose to that penalty.

Teaching note. Akaike, H. (1974). A new look at the statistical model identification. *IEEE Transactions on Automatic Control, 19*(6), 716-723. https://doi.org/10.1109/TAC.1974.1100705


```python
import math

xs = [1, 2, 3]
ys = [3, 5, 8]
n = 3

sum_x = 0
sum_y = 0
sum_xy = 0
sum_xx = 0
for i in range(n):
    sum_x = sum_x + xs[i]
    sum_y = sum_y + ys[i]
    sum_xy = sum_xy + xs[i] * ys[i]
    sum_xx = sum_xx + xs[i] ** 2

slope_only = sum_xy / sum_xx
slope = (n * sum_xy - sum_x * sum_y) / (n * sum_xx - sum_x ** 2)
intercept = sum_y / n - slope * (sum_x / n)

pred_slope = []
pred_line = []
for x in xs:
    pred_slope.append(slope_only * x)
    pred_line.append(intercept + slope * x)

def sse(predictions):
    total = 0
    for i in range(n):
        total = total + (ys[i] - predictions[i]) ** 2
    return total

def log_likelihood(sum_of_squares):
    return -n / 2 * math.log(2 * math.pi) - sum_of_squares / 2

def aic(sum_of_squares, k):
    return -2 * log_likelihood(sum_of_squares) + 2 * k

sse_slope = sse(pred_slope)
sse_line = sse(pred_line)
print("slope SSE", round(sse_slope, 3))
print("line SSE", round(sse_line, 3))
print("slope log L", round(log_likelihood(sse_slope), 3))
print("line log L", round(log_likelihood(sse_line), 3))
print("slope AIC", round(aic(sse_slope, 1), 3))
print("line AIC", round(aic(sse_line, 2), 3))
```

Compare the two sums of squares. The smaller error is the closer fit to these three points.

Compare the two log likelihoods. Higher means the points are more probable under that model. With the spread fixed, this ranking matches the error.

Compare the two AIC values. This lesson prefers the smaller one. The penalty is why that ranking can disagree with the error.
