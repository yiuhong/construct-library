# Learning curves

How can I plot a prediction trial by trial?

A learning curve plots the prediction after each trial. The update is the Rescorla–Wagner step from the incremental-learning lesson: add a step size times the prediction error (Rescorla & Wagner, 1972). The full reference is in that lesson.

A larger step size climbs faster. Both curves below start at 0, the outcome is 1 on every trial, and there are six trials. One step size is 0.2. The other is 0.6.

Run the cell. The printed lists are the predictions after each trial. A player under the figure draws those same curves one trial at a time. Press play, or drag the slider. The first plot in a session can take a moment.


```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

def curve(step, trials):
    prediction = 0.0
    values = []
    for trial in range(trials):
        prediction = prediction + step * (1 - prediction)
        values.append(prediction)
    return values

slow = curve(0.2, 6)
fast = curve(0.6, 6)
print([round(v, 3) for v in slow])
print([round(v, 3) for v in fast])

fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.set_xlim(1, 6)
ax.set_ylim(0, 1)
ax.set_xlabel("Trial")
ax.set_ylabel("Prediction")
slow_line, = ax.plot([], [], marker="o", label="Step 0.2")
fast_line, = ax.plot([], [], marker="o", label="Step 0.6")
ax.legend()

def draw(frame):
    n = frame + 1
    trials = list(range(1, n + 1))
    slow_line.set_data(trials, slow[:n])
    fast_line.set_data(trials, fast[:n])
    ax.set_title("After trial " + str(n))
    return (slow_line, fast_line)

anim = FuncAnimation(fig, draw, frames=6, interval=500, repeat=False)
plt.close(fig)
HTML(anim.to_jshtml())
```

## Reading the two curves

The slow curve is still moving at trial 6. The fast curve is already close to 1. They are heading for the same outcome. The step size changes the speed, not the outcome these trials keep producing.

A curve like this is a prediction of the model, trial by trial. It is not yet a comparison with a person's responses. That comparison would need a list of responses and a rule for how far the curve sits from them.
