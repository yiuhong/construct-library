# A psychometric curve

How can I map a stimulus level to the probability of one response?

A psychometric curve maps a stimulus level to the probability of a response. A logistic curve is a smooth step. Its midpoint $m$ is the stimulus where the probability is $1/2$. Its scale $s$ sets how quickly the probability rises. A smaller scale is a steeper curve.

$$
p(x) = \frac{1}{1 + e^{-(x - m)/s}}
$$

With $m = 0$ and $s = 1$, the probability at 0 is $1/2$. At a stimulus of 2 the exponent is $-2$, and the probability is about 0.881. The curve is this formula. The next cell evaluates it and draws it. The first plot in a session can take a moment.


```python
import math
import numpy as np
import matplotlib.pyplot as plt

def psychometric(x, midpoint, scale):
    return 1 / (1 + math.exp(-(x - midpoint) / scale))

for stimulus in [-2, 0, 2]:
    print(stimulus, round(psychometric(stimulus, 0, 1), 3))

grid = np.linspace(-4, 4, 81)
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(grid, [psychometric(x, 0, 1) for x in grid])
ax.set_xlabel("Stimulus")
ax.set_ylabel("Probability")
plt.show()
```

## Fitting the midpoint

The scale stays at 1. Five stimuli, $-2, -1, 0, 1, 2$, have observed proportions $0.05, 0.12, 0.27, 0.50, 0.73$. The loss is the sum of squared gaps between the curve and those proportions. The midpoint starts at $-1$ and steps against the slope of that loss. The step size is 2.2.

The probability is still $1/2$ at the midpoint. Sliding the midpoint slides the whole curve. A smaller scale would make the same slide steeper. This fit does not change the scale.

Run the cell. It prints the midpoint after each step. A player under the figure draws the curve at that midpoint, with the five proportions as points. Press play, or drag the slider.


```python
import math
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

stimuli = [-2, -1, 0, 1, 2]
observed = [0.05, 0.12, 0.27, 0.50, 0.73]

def psychometric(x, midpoint):
    return 1 / (1 + math.exp(-(x - midpoint)))

def loss(midpoint):
    total = 0
    for i in range(5):
        total = total + (psychometric(stimuli[i], midpoint) - observed[i]) ** 2
    return total

def slope_of_loss(midpoint):
    h = 1e-4
    return (loss(midpoint + h) - loss(midpoint - h)) / (2 * h)

midpoints = [-1.0]
for i in range(5):
    midpoints.append(midpoints[-1] - 2.2 * slope_of_loss(midpoints[-1]))
print([round(m, 3) for m in midpoints])
print([round(loss(m), 4) for m in midpoints])

grid = [i / 10 for i in range(-30, 31)]
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(stimuli, observed, marker="o", linestyle="none")
curve, = ax.plot([], [])
ax.set_xlim(-3, 3)
ax.set_ylim(0, 1)
ax.set_xlabel("Stimulus")
ax.set_ylabel("Probability")

def draw(frame):
    midpoint = midpoints[frame]
    curve.set_data(grid, [psychometric(x, midpoint) for x in grid])
    ax.set_title("Midpoint " + str(round(midpoint, 3)))
    return (curve,)

anim = FuncAnimation(fig, draw, frames=len(midpoints), interval=500, repeat=False)
plt.close(fig)
HTML(anim.to_jshtml())
```
