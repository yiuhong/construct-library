# Two independent groups

How can two separate groups be compared when the comparison uses ranks rather than a difference of means?

This notebook is in Non-parametric tests, in the module Comparing groups. Introductory statistics compared two groups by their means. Here each score is a different student, and the test is the Mann–Whitney test. It counts, over every pair with one score from each group, how often the first group's score is higher.

| sitting | scores |
| --- | --- |
| morning | 2, 4, 5, 6 |
| afternoon | 3, 7, 8, 9 |

Pooled and ordered, the eight scores and their ranks are:

| score | 2 | 3 | 4 | 5 | 6 | 7 | 8 | 9 |
| --- | --- | --- | --- | --- | --- | --- | --- | --- |
| sitting | morning | afternoon | morning | morning | morning | afternoon | afternoon | afternoon |
| rank | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |

The morning rank sum is $1 + 3 + 4 + 5 = 13$. There are $4 \times 4 = 16$ pairs. The morning score is higher in 3 of them: 4 against 3, 5 against 3, and 6 against 3. That count is the Mann–Whitney statistic for the morning group when those scores are passed first. The two-sided p-value on these eight scores is 0.2.

Four and four chosen scores, with no ties, are not two randomised sittings. A tie would have counted as half a win. This table has none, so the count is a whole number. The test does not assume a normal shape. It asks whether a morning score and an afternoon score are equally likely to be the higher one. A published comparison would state how the groups were formed. The statistic is not a difference of means.

Teaching note. Mann, H. B., & Whitney, D. R. (1947). On a test of whether one of two random variables is stochastically larger than the other. *The Annals of Mathematical Statistics, 18*(1), 50-60. https://doi.org/10.1214/aoms/1177730491


## Pairwise wins

The cell ranks the eight scores in one list, sums the morning ranks, and counts the pairs in which the morning score is higher. `mannwhitneyu` prints that count as its statistic.


```python
# Two sittings, four students each. No one appears in both lists.
from scipy.stats import mannwhitneyu, rankdata

morning = [2, 4, 5, 6]
afternoon = [3, 7, 8, 9]
# One pooled list, so the ranks are places in the combined order.
pooled = morning + afternoon
# float() prints an ordinary number rather than a numpy scalar.
ranks = [float(rank) for rank in rankdata(pooled)]
print(ranks)
# The first four pooled scores are the morning scores. Add their ranks.
morning_rank_sum = 0
for rank in list(ranks)[: len(morning)]:
    morning_rank_sum = morning_rank_sum + rank
print(morning_rank_sum)
# One win when the morning score is strictly higher. This table has no ties, so none are split.
wins = 0
for morning_score in morning:
    for afternoon_score in afternoon:
        if morning_score > afternoon_score:
            wins = wins + 1
print(wins)
result = mannwhitneyu(morning, afternoon, alternative="two-sided")
print(result.statistic)
print(result.pvalue)
```
