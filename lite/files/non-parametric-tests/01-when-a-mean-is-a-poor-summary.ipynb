# When a mean is a poor summary

What happens to a mean when one rating sits far from the others?

This notebook opens Non-parametric tests, in the module Ranks and related scores. Descriptive statistics already set a mean beside a median when one score sits far out. The list there was `[2, 3, 4, 5, 20]`. The contrast here is the same kind of contrast. It is a reason to look past the mean, not a second definition of it.

Seven people answered one questionnaire item. The ratings were chosen.

| person | rating |
| --- | --- |
| A | 2 |
| B | 2 |
| C | 2 |
| D | 3 |
| E | 3 |
| F | 3 |
| G | 9 |

Six ratings sit at 2 or 3. The 9 pulls the mean up to $24/7$, about 3.43, which is above every rating except G. The middle rating, once the list is sorted, stays at 3.

Seven chosen ratings, with the 9 written in so the pull is easy to see, are not a sample. There is no tie to resolve, because the low ratings repeat a value rather than sharing a rank, and there is no design. A published summary would say how the people were selected. It would not stop at one chosen item, and it would not treat this mean as a finding.


## The mean above the cluster

The cell divides the sum by 7, takes the middle rating of the sorted list, and counts how many ratings sit strictly below the mean. The figure marks each rating. Repeated ratings are stacked so the cluster is visible. The first plot in a session can take a moment while the plotting library downloads.


```python
# Seven chosen ratings. The 9 is there so the pull on the mean is easy to see.
import matplotlib.pyplot as plt

ratings = [2, 2, 2, 3, 3, 3, 9]
# The mean, as Descriptive statistics defined it: the sum divided by how many ratings.
total = 0
for rating in ratings:
    total = total + rating
mean = total / len(ratings)
# The median of an odd-length list is the middle value after sorting.
ordered = sorted(ratings)
median = ordered[len(ordered) // 2]
# How many ratings sit strictly below that mean.
below = 0
for rating in ratings:
    if rating < mean:
        below = below + 1
print(mean)
print(median)
print(below)

# Stack repeated ratings so six dots are not drawn on top of one another.
fig, ax = plt.subplots(figsize=(6.4, 3.2))
seen = {}
xs = []
ys = []
for rating in ratings:
    seen[rating] = seen.get(rating, 0) + 1
    xs.append(rating)
    ys.append(seen[rating])
ax.scatter(xs, ys)
ax.axvline(mean, label="Mean")
ax.axvline(median, linestyle="--", label="Median")
ax.set_xlabel("Rating")
ax.set_ylabel("Ratings at that value")
ax.legend()
plt.show()
```
