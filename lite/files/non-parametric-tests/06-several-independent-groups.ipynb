# Several independent groups

What does one rank test report when more than two separate groups are compared at once?

This notebook is in Non-parametric tests, in the module Comparing groups. Three groups sat the same short test. Each score is a different student. The Kruskal–Wallis test ranks every score in one list, then compares the rank sums of the groups. It does not report a difference of means, and it does not yet say which group differs.

The scores were chosen so the groups line up.

| group | scores |
| --- | --- |
| no notes | 2, 3, 5 |
| short notes | 4, 6, 7 |
| worked example | 8, 9, 10 |

The ranks run from 1 to 9. The rank sums are 7, 14, and 24. `kruskal` turns those sums into one statistic, about 6.49, and a p-value from a chi-square approximation with 2 degrees of freedom. On this table that p-value is about 0.039. With three scores in each group, the approximation is thin, and the groups were typed in this order. A small p-value here is a property of the illustration.

Nine chosen scores and no ties differ from a published Kruskal–Wallis test. A tie would change the ranks and the reference distribution. This table has none. A published analysis would use a larger sample, would not treat the chi-square approximation as exact at this size, and would not stop here if the question was which group differs.

Teaching note. Kruskal, W. H., & Wallis, W. A. (1952). Use of ranks in one-criterion variance analysis. *Journal of the American Statistical Association, 47*(260), 583-621. https://doi.org/10.1080/01621459.1952.10483441


## One ranking, three rank sums

The cell ranks the nine scores together, adds the ranks inside each group, and calls `kruskal`.


```python
# Three separate groups. The scores were chosen so the groups line up.
from scipy.stats import kruskal, rankdata

no_notes = [2, 3, 5]
short_notes = [4, 6, 7]
worked_example = [8, 9, 10]
pooled = no_notes + short_notes + worked_example
# float() prints an ordinary number rather than a numpy scalar.
ranks = [float(rank) for rank in rankdata(pooled)]
print(ranks)
# Split the pooled ranks back into the three groups, in the same order, and add each group.
groups = [no_notes, short_notes, worked_example]
start = 0
for group in groups:
    rank_sum = 0
    for rank in ranks[start : start + len(group)]:
        rank_sum = rank_sum + rank
    start = start + len(group)
    print(rank_sum)
result = kruskal(no_notes, short_notes, worked_example)
print(result.statistic)
print(result.pvalue)
```
