# Ranks instead of raw scores

What is kept, and what is dropped, when a score is replaced by its place in the list?

This notebook is in Non-parametric tests, in the module Ranks and related scores. Five test scores were chosen, including one tie.

| student | score |
| --- | --- |
| A | 4 |
| B | 11 |
| C | 7 |
| D | 7 |
| E | 15 |

Sorted, the scores are 4, 7, 7, 11, 15. The 4 is in 1st place. The two 7s occupy 2nd and 3rd, so each is given the average, $(2 + 3) / 2 = 2.5$. The 11 is 4th and the 15 is 5th. The gap from 11 to 15 is four points of score and only one step of rank. Adding 20 to every score leaves those places unchanged.

Five chosen scores, with one tie written in, are not a normed test. Nothing here assumes a shape for the scores. The spacing between them is simply discarded. A published ranking would say how ties are handled, and it would rank a sample rather than a list typed so the shared rank is obvious.


## Average ranks, and a gap that shrinks

The cell writes the shared rank of the two 7s by hand, then asks `rankdata` for the whole list. The first time this session imports scipy, that library downloads. That can take a moment. Adding 20 is a second list, scored only to show that the ranks do not move.


```python
# Five chosen scores, in student order. The two 7s are the tie.
from scipy.stats import rankdata

scores = [4, 11, 7, 7, 15]
# Sorted order. Places will be read off this list, not off the student order.
ordered = sorted(scores)
print(ordered)
# The two 7s occupy 2nd and 3rd place. They share the average of those places.
shared_rank = (2 + 3) / 2
print(shared_rank)
# rankdata uses that same average for ties. The result follows the original student order.
# float() prints an ordinary number rather than a numpy scalar.
ranks = [float(rank) for rank in rankdata(scores)]
print(ranks)
# The gap in score from 11 to 15, and the gap in rank for those same two scores.
score_gap = 15 - 11
rank_gap = 5 - 4
print(score_gap)
print(rank_gap)
# Add 20 to every score. The order, and so the ranks, stay the same.
shifted = []
for score in scores:
    shifted.append(score + 20)
print([float(rank) for rank in rankdata(shifted)])
```
