# Two related samples

What changes when the size of each paired difference is kept, as a rank, rather than only its sign?

This notebook is in Non-parametric tests, in the module Ranks and related scores. Six people rated the same item in a quiet room and a noisy room. The ratings were chosen. The comparison is the Wilcoxon signed-rank test. Rank the absolute changes, attach each rank to the sign of its change, and sum the two sides.

| person | quiet | noisy | noisy - quiet |
| --- | --- | --- | --- |
| A | 3 | 4 | +1 |
| B | 5 | 8 | +3 |
| C | 4 | 2 | -2 |
| D | 6 | 6 | 0 |
| E | 2 | 6 | +4 |
| F | 3 | 8 | +5 |

The zero is set aside. Ordered, the five absolute changes are 1, 2, 3, 4, and 5, so those are their ranks. In the order of the non-zero changes the ranks are 1, 3, 2, 4, and 5. The positive ranks are 1, 3, 4, and 5, which sum to 13. The only negative rank is 2. For a two-sided test, `wilcoxon` reports the smaller of those two sums, which is 2, and a p-value of 0.1875. With five non-zero changes, that p-value is not small. The table was chosen. It does not show that noise leaves ratings unchanged.

Six chosen pairs, one zero, and no ties among the absolute changes differ from a published signed-rank test. The null used here is that the changes are symmetric about zero. Five pairs cannot check that. A published analysis would say how zeros are handled, and it would not treat five pairs as a sample.

Teaching note. Wilcoxon, F. (1945). Individual comparisons by ranking methods. *Biometrics Bulletin, 1*(6), 80-83. https://doi.org/10.2307/3001968


## The two rank sums

The cell drops the zero, ranks the absolute changes, and adds each side. `wilcoxon` is then called on the changes themselves, zero included. Its default sets that zero aside, and its two-sided statistic is the smaller sum.


```python
# Quiet and noisy ratings for the same six people. Positive means the noisy rating is higher.
from scipy.stats import rankdata, wilcoxon

quiet = [3, 5, 4, 6, 2, 3]
noisy = [4, 8, 2, 6, 6, 8]
changes = []
for calm, loud in zip(quiet, noisy):
    changes.append(loud - calm)
print(changes)
# A zero has no sign, so it is left out of the ranking.
kept = []
for change in changes:
    if change != 0:
        kept.append(change)
abs_changes = []
for change in kept:
    abs_changes.append(abs(change))
# float() prints an ordinary number rather than a numpy scalar.
ranks = [float(rank) for rank in rankdata(abs_changes)]
print(ranks)
# Add the ranks whose changes are positive, and the rank whose change is negative.
positive_sum = 0
negative_sum = 0
for change, rank in zip(kept, ranks):
    if change > 0:
        positive_sum = positive_sum + rank
    else:
        negative_sum = negative_sum + rank
print(positive_sum)
print(negative_sum)
# Two-sided: wilcoxon reports the smaller of those two sums. The zero stays in the list and is dropped by the default.
result = wilcoxon(changes)
print(result.statistic)
print(result.pvalue)
```
