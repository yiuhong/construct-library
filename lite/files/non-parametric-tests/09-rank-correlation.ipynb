# Rank correlation

Do two rankings rise together?

This notebook is in Non-parametric tests, in the module Association and limits. Five students have a quiz score and a project score. The question is whether the ranks rise together, not a line for a mean. Spearman's rank correlation is the ordinary correlation applied to the two rank columns. Kendall's coefficient counts pairs of students who are in the same order on both rankings. It is computed once below and not developed.

The scores were chosen, with no ties.

| student | quiz | project |
| --- | --- | --- |
| A | 2 | 3 |
| B | 4 | 5 |
| C | 6 | 4 |
| D | 8 | 9 |
| E | 10 | 8 |

Quiz ranks are 1, 2, 3, 4, 5. Project ranks are 1, 3, 2, 5, 4. The rank differences are 0, -1, +1, -1, +1. The sum of their squares is 4. Spearman's coefficient on five rows is

$$
1 - \frac{6 \times 4}{5 \times (25 - 1)} = 1 - \frac{24}{120} = 0.8
$$

`spearmanr` agrees with 0.8. Of the 10 pairs of students, 8 are in the same order on both rankings and 2 are not, so Kendall's coefficient is $(8 - 2) / 10 = 0.6$. `kendalltau` prints that 0.6. The lesson uses Spearman. Kendall is only that one comparison.

Five chosen students and no ties differ from a published association. A tie would change both coefficients. The p-values assume the five rows are independent, and five rows are a thin basis for either p-value. A published analysis would separate the sampling plan from the coefficient. The printed p-values are not a finding about quizzes and projects.

Teaching note. Spearman, C. (1904). The proof and measurement of association between two things. *The American Journal of Psychology, 15*(1), 72-101. https://doi.org/10.2307/1412159


## Two rank columns

The cell ranks each column, sums the squared rank differences, and applies the five-row formula. `spearmanr` and `kendalltau` are called on the raw scores, which carry no ties, so ranking them does not have to be done first.


```python
# Five chosen students. Quiz and project, with no ties in either column.
from scipy.stats import kendalltau, rankdata, spearmanr

quiz = [2, 4, 6, 8, 10]
project = [3, 5, 4, 9, 8]
# float() prints an ordinary number rather than a numpy scalar.
quiz_ranks = [float(rank) for rank in rankdata(quiz)]
project_ranks = [float(rank) for rank in rankdata(project)]
print(quiz_ranks)
print(project_ranks)
# Square each rank difference and add those squares.
sum_of_squares = 0
for quiz_rank, project_rank in zip(quiz_ranks, project_ranks):
    difference = quiz_rank - project_rank
    sum_of_squares = sum_of_squares + difference * difference
print(sum_of_squares)
# The five-row formula. n is 5, and n squared minus 1 is 24.
spearman = 1 - (6 * sum_of_squares) / (5 * (25 - 1))
print(spearman)
spearman_result = spearmanr(quiz, project)
kendall_result = kendalltau(quiz, project)
# Round the library value to one decimal. It matches the 0.8 from the formula.
print(round(float(spearman_result.statistic), 1))
print(float(kendall_result.statistic))
```
