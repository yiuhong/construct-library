# Several related samples

How does the ranking change when the same people appear in every condition?

This notebook is in Non-parametric tests, in the module Comparing groups. Four people rated three study conditions. Each row is one person, so the columns are related. The Friedman test ranks the three ratings inside each person, then compares the rank totals of the conditions. Kruskal–Wallis ranked across different people. This ranking stays inside the row.

The ratings were chosen.

| person | list | notes | practice |
| --- | --- | --- | --- |
| A | 2 | 3 | 5 |
| B | 3 | 4 | 4 |
| C | 1 | 4 | 5 |
| D | 2 | 3 | 6 |

Inside A, C, and D the ranks are 1, 2, and 3. Inside B the two 4s share 2.5. The rank totals are 4 for list, 8.5 for notes, and 11.5 for practice. `friedmanchisquare` reports a statistic of 7.6 on these totals. The tie of two 4s is why that statistic is not the untied formula on its own. The p-value is about 0.022, from a chi-square approximation. Four people is a thin basis for that approximation, and the ratings were chosen so practice tends to sit at the top of the row.

Four chosen people, one within-person tie, and three conditions differ from a published Friedman test. A published analysis would include more people and a plan for ties. The result still does not name which conditions differ.

Teaching note. Friedman, M. (1937). The use of ranks to avoid the assumption of normality implicit in the analysis of variance. *Journal of the American Statistical Association, 32*(200), 675-701. https://doi.org/10.1080/01621459.1937.10503522


## Ranks inside each person

The cell ranks each row on its own, adds down the columns, and calls `friedmanchisquare` on the three condition lists.


```python
# Four people, three conditions. Each row is one person.
from scipy.stats import friedmanchisquare, rankdata

rows = [
    [2, 3, 5],
    [3, 4, 4],
    [1, 4, 5],
    [2, 3, 6],
]
# Rank inside the person. Person B's two 4s share a rank.
# float() prints an ordinary number rather than a numpy scalar.
ranked_rows = []
for row in rows:
    ranked_rows.append([float(rank) for rank in rankdata(row)])
print(ranked_rows)
# Three conditions. Add the rank in each column.
totals = [0, 0, 0]
for ranked in ranked_rows:
    for index, rank in enumerate(ranked):
        totals[index] = totals[index] + rank
print(totals)
# The columns, in the same order, for the Friedman call.
list_scores = [2, 3, 1, 2]
notes_scores = [3, 4, 4, 3]
practice_scores = [5, 4, 5, 6]
result = friedmanchisquare(list_scores, notes_scores, practice_scores)
print(result.statistic)
print(result.pvalue)
```
