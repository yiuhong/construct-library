# Counts in a table

What is being tested when the table holds counts of people, rather than ranks of scores?

This notebook is in Non-parametric tests, in the module Association and limits. Twenty students each picked one of two worksheet styles. The cells are counts of students. They are not ratings, and they are not ranks. A chi-square test and Fisher's exact test ask whether style and group split those counts in a way that departs from independence. Neither test ranks the students. This is not a Mann–Whitney test, and it is not a rank correlation.

The counts were chosen.

|  | style A | style B |
| --- | --- | --- |
| group 1 | 8 | 2 |
| group 2 | 3 | 7 |

Each group has 10 students. Style A was chosen 11 times and style B 9 times. If group and style were independent, the top-left expected count would be $10 \times 11 / 20 = 5.5$. The other expected counts are 4.5, 5.5, and 4.5.

`chi2_contingency` is called with the continuity correction turned off, so the statistic is the sum of squared gaps between observed and expected counts, each divided by its expected count. On a 2×2 table, scipy's default applies Yates's correction instead. That default is turned off here so the printed statistic matches the sum just described. `fisher_exact` reads the same four counts. Its statistic is an odds ratio, not a correlation.

Twenty chosen students, and an expected count of 4.5, differ from a published test of a table. There are no score-ties to handle, because these cells are not scores. The chi-square approximation is uncomfortable when an expected count is this small, which is one reason a published analysis might prefer Fisher, and might argue that choice rather than illustrate both. A published analysis would also say how the groups were formed and whether the margins were fixed. These calls do not turn the table into a rank test.


## Expected counts, then two calls for counts

The cell builds the table, checks that it sums to 20, and prints the expected counts under independence. The chi-square call turns the 2×2 correction off. Fisher's exact test is the second call on the same counts.


```python
# Counts of students, not ratings. Rows are groups. Columns are worksheet styles.
import pandas as pd
from scipy.stats import chi2_contingency, fisher_exact

counts = pd.DataFrame(
    [[8, 2], [3, 7]],
    index=["group 1", "group 2"],
    columns=["style A", "style B"],
)
print(counts)
# The number of students is the sum of the four cells.
total = 0
for value in counts.to_numpy().ravel():
    total = total + int(value)
print(total)
# correction=False skips Yates's adjustment, so the statistic matches the squared gaps.
table = counts.to_numpy()
chi = chi2_contingency(table, correction=False)
print(chi.expected_freq)
print(chi.statistic)
print(chi.pvalue)
# The same counts. The statistic here is an odds ratio.
exact = fisher_exact(table)
print(exact.statistic)
print(exact.pvalue)
```
