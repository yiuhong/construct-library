# Matplotlib figures and labels

How can I draw points on labelled axes, and join them?

A figure is the page. An axes is the region you draw on. `plt.subplots` returns both. `fig` is the figure. `ax` is the axes.

Imports and plots drew one line of scores. This lesson adds a title, draws the pairs as points, joins those points, and then grows the line one point at a time.


## Points, with labels

The two lists are paired. The first time goes with the first score. `scatter` marks each pair. `set_xlabel`, `set_ylabel`, and `set_title` write the text on the axes.

The first plot in a session can take a moment while the plotting library downloads.


```python
# One point per practice time, with a title and names on the axes. The points are not joined.
# Matplotlib draws the figure.
import matplotlib.pyplot as plt

# Practice time for each session, in minutes.
minutes = [10, 20, 30, 40, 50]
# The score from that session, in the same order.
scores = [3, 4, 6, 6, 9]

# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# One point per pair. A scatter plot does not join the points.
ax.scatter(minutes, scores)
# Name the horizontal axis, including the units.
ax.set_xlabel("Practice time (minutes)")
# Name the vertical axis.
ax.set_ylabel("Score")
# A title for the figure.
ax.set_title("Practice and score")
# Show the figure under this cell.
plt.show()
```

## Join the points

`plot` joins the pairs in list order. The points and the line are the same pairs. This cell uses `plt` from the cell above.


```python
# The same points, joined in the order of the practice times. The scatter marks each point, and the line connects them.
# The same practice times as the cell above.
minutes = [10, 20, 30, 40, 50]
# The same scores.
scores = [3, 4, 6, 6, 9]

# A new figure.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The points.
ax.scatter(minutes, scores)
# A line through those points, in list order.
ax.plot(minutes, scores)
# Name the horizontal axis.
ax.set_xlabel("Practice time (minutes)")
# Name the vertical axis.
ax.set_ylabel("Score")
# A title.
ax.set_title("Practice and score")
# Show the figure under this cell.
plt.show()
```

## A line that grows

Matplotlib can redraw the same axes, one frame at a time. Each frame adds one point and extends the line.

Run the cell, then press play under the figure. The slider moves through the five frames. The player is HTML in the page.

Redrawing the frames prints a repeated warning from this plotting build. The cell hides that warning so the player stays easy to see.


```python
# Reveal one point at a time. Each frame of the player adds the next practice time.
import warnings
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML

# Hide a warning from this version of Matplotlib, so the player can start cleanly. The warning is about an old way of passing coordinates, and it does not change the figure.
warnings.filterwarnings("ignore", message="The x parameter as float was deprecated")
warnings.filterwarnings("ignore", message="The y parameter as float was deprecated")

# Practice times, revealed from left to right.
minutes = [10, 20, 30, 40, 50]
# The matching scores.
scores = [3, 4, 6, 6, 9]

# The figure the player will draw into.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# Fix the horizontal range, so the axes do not jump as points appear.
ax.set_xlim(5, 55)
# Fix the vertical range as well.
ax.set_ylim(0, 10)
# Name the horizontal axis.
ax.set_xlabel("Practice time (minutes)")
# Name the vertical axis.
ax.set_ylabel("Score")
# A title.
ax.set_title("One point at a time")
# An empty line. The frames fill it in.
drawn = ax.plot([], [], marker="o")
# plot gives back a list. The line is the first item.
line = drawn[0]

# One frame of the animation. frame starts at 0.
def grow(frame):
    # How many points to show, from the first up to this frame.
    n = frame + 1
    # Keep only the first n times and scores.
    line.set_data(minutes[:n], scores[:n])

# Build a player under the figure. Press play, or drag the slider, to step through the frames.
player = animation.FuncAnimation(fig, grow, frames=len(minutes), interval=500, blit=False)
page = player.to_jshtml()
# Close the static figure so it is not also shown above the player.
plt.close(fig)
# The player itself, which the notebook shows under the cell.
HTML(page)
```
