# A straight-line fit

How can I fit a straight line through a few points?

You will fit a slope and an intercept to the six Monday points, and draw that line on the scatter. By the end you should be able to read the slope as the change in the weekly quiz when revision hours increase by 1, and the intercept as the line's height at 0 hours.

scipy fits one straight line. On these six points the quiz is the revision hours plus 3, so the slope is 1 and the intercept is 3. At 1 hour the line is `1 * 1 + 3`, which is 4, the first quiz score. At 0 hours the line is 3. Nobody in this Monday slice revised for 0 hours, so 3 is the line, not a recorded score. The points sit on the line. That is a property of this slice. It is a first look at a fit, not a result about revision.

The previous lesson plotted these points from the table. The line is a summary of those same points. The arithmetic stays on whole numbers so you can check the slope and the intercept by hand. The first time this session imports scipy, that library downloads. That can take a moment.


## Slope and intercept

`linregress` fits the line. `fit.slope` and `fit.intercept` are the two numbers. Revision hours are 1, 2, 3, 3, 4, and 5. The weekly quiz is 4, 5, 6, 6, 7, and 8. Each quiz is three more than the hours beside it.


```python
# The slope and the intercept of the straight line through the Monday points. The slope is the rise per extra revision hour, and the intercept is the height at 0 hours.
# SciPy's stats tools include a straight-line fit.
from scipy import stats

# Monday revision hours, in weekly quiz class order.
revision_hours = [1, 2, 3, 3, 4, 5]
# The weekly quiz on those rows.
weekly_quiz = [4, 5, 6, 6, 7, 8]
# Fit one straight line. The result holds the slope and the intercept.
fit = stats.linregress(revision_hours, weekly_quiz)
# The fitted slope. On this slice it is 1.
print(fit.slope)
# The fitted intercept. On this slice it is 3.
print(fit.intercept)
```

## The line on the points

The points are drawn with `scatter`. The fitted line is drawn with `plot`. For each revision time, the line's quiz is the slope times the hours, plus the intercept. The cell also prints the line at 0 hours, which is the intercept. No Monday student recorded 0 hours.


```python
# The Monday points, and the line's height at each revision time. The height is the slope times the hours, plus the intercept.
import matplotlib.pyplot as plt
from scipy import stats

# The same Monday revision hours as the cell above.
revision_hours = [1, 2, 3, 3, 4, 5]
# The same weekly quiz scores.
weekly_quiz = [4, 5, 6, 6, 7, 8]
# Fit the line again, so this cell stands on its own.
fit = stats.linregress(revision_hours, weekly_quiz)
# One predicted quiz per revision time, filled by the loop.
quiz_line = []
# Take each revision time in turn.
for hours in revision_hours:
    # The line's height at this revision time.
    quiz_line.append(fit.slope * hours + fit.intercept)
# The line at 0 hours. That height is the intercept, and it is not a recorded score.
print(fit.slope * 0 + fit.intercept)

# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The observed points, not joined.
ax.scatter(revision_hours, weekly_quiz)
# The fitted line through those revision times.
ax.plot(revision_hours, quiz_line)
# Name the horizontal axis.
ax.set_xlabel("Revision hours")
# Name the vertical axis.
ax.set_ylabel("Weekly quiz")
ax.set_ylim(0, 10)
# A title.
ax.set_title("A straight line through the Monday points")
# Show the figure under this cell.
plt.show()
```
