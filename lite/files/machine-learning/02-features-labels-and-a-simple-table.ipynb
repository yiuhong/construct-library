# Features, labels, and a simple table

Which columns is the model allowed to use, and which column is it asked to name?

This notebook is in Introduction to machine learning, in the module Learning from data. A feature is a column the model may use. The label is the column it is asked to predict. In the table below, hours and format are features. The score is the label.

| student | hours | format | score |
| --- | --- | --- | --- |
| A | 1 | written | 4 |
| B | 3 | spoken | 6 |
| C | 5 | written | 8 |

Hours are numeric. Format is a category, written or spoken. A model that expects numbers cannot take the word "spoken" as it stands. Two short preparations cover this table: scale the numeric column, and encode the category as columns of 0 and 1.

Three students, one numeric feature, and one category with two levels are not a questionnaire. No value is missing. The scaler is fitted on all three rows, including any row a later notebook would hold out. That is a leak if the held-out row is meant to stay unseen, and the next notebook separates the rows before the fit. The spread used by this scaler divides by the number of rows. A sample standard deviation divides by one less. With three rows those two spreads differ. The lesson needs the centre, which is the mean, and the fact that the column is rescaled.

Importing scikit-learn downloads it the first time.


## Scale the hours, and encode the format

The cell centres the hours and divides by their spread, then turns the two formats into two columns. Alphabetical order puts spoken before written.


```python
# Hours are a numeric feature. Format is a category. Score is the label, so it is not scaled or encoded here.
import numpy as np
from sklearn.preprocessing import OneHotEncoder, StandardScaler

hours = np.array([[1.0], [3.0], [5.0]])
formats = np.array([["written"], ["spoken"], ["written"]])
scores = [4, 6, 8]

# Centre the hours and divide by their spread. All three rows are used to set that scale.
scaler = StandardScaler()
scaled_hours = scaler.fit_transform(hours)
print(round(float(scaler.mean_[0]), 3))
print([round(float(value), 3) for value in scaled_hours.ravel()])

# Two format labels become two numeric columns.
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(formats)
print([str(name) for name in encoder.categories_[0]])
print(encoded.tolist())
# How many columns the category became.
print(encoded.shape[1])
# The label, left as the score the model would be asked to name.
print(scores)
```

## What those columns are

The mean of the hours is 3, so the middle student, who practised for 3 hours, sits at 0 after centring. The other two sit equally far below and above. The printed values near $-1.225$ and $1.225$ are those distances after dividing by the spread. They are not a second feature.

Spoken is the first category in alphabetical order, so its row is `[1, 0]`. Written is `[0, 1]`. The score stays a single column. It is the label, not another feature.
