# Training and testing

Why score a prediction on a row the fit did not use?

This notebook is in Introduction to machine learning, in the module Learning from data. The previous notebook let the preparation see every row. A test asks for the opposite. Some rows set the prediction. A row that was held out only scores it.

Four students, and one feature, hours:

| student | hours | score | role |
| --- | --- | --- | --- |
| A | 0 | 1 | training |
| B | 1 | 3 | training |
| C | 2 | 5 | training |
| D | 3 | 9 | test |

The first three scores sit on $\text{score} = 1 + 2 \times \text{hours}$, the line from the opening notebook. Fitted on those three alone, the line names 7 for a student with 3 hours. Student D scored 9. The absolute error on that row is 2. The training rows sit on the line, so their error is 0. That 0 is not the test.

One feature, three training rows, and one test row are an illustration of the split. The split was chosen by hand, and the test score was chosen while writing the lesson, so this is not a result to report. A full test uses more rows, and the split is fixed before anyone looks at the held-out score. Importing scikit-learn downloads it the first time.


## Fit on three rows, score the fourth

The cell fits the line on students A, B, and C, then scores student D only.


```python
# Three rows set the line. The fourth row is not used until the prediction is scored.
import numpy as np
from sklearn.linear_model import LinearRegression

train_hours = np.array([[0.0], [1.0], [2.0]])
train_scores = np.array([1.0, 3.0, 5.0])
test_hours = np.array([[3.0]])
test_score = 9.0

line = LinearRegression()
line.fit(train_hours, train_scores)
# The score the line names at 3 hours, from the training rows only.
prediction = float(line.predict(test_hours)[0])
# How far that name is from the held-out score.
absolute_error = abs(test_score - prediction)
# What the same line says about the rows it was fitted on.
training_predictions = [round(float(value), 3) for value in line.predict(train_hours)]
print(round(prediction, 3))
print(round(absolute_error, 3))
print(training_predictions)
```
