# Grouping similar observations

Can rows be grouped when no label has been written down?

This notebook is in Introduction to machine learning, in the module Unsupervised learning and limits. The earlier notebooks had a label: a score, or pass and revise. Here the table has only two test scores. Nothing in the table says which group a student belongs to. The grouping is a guess that students close to one another in the plane of the two scores belong together.

| student | vocabulary | arithmetic |
| --- | --- | --- |
| A | 1 | 1 |
| B | 1 | 2 |
| C | 2 | 1 |
| D | 8 | 8 |
| E | 9 | 8 |
| F | 8 | 9 |

Two groups are requested. Each student is placed with the nearer of two centres, and the centres are the means of the students placed with them. The low-score centre is near $(1.33, 1.33)$ and the high-score centre is near $(8.33, 8.33)$. Three students sit with each centre. The two clouds were placed far apart so the grouping is obvious. A start in a tighter cloud can come out differently; this fit fixes the start so the notebook repeats.

Six students, two scores, and a request for exactly two groups are not a cluster analysis. No label was used, and nothing here shows that two is the right number of groups, or that the groups are kinds of student. A full analysis would have to argue for the number of groups, and it would not start from two clouds drawn to be obvious. Importing scikit-learn downloads it the first time. The first plot in a session can take a moment while the plotting library downloads.


## Two centres

The cell fits the two groups, prints the lower centre and how many students sit with it, and draws the six scores with the two centres.


```python
# Group the scores without using a label. Two centres are requested.
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

scores = np.array([
    [1.0, 1.0],
    [1.0, 2.0],
    [2.0, 1.0],
    [8.0, 8.0],
    [9.0, 8.0],
    [8.0, 9.0],
])
groups = KMeans(n_clusters=2, n_init=10, random_state=0)
groups.fit(scores)
centres = groups.cluster_centers_
# The low-score group is the centre with the smaller vocabulary mean.
low = int(np.argmin(centres[:, 0]))
n_low = int(np.sum(groups.labels_ == low))
print(groups.n_clusters)
print([round(float(value), 2) for value in centres[low]])
print(n_low)

# The six students, and the two centres.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.scatter(scores[:, 0], scores[:, 1], label="Students")
ax.scatter(centres[:, 0], centres[:, 1], marker="x", label="Centres")
ax.set_xlabel("Vocabulary")
ax.set_ylabel("Arithmetic")
ax.legend()
plt.show()
```
