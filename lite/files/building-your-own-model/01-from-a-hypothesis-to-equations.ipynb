# From a hypothesis to equations

How can I turn a short hypothesis into an equation I can calculate?

A verbal hypothesis needs a quantity and a rule before it can be simulated.

Hypothesis: confidence grows with the evidence, and it cannot rise above 1.

Let the evidence be $e$ and the weight be $w$. One translation is

$$
\text{confidence} = \min(1,\, w e)
$$

The weight 0.4 and the evidence 1 give confidence 0.4. The same weight and evidence 3 give $1.2$ before the cap, and 1 after it. The cap is part of the hypothesis. Leaving it out would let confidence become 1.2, which the sentence ruled out.


```python
# Confidence is the weight times the evidence, and it is not allowed to rise above 1. A product above 1 is capped.
# The rule from the hypothesis. It takes a weight and an amount of evidence.
def confidence(weight, evidence):
    # The product, before any cap is applied.
    raw = weight * evidence
    # The cap applies only when the product is above 1.
    if raw > 1:
        return 1
    # Otherwise the product itself is the confidence.
    return raw

# A modest product, then a larger one that meets the cap.
print(confidence(0.4, 1))
print(confidence(0.4, 3))
```

## What the equation has not said

The equation says how confidence depends on one number called evidence. It does not say where the evidence came from, what the weight should be, or what happens on the next trial. Those would be further sentences, each with its own quantity.

A sentence that cannot be tied to a quantity yet is still a hypothesis. It is not yet a model you can simulate. "People are sometimes unsure" does not name the number that changes, or the rule for the change. The capped product above does both, for this one claim.
