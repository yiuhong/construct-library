# Simulate, fit, and recover

How can I check a fit against a parameter I chose?

Recovery asks whether a fit can find a parameter that you chose. Simulate data from a known parameter, fit the same rule, and compare the estimate with the known value. A small gap means the procedure can see that parameter in data of this kind. It does not, by itself, show that the rule is the right model of a person.

The rule is $y = ax$. The true slope is 3. The inputs are $x = 1, 2, 3, 4$. With no disturbance, $y = 3, 6, 9, 12$. The least-squares slope through the origin is

$$
a = \frac{\sum_i x_i y_i}{\sum_i x_i^2}
$$

That slope is 3 again. The absolute error is 0. The fit can recover a slope that was written into the data without noise.


```python
# Outcomes that sit on the true slope, then the slope a fit recovers from them.
xs = [1, 2, 3, 4]
true_slope = 3
ys = [true_slope * x for x in xs]

total_xy = 0
total_xx = 0
for i in range(4):
    total_xy = total_xy + xs[i] * ys[i]
    total_xx = total_xx + xs[i] ** 2
recovered = total_xy / total_xx
print(ys)
print(recovered, abs(recovered - true_slope))
```

## The same rule, with a disturbance

Add a listed disturbance $[1, -1, 0, 0]$ to those four outcomes. The outcomes are now 4, 5, 9, and 12. The fit uses the same formula. It no longer returns 3 exactly. The gap is the cost of this disturbance, on this list, for this rule.

A larger gap would mean that this fitting procedure, on lists like this one, cannot see the slope sharply. A gap of 0 in the noiseless cell only says that the algebra reverses when nothing has been added.


```python
# The same line, with a small disturbance added to each outcome.
xs = [1, 2, 3, 4]
true_slope = 3
disturbance = [1, -1, 0, 0]
ys = [true_slope * xs[i] + disturbance[i] for i in range(4)]

total_xy = 0
total_xx = 0
for i in range(4):
    total_xy = total_xy + xs[i] * ys[i]
    total_xx = total_xx + xs[i] ** 2
recovered = total_xy / total_xx
print(ys)
print(round(recovered, 3), round(abs(recovered - true_slope), 3))
```
