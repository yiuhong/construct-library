# Range and standard deviation

How can I describe how far the values sit from one another?

The range is the largest value minus the smallest. It uses only those two values. The variance uses every value: square each distance from the mean, then average the squares. The standard deviation is the square root of that variance, so the spread is back in the units of the measurements.

$$
\mathrm{SD}(x) = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2}
$$

This divides by $n$. It describes this list. For `[2, 4, 4, 6]` the mean is 4, the range is $6 - 2 = 4$, and the variance is $(4 + 0 + 0 + 4) / 4 = 2$.


## A short list

The cell walks the squared distances one value at a time, then prints the range, the variance, and the standard deviation.


```python
# Range uses the two ends. Variance averages every squared distance from the mean.
import math

# Four measurements.
values = [2, 4, 4, 6]
# The ends of the list.
lowest = min(values)
highest = max(values)
# The range is the gap between those ends.
value_range = highest - lowest
# The mean the distances are measured from.
mean = sum(values) / len(values)

# A running total of the squared distances.
total = 0
# One value at a time.
for value in values:
    # How far this value sits from the mean.
    distance = value - mean
    # Square it, so a value below the mean still adds a positive amount.
    total = total + distance ** 2
# The variance of this list divides by how many values there are.
variance = total / len(values)
# The square root returns the spread to the units of the measurements.
standard_deviation = math.sqrt(variance)
print(value_range)
print(variance)
print(standard_deviation)
```
