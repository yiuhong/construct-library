# Centre and spread

Where is the weekly quiz centred, and how wide is the spread?

You will calculate the mean and the median of the weekly quiz, then the range and the standard deviation of that same column. By the end you should be able to say what the recorded score of 18 does to the mean, and what the range uses.

One centre can hide a score that does not sit with the others. The spread says whether the class is tight around that centre or stretched away from it.

The previous lesson counted a questionnaire answer and left the quiz scores in the table. Those twelve scores are the column this lesson summarises.


## The mean and the median

The mean is the sum divided by how many scores there are.

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

The median is the middle once the scores are sorted. With twelve scores, average the 6th and the 7th. The scores are 4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, and 18, in the same row order as the weekly quiz class. They sum to 84, so the mean is $84/12 = 7$. Sorted, the 6th and 7th scores are both 6, so the median is 6. The recorded 18 pulls the mean above the median.


```python
# The mean uses every weekly quiz score. The median uses the two middle scores after sorting.
import matplotlib.pyplot as plt

# The weekly_quiz column, in the weekly quiz class row order.
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
# How many scores.
n = len(weekly_quiz)
# The mean is the sum divided by that count.
mean = sum(weekly_quiz) / n
# Sort a copy so the middle two scores can be read off.
ordered = sorted(weekly_quiz)
# With twelve scores, the middle pair is the 6th and the 7th, at indexes 5 and 6.
median = (ordered[5] + ordered[6]) / 2
print(mean)
print(median)

# Place each score along the class, and mark the two centres.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# One point per student, in row order.
ax.plot(range(1, n + 1), weekly_quiz, marker="o", linestyle="none", label="Weekly quiz")
# A dashed line at the mean.
ax.axhline(mean, color="black", linestyle="--", label="Mean")
# A dotted line at the median.
ax.axhline(median, color="grey", linestyle=":", label="Median")
# The horizontal axis is the place in the class list.
ax.set_xlabel("Place in the class")
# The vertical axis is the quiz score.
ax.set_ylabel("Weekly quiz")
# A key, so the points and the two centres can be told apart.
ax.legend()
# Show the figure under this cell.
plt.show()
```

## The same centres without the 18

Hold out the last row, the Thursday score recorded as 18. The other eleven scores sum to 66, and $66/11 = 6$, the same number as the median of the full list. The mean of the eleven sits on the median. Putting the 18 back moves the mean up to 7 and leaves the median at 6, because that score is not one of the two middle values.


```python
# The mean of the weekly quiz with the recorded 18 held out.
# Every score except the last row.
weekly_quiz_held_out = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8]
# The mean of those eleven scores.
held_out_mean = sum(weekly_quiz_held_out) / len(weekly_quiz_held_out)
print(held_out_mean)
```

## Range and standard deviation

The range is the largest score minus the smallest. It uses only those two scores. The variance uses every score: square each distance from the mean, then divide by $n$. Dividing by $n$ describes this class. It is not yet an estimate for a larger population. The standard deviation is the square root of that variance, so the spread is back in quiz points.

$$
\mathrm{SD}(x) = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2}
$$

On all twelve scores the range is $18 - 4 = 14$. The squared distances from the mean of 7 sum to 152, so the variance is $152/12$ and the standard deviation is the square root of that, about 3.6 quiz points. Most of that sum of squares is the one squared distance from 7 to 18.


```python
# Range uses the two ends. Variance averages every squared distance from the mean.
import math

# The weekly_quiz column, including the recorded 18.
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
# The ends of the column.
lowest = min(weekly_quiz)
highest = max(weekly_quiz)
# The range is the gap between those ends.
quiz_range = highest - lowest
# The mean the distances are measured from.
mean = sum(weekly_quiz) / len(weekly_quiz)

# A running total of the squared distances.
total = 0
# One score at a time.
for score in weekly_quiz:
    # How far this score sits from the mean.
    distance = score - mean
    # Square it, so a score below the mean still adds a positive amount.
    total = total + distance ** 2
# The variance of this class divides by how many scores there are.
variance = total / len(weekly_quiz)
# The square root returns the spread to quiz points.
standard_deviation = math.sqrt(variance)
print(quiz_range)
print(total)
print(variance)
print(standard_deviation)
```
