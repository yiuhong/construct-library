# A table of counts

How can I turn a list of labels into counts and proportions?

This notebook opens Descriptive statistics and data visualisation. The lessons here describe a set of measurements and choose a display for them. They assume you can already make a figure.

Five labels are recorded below. A count is how many times a label appears. A proportion is that count divided by how many labels there are. For these five labels, "yes" appears three times, so its proportion is $3/5 = 0.6$.


## Counts, then a bar

The loop adds one to a count for each label. The table keeps the label, the count, and the proportion together. A bar chart is a direct picture of those counts: the height of a bar is the count. The first time this session imports pandas or matplotlib, that library downloads. That can take a moment.


```python
# Count the labels, divide by how many there are, and draw one bar per label.
import matplotlib.pyplot as plt
import pandas as pd

# The five recorded labels.
labels = ["yes", "no", "yes", "yes", "no"]

# Start both counts at zero. The loop adds one for each label.
yes_count = 0
no_count = 0
# One label at a time.
for label in labels:
    # A yes adds to the yes count. The other label in this list is no.
    if label == "yes":
        yes_count = yes_count + 1
    else:
        no_count = no_count + 1

# How many labels were recorded.
n = len(labels)
# A proportion is the count divided by that length.
yes_proportion = yes_count / n
no_proportion = no_count / n
print(yes_count)
print(yes_proportion)

# One row per label, with its count and its proportion.
table = pd.DataFrame({
    "label": ["yes", "no"],
    "count": [yes_count, no_count],
    "proportion": [yes_proportion, no_proportion],
})
print(table)

# The bar height is the count. Categories, not a number line, sit on the horizontal axis.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# One bar per label. The height is the count.
ax.bar(["yes", "no"], [yes_count, no_count])
# The horizontal axis names the labels.
ax.set_xlabel("Label")
# The vertical axis is the count.
ax.set_ylabel("Count")
# Show the figure under this cell.
plt.show()
```
