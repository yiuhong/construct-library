# Errors and interrupting a run

How can I read an error, and stop a cell that is still running?

A mistake stops that cell and prints a traceback under it. Read the last line first. It names the error. The lines above it point at the code that raised it.

Run the next cell. It divides by zero, on purpose, and then stops.


```python
print(1 / 0)
```

## Reading the traceback

The last line says `ZeroDivisionError: division by zero`. That is the error. An arrow points at `print(1 / 0)`, the line that divided by zero.

Change the `0` to `2` and press Shift+Enter. The traceback is replaced by the printed result.


## Interrupting a run

While a cell is running, its prompt on the left shows `[*]`. Open the Kernel menu and choose Interrupt Kernel to ask the kernel to stop that cell. Then change the cell if you need to, and run it again.

On this site the kernel runs inside the page, on the browser's main thread. A tight loop can freeze the tab before Interrupt can run, so this lesson does not include one. A cell that raises, like the one above, finishes on its own.

If the kernel stops answering, open the Kernel menu and choose Restart Kernel. Then run the cells you still need. Restart clears the names the kernel was holding.


## What you practised

You ran a cell that divided by zero and read the last line of the traceback first. You replaced the zero and ran the cell again. You also saw Interrupt Kernel, for a cell that is still running, and Restart Kernel, which clears the names.
