# Imports and plots

How can I load a library and draw a plot?

`import` loads a library into the kernel. Later cells can use the names it defines.

This Python runs in the browser. The first time a cell needs numpy or matplotlib, those libraries download. That can take a moment, and the first plot can take a moment too. Later cells in the same session reuse what already loaded.

Run the next cell and wait for the version number.


```python
# Load NumPy and Matplotlib, then show which NumPy this session has.
import numpy as np
import matplotlib.pyplot as plt

print(np.__version__)
```

## An array

NumPy holds numbers in an array. The next cell builds three scores and prints their mean.


```python
# The mean of three scores, held together as one array.
scores = np.array([2, 4, 7])
print(scores.mean())
```

## A plot of those scores

A plot is output. It appears under the cell that drew it. This cell uses `scores` from the cell above, so run that one first if you have not.


```python
# The scores from the cell above, one point per trial.
trials = np.array([1, 2, 3])
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(trials, scores, marker="o")
ax.set_xlabel("Trial")
ax.set_ylabel("Score")
# Tick marks only at the three trials, rather than at fractional trial numbers.
ax.set_xticks(trials)
plt.show()
```
