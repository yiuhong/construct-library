# Imports and plots

How can I load a library and draw a plot?

`import` loads a library into the kernel. Later cells can use the names it defines.

This Python runs in the browser. The first time a cell needs numpy or matplotlib, those libraries download. That can take a moment, and the first plot can take a moment too. Later cells in the same session reuse what already loaded.

Run the next cell and wait for the version number.


```python
# Load NumPy and Matplotlib for the later cells, then show which NumPy this session has. The version is a check that the import worked.
# NumPy, under the short name np.
import numpy as np
# Matplotlib's plotting tools, under the short name plt.
import matplotlib.pyplot as plt

# The NumPy version this session loaded.
print(np.__version__)
```

## An array

NumPy holds numbers in an array. The next cell builds three scores and prints their mean.


```python
# The mean of three scores, held together as one array. An array can take the mean of its own values.
# Three scores in one NumPy array. This needs the import from the earlier cell.
scores = np.array([2, 4, 7])
# The mean of those three scores.
print(scores.mean())
```

## A plot of those scores

A plot is output. It appears under the cell that drew it. This cell uses `scores` from the cell above, so run that one first if you have not.


```python
# The scores from the cell above, one point per trial. The trials are numbered 1, 2, and 3.
# One trial number per score.
trials = np.array([1, 2, 3])
# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# A point for each trial, joined in trial order. scores comes from the cell above.
ax.plot(trials, scores, marker="o")
# Name the horizontal axis.
ax.set_xlabel("Trial")
# Name the vertical axis.
ax.set_ylabel("Score")
# Tick marks only at the three trials, rather than at fractional trial numbers that were never run.
ax.set_xticks(trials)
# Show the figure under this cell.
plt.show()
```
