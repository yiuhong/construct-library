# Cell order and the kernel

How can I tell what the kernel still remembers?

The kernel is the Python session for this notebook. It keeps the names you have defined, in the order you run cells. That order can differ from the order on the page.

Click the next cell and press Shift+Enter. That stores `count`. Then run the cell under it. It prints the stored number.


```python
# Store a count under the name count. The kernel keeps that name after this cell finishes, so a later cell can read it.
count = 1
```

```python
# Read the count stored by the cell above. Run that cell first. If you have not, Python has no name count yet and this cell fails.
print(count)
```

## Running out of order

Go back to the first code cell. Change `1` to `4` and press Shift+Enter. Then run the print cell again. The new number appears, because the kernel now holds the new `count`.

The page does not run earlier cells for you. If the print cell runs before any assignment, Python reports `NameError`: the name is not defined yet.


## The kernel keeps a running value

Run the next cell once. It adds one to `count` and prints the result. Run it again. The number goes up by one each time. That value lives in the kernel, not in the cell as a saved answer.


```python
# Add 1 to the count the kernel is already holding, then show the new value. Run the cell again and the count grows once more.
# Replace the stored count with one more than it was.
count = count + 1
# The value just stored.
print(count)
```

## Restart

Open the Kernel menu and choose Restart Kernel. That clears the kernel's memory. The text in the cells stays. `count` is gone until you run the assignment cell again, then the cells that use it.

What the notebook still remembers is whatever the kernel still holds.


## What you practised

You stored a name, printed it from another cell, and changed it by running the cells in an order you chose. You saw the same cell add one each time, and that Restart Kernel clears those names.
