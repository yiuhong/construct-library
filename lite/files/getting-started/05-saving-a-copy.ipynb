# Saving a copy

How can I keep a copy of a notebook I have edited in this browser?

JupyterLite keeps your edits in this browser, in its local storage (IndexedDB). Reload the page and the edit is still here. It is not written back to the repository.

Clearing this site's saved data, or opening the library in another browser, brings back the published notebook without those edits.


## Download a file you can keep

Open the File menu and choose Download. The browser saves a `.ipynb` file. That file is a copy you hold. The repository is unchanged.

You can also right-click this notebook in the file list on the left and choose Download.

Change the text in the next cell, press Shift+Enter, then download. The file includes that edit. The published lesson does not.


```python
# A short note stored as text. Edit the words between the quotes, run the cell so the print matches, and then download the notebook.
# The text this notebook will remember after you edit it.
note = "Edited in this browser."
# Show that text under the cell.
print(note)
```

## What the download is

The downloaded file is the copy to keep. Choosing Save in the File menu only updates the copy in this browser. That copy is still not written back to the repository.
