# Jupyter Notebook basics

How can I run a cell and see what it prints?

A notebook is a column of cells. This cell is Markdown: formatted text. Other cells hold code.

Click the next cell, then press Shift+Enter. That runs the Python in the cell. Printed text appears under the cell that produced it.


```python
# Run this cell with Shift+Enter. The text in quotes is printed under the cell, and the quotes themselves are not printed.
print("Printed by this cell.")
```

## Markdown and code

Double-click this text to edit the Markdown. Press Shift+Enter to render it again.

A code cell does not render prose. Shift+Enter runs its Python. Change the number below, then press Shift+Enter again. The new printed line replaces the old one, still under that same cell.


```python
# Change the stored number and run the cell again. The printed line uses whatever number is stored, so the result follows your edit.
# The number stored under the name value.
value = 2
# One more than the stored number.
print(value + 1)
```

## A plot

A plot is output too. The figure appears under the cell that drew it.

The first time you run the next cell, the plotting library may take a moment to download.


```python
# One bar for heads and one for tails. The two heights are equal, so the bars match.
# Matplotlib draws the bars.
import matplotlib.pyplot as plt

# A figure and one set of axes, sized for the lesson page.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# One bar per label. The second list is the height of each bar.
ax.bar(["Heads", "Tails"], [1, 1])
# Name the vertical axis, so the heights read as counts.
ax.set_ylabel("Count")
# Show the figure under this cell.
plt.show()
```
