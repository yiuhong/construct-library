# Jupyter Notebook basics

How can I run a cell and see what it prints?

A notebook is a column of cells. This cell is Markdown: formatted text. Other cells hold code.

Click the next cell, then press Shift+Enter. That runs the Python in the cell. Printed text appears under the cell that produced it.


```python
print("Printed by this cell.")
```

## Markdown and code

Double-click this text to edit the Markdown. Press Shift+Enter to render it again.

A code cell does not render prose. Shift+Enter runs its Python. Change the number below, then press Shift+Enter again. The new printed line replaces the old one, still under that same cell.


```python
value = 2
print(value + 1)
```

## A plot

A plot is output too. The figure appears under the cell that drew it.

The first time you run the next cell, the plotting library may take a moment to download.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.bar(["Heads", "Tails"], [1, 1])
ax.set_ylabel("Count")
plt.show()
```
