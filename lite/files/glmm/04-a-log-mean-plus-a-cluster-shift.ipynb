# A log mean plus a cluster shift

How does a cluster shift enter a model for a mean count?

You will average the practice items for each person, then read a mean count from a shift on the log scale. By the end you should be able to turn a stated shift into a positive mean count, and match that mean to people on this panel.

A count's mean has to stay positive. Adding a number on the log of the mean multiplies the mean, so the mean does not cross below zero.

The previous lesson moved a pass probability with a shift on the logit. Practice is not a pass. It is how many practice items the person completed that week, on the same six people, and the link for a mean count is the log.


## The practice column

The practice counts, week by week, are the ones in the repeated-quiz panel:

| person | week 1 | week 2 | week 3 | week 4 |
| --- | --- | --- | --- | --- |
| P1 | 1 | 0 | 1 | 2 |
| P2 | 1 | 1 | 1 | 1 |
| P3 | 1 | 2 | 2 | 3 |
| P4 | 2 | 1 | 3 | 2 |
| P5 | 2 | 3 | 3 | 4 |
| P6 | 4 | 3 | 3 | 2 |

P1's four counts sum to 4, so the mean count is 1. A zero in P1's second week is a count of no items, not a missing row. The six person means are 1, 1, 2, 2, 3, and 3. The mean of those person means is 2. These means were counted from the chosen column. They were not fitted.


```python
# One mean practice count per person, then the mean of those means.
practice = {
    "P1": [1, 0, 1, 2],
    "P2": [1, 1, 1, 1],
    "P3": [1, 2, 2, 3],
    "P4": [2, 1, 3, 2],
    "P5": [2, 3, 3, 4],
    "P6": [4, 3, 3, 2],
}
person_means = []
for person in practice:
    counts = practice[person]
    # The person's total divided by four weeks.
    person_means.append(sum(counts) / len(counts))
mean_of_people = sum(person_means) / len(person_means)
print(person_means)
print(mean_of_people)
```

## The log of the mean

The log link writes the line as the log of the mean count:

$$
\log(\text{mean}) = \text{fixed linear predictor} + \text{cluster shift}
$$

$$
\text{mean} = e^{\text{fixed linear predictor} + \text{cluster shift}}
$$

Set the fixed linear predictor to 0. The mean count is $e$ to the power of the shift. A shift of 0 gives a mean count of 1, the mean for P1 and for P2. A shift of $\log 2$ gives a mean count of 2, the mean for P3 and for P4. Adding $\log 2$ multiplies the mean by 2.

The shifts were chosen so those means can be read. They were not estimated. A full model would estimate a line for the practice counts with a count likelihood, such as the Poisson, and a distribution for the person shifts. This cell only reads the mean the log link produces. The same six people are still a small panel.

Teaching note. Nelder, J. A., & Wedderburn, R. W. M. (1972). Generalized linear models. *Journal of the Royal Statistical Society. Series A (General), 135*(3), 370-384. https://doi.org/10.2307/2344614


```python
# A chosen cluster shift on the log-mean scale. The fixed part is 0. Nothing is fitted.
import math

def predicted_mean(cluster_shift):
    # The line is the fixed part plus the shift.
    linear_predictor = 0 + cluster_shift
    # Undo the log. The result is the mean count.
    return math.exp(linear_predictor)

# A shift of 0 matches P1 and P2. log 2 matches P3 and P4.
print(predicted_mean(0))
print(predicted_mean(math.log(2)))
```
