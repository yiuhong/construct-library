# Rows are not extra clusters

Why is the number of rows a poor count of how many people I measured?

The same six scores are `[2, 4, 6]` for person A and `[5, 5, 8]` for person B. The mean of all six rows is $30/6 = 5$. Person A's mean is 4 and person B's mean is 6. The mean of those two person means is also 5, because the clusters are the same size.

The matching averages do not mean there are six people. A statement about people uses two clusters. Treating each row as its own person would overstate how many people were measured.


## Row mean, and the mean of the person means

The cell averages the rows, then averages the two person means.


```python
# The mean of the rows, and the mean of the person means. The sample size for people is the number of people.
person_a = [2, 4, 6]
person_b = [5, 5, 8]
# Stack the rows.
rows = person_a + person_b
# The mean if every row is treated as one observation.
row_mean = sum(rows) / len(rows)
# One mean per person.
mean_a = sum(person_a) / len(person_a)
mean_b = sum(person_b) / len(person_b)
# The mean of the cluster means.
mean_of_people = (mean_a + mean_b) / 2
# How many people, which is not the number of rows.
n_people = 2
print(n_people)
print(row_mean)
print(mean_of_people)
```
