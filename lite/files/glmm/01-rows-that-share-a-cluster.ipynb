# Rows that share a cluster

What is a cluster in a table of repeated measurements?

This notebook opens Advanced modelling with GLMM, in the module Clustered data. A later course can start here, after a line for the mean is already familiar.

Two people each have three trials. The table has six rows. A cluster is the set of rows that share a person. There are two clusters, not six. The scores are only an illustration of that layout.

| person | trial | score |
| --- | --- | --- |
| A | 1 | 2 |
| A | 2 | 4 |
| A | 3 | 6 |
| B | 1 | 5 |
| B | 2 | 5 |
| B | 3 | 8 |


## Count clusters and rows

The cell builds the table and counts how many distinct people it contains. The first time this session imports pandas, that library downloads. That can take a moment.


```python
# Six rows, and the person each row belongs to.
import pandas as pd

# One row per trial. The person column repeats.
table = pd.DataFrame({
    "person": ["A", "A", "A", "B", "B", "B"],
    "trial": [1, 2, 3, 1, 2, 3],
    "score": [2, 4, 6, 5, 5, 8],
})
# How many rows.
n_rows = len(table)
# How many distinct people. Each distinct person is one cluster.
n_clusters = table["person"].nunique()
print(table)
print(n_clusters)
print(n_rows)
```
