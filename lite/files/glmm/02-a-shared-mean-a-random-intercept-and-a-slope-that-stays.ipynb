# A shared mean, a random intercept, and a slope that stays

How can one equation share a mean across people, shift a person up or down, and still use one slope for week?

You will write each person's quiz mean as a shared mean plus a shift, read the gap between two people as a gap between shifts, and then add a slope for week that every person shares. By the end you should be able to build a cluster mean from the shift, and a week's stated score from that height plus the shared slope.

The shared mean is the part every person has in common. The shift belongs to one person and moves that person's line up or down. The slope is the change in the stated score when the week moves on, and in this illustration the slope does not change from person to person.

The previous lesson left a gap of 4 between P6's mean and P1's mean. That gap is what the shift carries. The week column was not in the equation yet.


## A shared mean and six shifts

The person means on the repeated-quiz panel are 3, 4, 5, 5, 6, and 7, so the shared mean is 5. The shift is the person mean minus 5, which gives $-2$, $-1$, $0$, $0$, $+1$, and $+2$. Those shifts sum to 0. They move people around the shared mean. They do not change the shared mean.

A linear mixed model writes a row as a shared part, a shift for the cluster, and a leftover:

$$
y = \text{fixed mean} + \text{cluster shift} + \text{leftover}
$$

Drop the leftover and the prediction is the cluster mean. For P1 that is $5 + (-2) = 3$. For P6 it is $5 + 2 = 7$. These shifts are the gaps of the person means from 5, written in so the split can be read. They are not the output of a fit. A fit would estimate the shifts, and it would estimate how spread out the shifts are across people. This cell does not do that.

Teaching note. Laird, N. M., & Ware, J. H. (1982). Random-effects models for longitudinal data. *Biometrics, 38*(4), 963-974. https://doi.org/10.2307/2529876


```python
# A shared mean, then a shift for each person. The shifts are the person means minus 5. Nothing is fitted.
means = [3, 4, 5, 5, 6, 7]
# The shared mean is the mean of the person means.
fixed_mean = sum(means) / len(means)
shifts = []
for mean in means:
    # How far this person's mean sits from the shared mean.
    shifts.append(mean - fixed_mean)
# The prediction drops the leftover. It is the fixed mean plus the shift.
prediction_p1 = fixed_mean + shifts[0]
prediction_p6 = fixed_mean + shifts[5]
print(fixed_mean)
print(shifts)
print(sum(shifts))
print(prediction_p1)
print(prediction_p6)
```

## The gap between two cluster means

A random intercept is this same shift added to a fixed intercept. Here the fixed intercept is the shared mean, 5. P1's cluster mean is 3 and P6's is 7. The difference is 4, and the difference of the shifts is also $2 - (-2) = 4$. The intercept is in both cluster means, so it cancels. What remains is how far the two shifts sit apart.


```python
# Each cluster mean is the fixed intercept plus that person's shift.
fixed_intercept = 5
shift_p1 = -2
shift_p6 = 2
mean_p1 = fixed_intercept + shift_p1
mean_p6 = fixed_intercept + shift_p6
# The shared intercept is in both means, so it drops out of the gap.
gap = mean_p6 - mean_p1
shift_gap = shift_p6 - shift_p1
print(mean_p1)
print(mean_p6)
print(gap)
print(shift_gap)
```

## One slope for every person

Week can enter the same kind of equation. The intercept and the slope below are chosen round values. The shift is still the person mean minus 5.

$$
\text{prediction} = (\text{intercept} + \text{shift}) + \text{slope} \times (\text{week} - 1)
$$

The intercept is 3 and the slope is 1. At week 1 the stated score is $3 + \text{shift}$. Each later week adds 1. P1's shift is $-2$, so week 1 is stated as 1 and week 4 as 4. Those two weeks match P1's quiz scores, 1 and 4. P4's shift is 0, so week 1 is stated as 3 and week 4 as 6. P4's quiz scores in those weeks are 2 and 7. The slope did not tilt to follow P4. The shift did not change the slope. It only set the height.

P1's week-2 quiz is 3 and the stated score is 2, so the leftover that week is 1. The line illustrates a shared slope and a person shift. It is not a fitted mixed model. A full analysis would estimate the intercept, the slope, and the shifts together, and it would estimate the spread of the shifts. Six people would still be a small panel for that estimate.


```python
# The shift changes the height. The slope stays shared. The intercept and the slope were chosen, not fitted.
import matplotlib.pyplot as plt

intercept = 3
slope = 1
shift_p1 = -2
shift_p4 = 0
weeks = [1, 2, 3, 4]
# The quiz scores, for the comparison with the stated line.
score_p1 = [1, 3, 4, 4]
score_p4 = [2, 5, 6, 7]

def stated(shift, week):
    # Height, then the shared step for each week after week 1.
    return (intercept + shift) + slope * (week - 1)

line_p1 = []
line_p4 = []
for week in weeks:
    line_p1.append(stated(shift_p1, week))
    line_p4.append(stated(shift_p4, week))
# Week 1 and week 4. P1 sits on the line. P4 does not.
print(line_p1[0], line_p1[3])
print(line_p4[0], line_p4[3])
# P1's week-2 quiz minus the stated score.
leftover_p1_week2 = score_p1[1] - line_p1[1]
print(leftover_p1_week2)

fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(weeks, line_p1, marker="o", label="P1 stated")
ax.plot(weeks, line_p4, marker="o", label="P4 stated")
ax.scatter(weeks, score_p1, label="P1 quiz")
ax.scatter(weeks, score_p4, label="P4 quiz")
ax.set_xticks(weeks)
# The horizontal axis is the week.
ax.set_xlabel("Week")
# The vertical axis is the quiz score, stated or observed.
ax.set_ylabel("Score")
ax.legend()
plt.show()
```
