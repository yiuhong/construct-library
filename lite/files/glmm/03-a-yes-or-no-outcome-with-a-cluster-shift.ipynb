# A yes-or-no outcome with a cluster shift

How do I summarise a pass when each person has four weeks, and how can a shift move that probability?

You will write a pass proportion for each person, then read a probability from a logit line plus a cluster shift. By the end you should be able to average person proportions without turning rows into people, and to turn a stated shift into a probability between 0 and 1.

A pass either happened or it did not, so the summary inside a person is a proportion. A proportion has to stay between 0 and 1. A straight line written on the proportion itself can leave that range. A generalised linear model keeps a line and puts a link between the line and the proportion. The link used here is the logit.

The previous lesson stated a quiz score, which can sit on an unrestricted line. On the same six people, a week is also a pass or a fail. P2 passed two of the four weeks. P3 passed three.


## A proportion for each person

The pass mark is still 5, on the same scores. P1's scores, 1, 3, 4, and 4, are all fails, so P1's proportion is 0. P2's scores, 2, 4, 5, and 5, give two passes, so the proportion is $1/2$. P3, P4, and P5 each pass three weeks, so each proportion is $3/4$. P6's scores, 5, 7, 8, and 8, are all passes, so the proportion is 1.

The mean of the six proportions is $(0 + 1/2 + 3/4 + 3/4 + 3/4 + 1) / 6 = 5/8$. Fifteen of the 24 rows are passes, and $15/24 = 5/8$ as well, because every person has four weeks. The pooled proportion still comes from six people.


```python
# A pass is a score of 5 or more. One proportion per person, then the mean of those proportions.
scores = {
    "P1": [1, 3, 4, 4],
    "P2": [2, 4, 5, 5],
    "P3": [3, 5, 6, 6],
    "P4": [2, 5, 6, 7],
    "P5": [4, 6, 7, 7],
    "P6": [5, 7, 8, 8],
}
proportions = []
passes = 0
n_rows = 0
for person in scores:
    person_passes = 0
    for score in scores[person]:
        n_rows = n_rows + 1
        # The pass mark on this panel.
        if score >= 5:
            person_passes = person_passes + 1
            passes = passes + 1
    # That person's passes divided by that person's weeks.
    proportions.append(person_passes / len(scores[person]))
mean_proportion = sum(proportions) / len(proportions)
pooled = passes / n_rows
print(proportions)
print(mean_proportion)
print(passes)
print(pooled)
```

## The logit, plus a shift

The line, written $\eta$, stays unrestricted, and the probability is read back from it:

$$
p = \frac{1}{1 + e^{-\eta}}
$$

$$
\eta = \text{fixed linear predictor} + \text{cluster shift}
$$

Set the fixed linear predictor to 0. Then $\eta$ is the cluster shift. A shift of $\log 3$ makes the odds 3, so the probability is $3/(1 + 3) = 0.75$, which is P3's pass proportion. The shift was chosen because it returns that proportion. It was not estimated from the six people. A shift of 0 leaves $\eta$ at 0 and the probability at $1/2$, which is P2's proportion. The shift moves the probability. It is not itself a probability.

A full model would estimate a linear predictor for the pass column, with a binomial likelihood, and it would estimate a distribution for the person shifts. That fit is not run here. Six people are not a population of clusters.

Teaching note. Nelder, J. A., & Wedderburn, R. W. M. (1972). Generalized linear models. *Journal of the Royal Statistical Society. Series A (General), 135*(3), 370-384. https://doi.org/10.2307/2344614


```python
# A chosen cluster shift on the logit scale. The fixed linear predictor is 0. Nothing is fitted.
import math

def probability(cluster_shift):
    # The line is the fixed part plus the shift.
    linear_predictor = 0 + cluster_shift
    # Map the line back to a probability.
    return 1 / (1 + math.exp(-linear_predictor))

# log 3 was chosen because it returns P3's proportion, 3/4.
print(probability(math.log(3)))
```
