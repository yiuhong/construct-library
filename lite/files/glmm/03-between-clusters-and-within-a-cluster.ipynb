# Between clusters and within a cluster

Which part of the variation sits between clusters, and which part sits inside one cluster?

Person A has scores `[2, 4]`. The cluster mean is 3, and the within-cluster range is $4 - 2 = 2$. Person B has scores `[6, 8]`. The cluster mean is 7, and the within-cluster range is also 2. The between-cluster gap is $7 - 3 = 4$.

A model that ignores the cluster mixes these two parts. The gap between people is not the same object as the gap between two trials from one person.


## Two gaps

The cell computes the cluster means, the gap between them, and the range inside person A. The figure draws the four scores by person. The first plot in a session can take a moment while the plotting library downloads.


```python
# A gap between cluster means, and a range inside one cluster.
import matplotlib.pyplot as plt

person_a = [2, 4]
person_b = [6, 8]
# Each cluster mean.
mean_a = sum(person_a) / len(person_a)
mean_b = sum(person_b) / len(person_b)
# Between clusters: how far the two means sit apart.
between = mean_b - mean_a
# Within person A: the gap from the smaller score to the larger.
within_a = max(person_a) - min(person_a)
print(between)
print(within_a)

# Person A near 1, person B near 2. Small offsets keep the two dots apart.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# Two scores for person A, slightly offset.
ax.scatter([0.95, 1.05], person_a, label="Person A")
# Two scores for person B.
ax.scatter([1.95, 2.05], person_b, label="Person B")
# Show the person names.
ax.set_xticks([1, 2], ["A", "B"])
# The horizontal axis is the person, which is the cluster.
ax.set_xlabel("Person")
# The vertical axis is the score.
ax.set_ylabel("Score")
# A key for the two people.
ax.legend()
# Show the figure under this cell.
plt.show()
```
