# A fixed mean plus a cluster shift

How can one equation separate a shared mean from a shift that belongs to the cluster?

This notebook is in Advanced modelling with GLMM, in the module Linear mixed models. A linear mixed model writes the outcome as a shared part plus a shift for the cluster, plus whatever is left over:

$$
y = \text{fixed mean} + \text{cluster shift} + \text{leftover}
$$

Set the fixed mean to 5, the shift for cluster A to $-1$, and the shift for cluster B to $+1$. The prediction for A is 4. The prediction for B is 6. The shifts sum to 0, so they move clusters around the shared mean rather than changing that mean.

The shifts here are chosen so the equation can be read. This lesson does not estimate them.

Teaching note. Laird, N. M., & Ware, J. H. (1982). Random-effects models for longitudinal data. *Biometrics, 38*(4), 963-974. https://doi.org/10.2307/2529876


## Add the shift to the shared mean

The cell adds each chosen shift to the fixed mean.


```python
# A shared mean, then a chosen shift for each cluster. Nothing is fitted.
fixed_mean = 5
# Chosen shifts. They sum to 0, so they sit around the shared mean.
shift_a = -1
shift_b = 1
# The prediction drops the leftover. It is the fixed mean plus the cluster shift.
prediction_a = fixed_mean + shift_a
prediction_b = fixed_mean + shift_b
print(prediction_a)
print(prediction_b)
```
