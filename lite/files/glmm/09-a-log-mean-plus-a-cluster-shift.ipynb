# A log mean plus a cluster shift

How does a cluster shift enter a model for a mean count?

A count uses a log link, as in the count notebook of Introductory statistics. The generalised mixed model adds the cluster shift on that log scale:

$$
\log(\text{mean}) = \text{fixed linear predictor} + \text{cluster shift}
$$

Set the fixed linear predictor to 0. The mean count is $e$ to the power of the shift. A shift of 0 gives a mean count of 1. A shift of $\log 2$ gives a mean count of 2. The shift multiplies the mean, because adding on the log scale is multiplying on the count scale.

This is still an illustration with a chosen shift. It is not a fitted mixed model for counts.


## Exponentiate the shift

`predicted_mean` reads the mean count from a fixed linear predictor of 0 plus the cluster shift.


```python
# A cluster shift on the log-mean scale. Exponentiating reads the mean count.
import math

def predicted_mean(cluster_shift):
    # The line is the fixed part plus the shift. The fixed part is 0 here.
    linear_predictor = 0 + cluster_shift
    # Undo the log. The result is the mean count.
    return math.exp(linear_predictor)

print(predicted_mean(0))
print(predicted_mean(math.log(2)))
```
