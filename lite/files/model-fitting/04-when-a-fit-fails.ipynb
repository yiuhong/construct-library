# When a fit fails

How can I notice a fit that has stopped on a bad answer?

A search can stop and still be a bad answer. Two failures are easy to spot on a small example.

The loss can be flat. Every $x$ below is 0, and $y = 1, 2, 3$. Then $ax$ is 0 for every slope $a$, and the sum of squared errors is $1 + 4 + 9 = 14$ for every $a$. The data do not pick out one slope. A report that quotes a single $a$ is adding a number the loss did not choose.


```python
def sse(a):
    total = 0
    for y in [1, 2, 3]:
        total = total + (y - a * 0) ** 2
    return total

print(sse(0))
print(sse(5))
```

## Outside the allowed range

A parameter can also leave the set it is allowed to live on. A probability has to sit between 0 and 1, including the ends. A search that proposes 1.2 has not returned a probability, even if a formula printed that number.

The previous lesson was the other failure: two valleys, and a start that chooses one of them. A practical check is small. Evaluate the loss at the reported answer and at a neighbour. If the neighbour is better, the search did not finish. If every neighbour ties, the parameter is not identified. If the answer is outside the allowed range, reject it.


```python
def is_probability(p):
    return 0 <= p <= 1

print(is_probability(0.75))
print(is_probability(1.2))
print(is_probability(-0.1))
```
