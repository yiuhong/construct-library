# Starting values and local minima

How can I see that a downhill search stops in the valley where it started?

Some losses have more than one valley. A search that only steps downhill stops in the valley where it started. That stop is a local minimum: the best value nearby. Another start can stop somewhere else.

$$
f(x) = (x - 1)^2 (x - 4)^2
$$

$f(x)$ is 0 at $x = 1$ and at $x = 4$. Both are minima. Between them the curve rises, so a downhill step will not cross from one valley into the other.

The next cell draws the curve. The first plot in a session can take a moment. It then starts a search at 0 and at 5.


```python
# A loss with two valleys. A downhill search stays in the valley where it starts, so the two starts can finish in different places.
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# The loss: a valley at 1 and a valley at 4.
def f(x):
    return (x - 1) ** 2 * (x - 4) ** 2

# x values from 0 to 5, for the drawing.
grid = np.linspace(0, 5, 51)
# A figure and one set of axes.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The loss curve, showing both valleys.
ax.plot(grid, f(grid))
# Name the horizontal axis.
ax.set_xlabel("x")
# Name the vertical axis.
ax.set_ylabel("loss")
# Show the curve before the two searches.
plt.show()

# Start once in the left valley's neighbourhood, and once in the right's.
for start in [0.0, 5.0]:
    # Walk downhill from this start. The lambda unwraps the one-item list minimize passes.
    found = minimize(lambda v: f(v[0]), x0=[start])
    # The start, where the search stopped, and the loss there.
    print(start, round(float(found.x[0]), 3), round(float(found.fun), 3))
```

## What to report

The search from 0 finishes at 1. The search from 5 finishes at 4. The loss is 0 at both, so the loss alone does not say which answer you have. A fit report needs the start as well as the parameter it stopped on.

A loss with one valley does not have this split. The sum of squared errors for a line through the origin, in the previous lesson, has one valley. A different start still finishes on the same slope.
