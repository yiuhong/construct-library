# What fit means, and when not to use it

What does a small gap between a model and a table say, and what does it not say?

Use the measurement model from the earlier notebook. Loadings 2, 1, and 2 imply covariances of 2, 4, and 2 for vocabulary-reading, vocabulary-spelling, and reading-spelling. Suppose the table in front of you instead has 2, 3, and 2. The gaps are 0, 1, and 0. The sum of squared gaps is 1.

A model with every loading set to 0 implies covariances of 0, 0, and 0. Against the same table, the sum of squared gaps is $2^2 + 3^2 + 2^2 = 17$.

The 1 is smaller than the 17. That is all it is. The table was chosen. There is no sample size, so the gap is not a chi-square, not an RMSEA, and not a CFI. Those indices need a likelihood and a sample. This notebook does not compute them. It adds three squared differences.

Do not treat the gap as evidence when the same numbers were used to draw the diagram and then to praise it. Do not treat a handful of people as the population the model is about. A small gap does not show that the story behind the arrows is true.


## Two sums of squared gaps

The cell subtracts each implied covariance from the chosen table and adds the squared gaps.


```python
# A close measurement model and a model with every loading at 0, scored on the same three covariances.
# Vocabulary-reading, vocabulary-spelling, reading-spelling.
observed = [2, 3, 2]
# Loadings 2, 1, and 2.
implied_close = [2, 4, 2]
# Every loading 0.
implied_zero = [0, 0, 0]

def sum_squared_gaps(implied):
    # Square each gap and add.
    total = 0
    for i in range(3):
        gap = observed[i] - implied[i]
        total = total + gap ** 2
    return total

close_gap = sum_squared_gaps(implied_close)
zero_gap = sum_squared_gaps(implied_zero)
print(close_gap)
print(zero_gap)
```
