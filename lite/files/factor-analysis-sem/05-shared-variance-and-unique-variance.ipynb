# Shared variance and unique variance

What part of a score is shared with a factor, and what part belongs to that score alone?

This notebook opens the module Factor analysis. A common-factor model writes an item as a shared part plus a part special to that item:

$$
\text{item} = \text{loading} \times \text{factor} + \text{unique}
$$

Spearman's account of a general factor used that split: a part shared across the tests, and a part special to one test. The rows below are chosen so the split can be read. The factor was written down. It was not estimated. Four scores are not a study of general intelligence, and this cell does not fit a factor model.

| student | factor | shared | unique | item |
| --- | --- | --- | --- | --- |
| A | $-1$ | $-2$ | $1$ | $-1$ |
| B | $0$ | $0$ | $-1$ | $-1$ |
| C | $0$ | $0$ | $-1$ | $-1$ |
| D | $1$ | $2$ | $1$ | $3$ |

The loading is 2, so the shared part is twice the factor. Students B and C sit at the factor's centre, so their shared part is 0. The unique parts were chosen to be uncorrelated with the factor, which is why the two sums of squares add.

Teaching note. Spearman, C. (1904). "General intelligence," objectively determined and measured. *The American Journal of Psychology, 15*(2), 201-292. https://doi.org/10.2307/1412107


## Add the shared part and the unique part

The cell rebuilds the table and adds the two sums of squares.


```python
# One item, written as twice the factor plus a unique part. The factor and the unique part are given, not estimated.
factor = [-1, 0, 0, 1]
loading = 2
unique = [1, -1, -1, 1]
# The shared part is the loading times the factor.
shared = []
item = []
for i in range(4):
    shared.append(loading * factor[i])
    item.append(shared[i] + unique[i])

def sum_of_squares(values):
    # Square each value and add.
    total = 0
    for value in values:
        total = total + value ** 2
    return total

shared_ss = sum_of_squares(shared)
unique_ss = sum_of_squares(unique)
item_ss = sum_of_squares(item)
print(shared)
print(item)
print(shared_ss)
print(unique_ss)
print(item_ss)
```
