# Many correlated variables

What does it mean when several test scores rise and fall together?

This notebook opens the course on principal components, factor analysis, and structural equation models, in the module Principal component analysis. Four students each took three short tests. The scores are chosen so the pattern is easy to see.

| student | vocabulary | reading | spelling |
| --- | --- | --- | --- |
| A | 2 | 1 | 2 |
| B | 4 | 3 | 2 |
| C | 6 | 5 | 8 |
| D | 8 | 7 | 8 |

Vocabulary and reading rise in step. Spelling rises too, with a tie in the middle. A positive sum of paired deviations means the two tests move the same way: subtract each test's mean, multiply each student's pair, and add.

Four students are not a questionnaire study. A real one would include many more people. These sums describe this table only. No correlation test is computed.


## Paired deviations

The cell rebuilds the table as three lists, in the same student order, and adds the paired deviations for each pair of tests.


```python
# Three short tests for four students. Each list is one test, in the same student order.
vocab = [2, 4, 6, 8]
reading = [1, 3, 5, 7]
spelling = [2, 2, 8, 8]

def mean(scores):
    # Add the scores, then divide by how many there are.
    total = 0
    for score in scores:
        total = total + score
    return total / len(scores)

def paired_sum(left, right):
    # Subtract each mean, multiply the pair, and add those products.
    left_mean = mean(left)
    right_mean = mean(right)
    total = 0
    for i in range(len(left)):
        total = total + (left[i] - left_mean) * (right[i] - right_mean)
    return total

# Vocabulary with reading, then vocabulary with spelling, then reading with spelling.
paired_reading = paired_sum(vocab, reading)
paired_spelling = paired_sum(vocab, spelling)
paired_reading_spelling = paired_sum(reading, spelling)
# How many tests, not how many students.
n_variables = len([vocab, reading, spelling])
print(paired_reading)
print(paired_spelling)
print(paired_reading_spelling)
print(n_variables)
```
