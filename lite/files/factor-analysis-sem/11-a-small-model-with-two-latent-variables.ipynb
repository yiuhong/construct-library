# A small model with two latent variables

How can two latent variables be tied together?

Verbal has two items, vocabulary and reading, with loadings 2 and 1. Number has two items, arithmetic and tables, with loadings 2 and 1. A path of 0.5 runs from verbal to number. No item loads on both latents.

The implied covariance of a verbal item and a number item multiplies three numbers: the verbal loading, the path, and the number loading. Vocabulary and arithmetic imply $2 \times 0.5 \times 2 = 2$. Reading and tables imply $1 \times 0.5 \times 1 = 0.5$.

The cell then searches for the path only. The loadings stay fixed, and the target covariance of vocabulary and arithmetic is 2. From a start of 0, the search returns 0.5. A full model would estimate the loadings as well. Four items and a chosen target do not stand in for that, and this search reports no fit index beyond the gap in that one covariance.


## The product, then a search for the path

The loadings stay at the chosen values. The loss is the squared gap between the implied covariance and 2.


```python
# Two latents and one path. Only the path is searched. The loadings stay fixed.
from scipy.optimize import minimize

loading_vocab = 2
loading_reading = 1
loading_arithmetic = 2
loading_tables = 1
path = 0.5
# Covariance of one verbal item and one number item.
cross_cov = loading_vocab * path * loading_arithmetic
reading_tables = loading_reading * path * loading_tables
print(cross_cov)
print(reading_tables)

def path_loss(value):
    # The path is the first entry. The loadings stay at 2 and 2. The target covariance is 2.
    implied = loading_vocab * value[0] * loading_arithmetic
    gap = implied - 2
    return gap ** 2

# Start the search at a path of 0.
result = minimize(path_loss, x0=[0.0])
fitted_path = float(result.x[0])
print(round(fitted_path, 3))
print(round(float(result.fun), 6))
```
