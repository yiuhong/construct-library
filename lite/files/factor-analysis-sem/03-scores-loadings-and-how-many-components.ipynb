# Scores, loadings, and how many components

What is a score, what is a loading, and how do you decide how many components to keep?

Six students and three tests. The table is built from two known directions, so the shares of spread come out as round numbers. A real set of tests would not be built that way.

A loading is the weight of one test on a component. The loadings of one component are a direction, and this cell scales that direction to length 1. A score is where one student sits on that component: the centred scores, weighted by the loadings.

The sign of a component is a choice. The cell flips it so vocabulary's loading is positive. The opposite sign would describe the same direction.

The squared singular values are the sums of squares along each component. Here they are 72, 8, and 0. The zero is there because the table used only two directions. Six students and three tests are a sketch. No rule for keeping components in a published analysis is applied.


## Loadings, scores, and shares

The cell centres the table, runs the decomposition, and prints the loadings, one student's scores, and the share on the first component.


```python
# Six students and three tests, built from two directions. The decomposition reads the loadings and the shares.
import numpy as np

# Two constructed directions. Each is already centred, and they are uncorrelated with each other.
first = [-2, -1, -1, 1, 1, 2]
second = [-1, 1, 0, 0, 1, -1]
vocab = []
reading = []
spelling = []
for i in range(6):
    # Vocabulary uses only the first direction.
    vocab.append(2 * first[i])
    # Reading and spelling share the first direction and split the second in opposite ways.
    reading.append(first[i] + second[i])
    spelling.append(first[i] - second[i])

# One row per student, one column per test.
table = np.column_stack([vocab, reading, spelling])
# Centre each test. These columns are already centred. The subtraction is the step a real table needs.
centred = table - table.mean(axis=0)
# s ** 2 is the sum of squares along each component. The rows of vt are the directions.
u, s, vt = np.linalg.svd(centred, full_matrices=False)
sums = s ** 2
print([round(float(value), 1) for value in sums])

# Weights of the three tests on the first component. Flip the sign so vocabulary's loading is positive.
loadings = vt[0].copy()
if loadings[0] < 0:
    loadings = -loadings
print([round(float(value), 3) for value in loadings])
# Where each student sits on that component.
scores = centred @ loadings
print([round(float(value), 3) for value in scores])

# The second component contrasts reading with spelling. Flip it so reading's loading is positive.
second_loadings = vt[1].copy()
if second_loadings[1] < 0:
    second_loadings = -second_loadings
# Round for display. A residue that rounds to zero is shown as 0.0, not -0.0.
print([0.0 if round(float(value), 3) == 0 else round(float(value), 3) for value in second_loadings])

# Share of the total sum of squares on the first component, and the sum of squares on the second.
share_first = float(sums[0] / sums.sum())
ss_second = float(sums[1])
print(round(share_first, 1))
print(round(ss_second, 1))
```
