# Path diagrams and indirect effects

How can one variable reach another through a third?

This notebook opens the module Structural equation models. A path diagram writes the arrows. Practice feeds skill, and skill feeds the score. Practice also has a direct arrow to the score. The coefficients are chosen.

```
practice --2--> skill --3--> score
practice --1--------------> score
```

Along one chain, the indirect effect is the product of the coefficients. Wright's path rules use that product. Here the indirect effect is $2 \times 3 = 6$. The direct path is 1, so the total effect is $6 + 1 = 7$.

Three variables and three chosen coefficients are not an estimated structural model. There is no latent variable yet, and no fit index.

Teaching note. Wright, S. (1934). The method of path coefficients. *The Annals of Mathematical Statistics, 5*(3), 161-215. https://doi.org/10.1214/aoms/1177732676


## Multiply along the chain, then add the direct path

The cell uses the three coefficients from the diagram.


````python
# An indirect effect is the product of the coefficients along the chain. The total adds the direct path.
practice_to_skill = 2
skill_to_score = 3
direct = 1
# The chain practice -> skill -> score.
indirect_effect = practice_to_skill * skill_to_score
# The chain plus the direct arrow.
total_effect = indirect_effect + direct
print(indirect_effect)
print(total_effect)
````
