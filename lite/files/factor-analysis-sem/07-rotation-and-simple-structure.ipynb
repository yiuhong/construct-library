# Rotation and simple structure

Why can two factors be turned, and what pattern is the turn aiming at?

An orthogonal turn mixes two factors and keeps them at right angles. The reproduced covariance of two items stays the same, because the turn does not change the products of the loadings once both factors are included. What changes is which factor an item lines up with.

The aim is a simple pattern, called simple structure in this lesson: each item lines up with one factor and has a loading of 0 on the other. The lesson does not reproduce a published rotation criterion.

Four items start a quarter-turn away from that pattern. Vocabulary and reading belong with the first factor. Arithmetic and tables belong with the second. Their loadings, before the turn back, are mixtures. A loop tries every whole-number angle from $-90$ degrees to $90$ degrees and keeps the angle that makes the smaller loading on each item as small as it can. The score in that loop is a teaching stand-in. Oblique rotation, which lets the factors correlate, is not tried.

Four chosen items and one angle are not a rotation of a fitted questionnaire.


## One turn, and a covariance that stays

The cell turns the starting loadings by $-45$ degrees and prints both patterns, rounded to three decimals. It also prints the reproduced covariance of vocabulary and reading before and after the turn.


```python
# Turn two factors by one angle. The reproduced covariance of two items should stay put.
import math

# Square root of one half, so a 45 degree mixture is exact.
half = math.sqrt(2) / 2
# Rows are vocabulary, reading, arithmetic, and tables. Columns are the two factors, before the turn back.
start = [
    [0.8 * half, -0.8 * half],
    [0.6 * half, -0.6 * half],
    [0.8 * half, 0.8 * half],
    [0.6 * half, 0.6 * half],
]

def rotate(loadings, degrees):
    # Mix the two factors by this angle. The new factors stay at right angles.
    theta = math.radians(degrees)
    cosine = math.cos(theta)
    sine = math.sin(theta)
    turned = []
    for row in loadings:
        first = row[0] * cosine + row[1] * sine
        second = -row[0] * sine + row[1] * cosine
        turned.append([first, second])
    return turned

def reproduced(loadings, i, j):
    # Covariance of two items from the products of their loadings on the two factors.
    return loadings[i][0] * loadings[j][0] + loadings[i][1] * loadings[j][1]

def shown(row):
    # Round for display. A residue that rounds to zero is shown as 0.0, not -0.0.
    tidy = []
    for value in row:
        rounded = round(value, 3)
        if rounded == 0:
            rounded = 0.0
        tidy.append(rounded)
    return tidy

turned_once = rotate(start, -45)
for row in turned_once:
    print(shown(row))
print(round(reproduced(start, 0, 1), 3))
print(round(reproduced(turned_once, 0, 1), 3))
```

## Search over angles

The score is the sum, over items, of the square of the loading that is smaller in absolute value. Zero means every item uses only one factor. The loop keeps the angle with the smallest score.


```python
# Try one angle at a time and keep the turn that lines each item up with one factor.
import math

half = math.sqrt(2) / 2
start = [
    [0.8 * half, -0.8 * half],
    [0.6 * half, -0.6 * half],
    [0.8 * half, 0.8 * half],
    [0.6 * half, 0.6 * half],
]

def rotate(loadings, degrees):
    # The same orthogonal mix as the cell above.
    theta = math.radians(degrees)
    cosine = math.cos(theta)
    sine = math.sin(theta)
    turned = []
    for row in loadings:
        first = row[0] * cosine + row[1] * sine
        second = -row[0] * sine + row[1] * cosine
        turned.append([first, second])
    return turned

def clutter(loadings):
    # How much each item still sits on both factors.
    total = 0
    for row in loadings:
        smaller = min(abs(row[0]), abs(row[1]))
        total = total + smaller ** 2
    return total

def reproduced(loadings, i, j):
    return loadings[i][0] * loadings[j][0] + loadings[i][1] * loadings[j][1]

# Every whole-number angle from -90 to 90, inclusive.
best_degrees = None
best_score = None
for degrees in range(-90, 91):
    score = clutter(rotate(start, degrees))
    if best_score is None or score < best_score:
        best_score = score
        best_degrees = degrees

turned = rotate(start, best_degrees)
# Arithmetic is the third item. Its loading on the first factor, after the chosen turn.
arithmetic_on_first = turned[2][0]
# Vocabulary and reading, after the turn.
cov_after = reproduced(turned, 0, 1)
def shown(row):
    # Round for display. A residue that rounds to zero is shown as 0.0, not -0.0.
    tidy = []
    for value in row:
        rounded = round(value, 3)
        if rounded == 0:
            rounded = 0.0
        tidy.append(rounded)
    return tidy

print(best_degrees)
print(shown(turned[2]))
print(0.0 if round(arithmetic_on_first, 6) == 0 else round(arithmetic_on_first, 6))
print(round(cov_after, 3))
```
