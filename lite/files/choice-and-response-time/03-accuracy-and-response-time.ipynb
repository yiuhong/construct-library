# Accuracy and response time

How can I get accuracy and response time from the same accumulator?

The accumulator from the previous lesson makes two predictions. The continuous random-walk account is Ratcliff's (Ratcliff, 1978). Here the noise is first set to 0, so the arithmetic is exact.

With noise at 0 and a positive drift, every step walks toward the upper bound, and the path cannot hit the lower bound. If the upper bound is the correct response, accuracy is 1. The response time is the number of steps until the total reaches the bound.

Bound 1 and drift 0.5: the total is 0.5, then 1. Two steps. Bound 1 and drift 0.25: four steps. A stronger drift toward the correct bound finishes sooner.


```python
def noiseless_steps(drift, bound):
    evidence = 0.0
    steps = 0
    while evidence < bound:
        evidence = evidence + drift
        steps = steps + 1
    return steps

print(noiseless_steps(0.5, 1))
print(noiseless_steps(0.25, 1))
```

## The same noise, two drifts

With noise, a weak drift can be carried into the wrong bound. The correct bound is the upper one, at 1, and the wrong bound is $-1$. Both paths below use the noise list $[-0.8, -0.8, 0.4]$.

Drift 0.2 reaches $-1.2$ on step 2. That is the wrong bound. Drift 0.9 is still positive after the same two shocks, and it reaches the upper bound on step 3. A stronger drift is not only faster when it is correct. It is also harder for a run of unfavourable noise to overturn.


```python
def run(drift, noise):
    evidence = 0.0
    for step, shock in enumerate(noise, start=1):
        evidence = evidence + drift + shock
        if evidence >= 1 or evidence <= -1:
            return step, round(evidence, 3)
    return None, round(evidence, 3)

print(run(0.2, [-0.8, -0.8, 0.4]))
print(run(0.9, [-0.8, -0.8, 0.4]))
```
