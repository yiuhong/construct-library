# An accumulator model

How can I make a choice by adding evidence until it hits a bound?

A choice can be a running total of evidence. Start at 0. Each step adds a drift, a steady push toward one response, and a noise term. Stop at the first step where the total reaches an upper bound or a lower bound. The bound that is hit is the choice. The number of steps is the response time.

Hitting a bound with a noisy running total is the decision rule in Ratcliff's continuous random-walk model, which links a choice to the time it takes (Ratcliff, 1978). The walk below is a short discrete version with a fixed list of noise values. It is not a fit to a set of response times.

Ratcliff, R. (1978). A theory of memory retrieval. *Psychological Review, 85*(2), 59–108.

Drift 0.4. Noise $[0.2, -0.3, 0.5, 0.1]$. Bounds at 1 and at $-1$.


```python
# Add drift and noise until the running total reaches a bound. The bounds are 1 and -1, and crossing either one stops the walk.
# The steady push added on every step.
drift = 0.4
# The disturbance on each step, already listed out.
noise = [0.2, -0.3, 0.5, 0.1]
# The running total starts at 0.
evidence = 0.0
# Step numbers begin at 1, paired with each disturbance.
for step, shock in enumerate(noise, start=1):
    # Add this step's drift and its disturbance.
    evidence = evidence + drift + shock
    # The step number and the total so far.
    print(step, round(evidence, 3))
    # Stop if the total has reached either bound.
    if evidence >= 1 or evidence <= -1:
        # The stopping step, and the total on that step.
        print("stop", step, round(evidence, 3))
        # Leave the loop. Later disturbances are not used.
        break
```

## Which bound, and when

The total is 0.6, then 0.7, then 1.6. It crosses the upper bound on step 3, so the choice is the upper response and the response time is 3 steps. The stored total on that step is 1.6, which is past the bound. The bound is the stopping rule. It is not a claim that the evidence landed exactly on 1.

The next cell draws that same path, one step at a time. The dashed lines are the two bounds. Run the cell. A player is drawn under the figure. Press play, or drag the slider.

A larger drift, or a closer bound, would finish in fewer steps. The next lesson uses that to predict accuracy and response time. The same noise list can also be pulled the other way if the drift is weak. One path is one trial. A model of a person is a collection of such trials.


```python
# The same path, drawn one step at a time. The player adds one step per frame, and the dashed lines are the two bounds.
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# The same drift as the cell above.
drift = 0.4
# The same disturbances.
noise = [0.2, -0.3, 0.5, 0.1]
# Evidence after each step, starting from 0 before any step.
path = [0.0]
# Build the path, and stop once a bound is crossed.
for shock in noise:
    path.append(path[-1] + drift + shock)
    if path[-1] >= 1 or path[-1] <= -1:
        break

# The figure the player draws into.
fig, ax = plt.subplots(figsize=(6.4, 3.4))
# The upper bound.
ax.axhline(1, color="0.5", linestyle="--")
# The lower bound.
ax.axhline(-1, color="0.5", linestyle="--")
# Fix the axes so they do not jump between frames.
ax.set_xlim(-0.2, 3.2)
ax.set_ylim(-1.6, 2.0)
# Name the horizontal axis.
ax.set_xlabel("Step")
# Name the vertical axis.
ax.set_ylabel("Evidence")
# An empty line. Each frame extends it.
line, = ax.plot([], [], marker="o")

# Show the path from the start up to this frame.
def draw(frame):
    # Steps 0 through this frame, and the evidence at each.
    line.set_data(list(range(frame + 1)), path[: frame + 1])
    # Name the frame by its step.
    ax.set_title("Step " + str(frame))
    return (line,)

# Build a player under the figure. Press play, or drag the slider, to step through the frames.
anim = FuncAnimation(fig, draw, frames=len(path), interval=600, repeat=False)
# Close the static figure so only the player is shown.
plt.close(fig)
# The player under the cell.
HTML(anim.to_jshtml())
```
