# Subplots

How do two related plots share one figure?

This notebook is in Matplotlib, in the module Figures and axes. `plt.subplots(1, 2)` builds one page and two axes, side by side. The call is one row and two columns. Matplotlib returns a flat array of length 2 when there is only one row, so `axes[0]` is the left panel and `axes[1]` is the right panel.

The two panels here do not share a vertical scale. A quiz score and a number of hours are different units. A shared scale would make those units look alike.


## Two panels

One student, four weeks. The left panel is the quiz. The right panel is hours of practice. The cell prints the shape of the axes array. The first plot in a session can take a moment while the plotting library downloads.


```python
# One figure, two axes. The quiz and the practice hours stay on their own scales.
import matplotlib.pyplot as plt

# The week number.
weeks = [1, 2, 3, 4]
# The quiz score that week.
quiz = [5, 6, 8, 7]
# Hours of practice that week. Not the same unit as the quiz.
practice = [2, 3, 3, 5]

# One row, two columns. The page is wide enough for both labels.
fig, axes = plt.subplots(1, 2, figsize=(8, 3.2), layout="constrained")
# The shape is (2,): one row is returned as a flat list of axes.
print(axes.shape)

# The left axes.
axes[0].plot(weeks, quiz, marker="o", color="#1c1c1c")
axes[0].set_xlabel("Week")
axes[0].set_ylabel("Quiz score")
axes[0].set_title("Quiz")
# Keep the score axis starting at zero.
axes[0].set_ylim(0, 10)

# The right axes. Its vertical label is hours, not score.
axes[1].plot(weeks, practice, marker="o", color="#1c1c1c")
axes[1].set_xlabel("Week")
axes[1].set_ylabel("Practice (hours)")
axes[1].set_title("Practice")
axes[1].set_ylim(0, 8)

# Show both panels under this cell.
plt.show()
```

## The same two series, one above the other

`plt.subplots(2, 1)` stacks the panels. Again there is only one column, so the array is flat: `axes[0]` is the top panel and `axes[1]` is the bottom panel. This cell uses `plt` from the cell above.


```python
# The same two series, stacked. Top is the quiz. Bottom is the practice.
weeks = [1, 2, 3, 4]
quiz = [5, 6, 8, 7]
practice = [2, 3, 3, 5]

# Two rows, one column.
fig, axes = plt.subplots(2, 1, figsize=(6.2, 4.6), layout="constrained")
# Still a flat list, because there is only one column.
print(axes.shape)

axes[0].plot(weeks, quiz, marker="o", color="#1c1c1c")
axes[0].set_ylabel("Quiz score")
axes[0].set_title("Quiz")
axes[0].set_ylim(0, 10)

axes[1].plot(weeks, practice, marker="o", color="#1c1c1c")
axes[1].set_xlabel("Week")
axes[1].set_ylabel("Practice (hours)")
axes[1].set_title("Practice")
axes[1].set_ylim(0, 8)

plt.show()
```
