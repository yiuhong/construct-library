# Histograms and KDE

What does a bin count, and what is left out of a smooth curve on the same scores?

You will count the twelve weekly quiz scores into bins, narrow those bins, and then draw a smooth curve on the same scores. By the end you should be able to say what the height of a histogram bar counts, and what this curve simplifies.

A histogram's height is how many scores fell in the bin. The width of the bin is a choice. A wide bin can hide a pile that a narrower bin shows. A smooth curve is a different picture of the same list. On twelve scores it is only a sketch.

The previous lesson counted a category. The weekly quiz is a number, so the bars here cut that number into bins. The scores, in weekly quiz class order, are 4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, and 18.


## Bins of width 2

The edges are 4, 6, 8, 10, 12, 14, 16, 18, and 20. A score that lands on an edge goes into the bin on the right, except the last edge, which is included. So 6 belongs to the bin that starts at 6, 8 belongs to the bin that starts at 8, and the recorded 18 belongs to the last bin.

The counts are 4, 5, 2, then four empty bins, then 1. The empty bins are the gap between 10 and 18. The height is the count, including a height of zero. The cell prints those eight counts.

The first plot in a session can take a moment while the plotting library downloads.


```python
# Count the twelve weekly quiz scores into bins of width 2, then draw one bar per bin.
import matplotlib.pyplot as plt

# Twelve weekly quiz scores, in class order. One was recorded as 18.
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
# Bin edges. The last edge is included. The others are the start of a bin.
edges = [4, 6, 8, 10, 12, 14, 16, 18, 20]

fig, ax = plt.subplots(figsize=(6.6, 3.6), layout="constrained")
# hist returns the counts, the edges, and the bars. Keep the counts.
counts, drawn_edges, bars = ax.hist(
    weekly_quiz, bins=edges, color="#4a4a4a", edgecolor="white"
)
ax.set_xlabel("Weekly quiz")
ax.set_ylabel("Count")
ax.set_title("Bins of width 2")
# The eight bin counts, as ordinary integers.
print([int(count) for count in counts])
plt.show()
```

## Bins of width 1

The same scores, with a bin for each integer from 4 up to 9. The three 6s become their own bar. In the wide bins they were part of a bar of five, together with the two 7s.

These edges stop at 9. The recorded 18 sits above that, so it is not in any of these bars. The wide bins were the picture that included it. Eleven scores fall inside these edges: two 4s, two 5s, three 6s, two 7s, and two 8s. A histogram of a short list is a picture of that list and of the edges you chose. This cell uses `plt` from the cell above.


```python
# The same scores, one bin per integer from 4 up to 9. The recorded 18 is outside these edges.
weekly_quiz = [4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18]
# Edges at every integer from 4 to 9. A score of 18 is above the last edge.
edges = [4, 5, 6, 7, 8, 9]

fig, ax = plt.subplots(figsize=(6.6, 3.6), layout="constrained")
counts, drawn_edges, bars = ax.hist(
    weekly_quiz, bins=edges, color="#4a4a4a", edgecolor="white"
)
ax.set_xlabel("Weekly quiz")
ax.set_ylabel("Count")
ax.set_title("Bins of width 1, up to 9")
print([int(count) for count in counts])
# How many of the twelve scores fell inside these edges.
print(sum(int(count) for count in counts))
plt.show()
```

## A smooth curve on the same scores

This is a picture of twelve chosen scores. It is not a density you would report for a class.

What is simplified:

- Small list. Twelve scores, one of them recorded as 18. The curve looks smoother than a list this short can support.
- Toy smoother. `gaussian_kde` puts a normal bump on every score and adds the bumps, at the library's default width. For one measurement that width's factor is the sample size raised to the power `-1/5`. For 12 scores, `12 ** (-1/5)` is the factor the cell prints beside `kde.factor`. The two printed numbers agree to many decimal places. A difference in the final digit is the arithmetic, not a different rule. The width was not chosen by comparing several curves, and it was not checked against the histogram before being kept.
- JupyterLite limits. This page can run `gaussian_kde`. It does not search bandwidths, correct the edges of a quiz marked out of 10, or refit after a decision about the recorded 18.

A fuller density estimate would still need more scores, a bandwidth you chose and checked, the histogram beside the curve, and care at the edges when a score cannot fall outside a range. The normal bumps can spill below 0 or past the mark of 10, and they also sit on the recorded 18.

The histogram beside the curve uses the width-2 edges from the first cell, with `density=True`. The bar areas then add to about 1, so the height is no longer the count. That rescaling lets the bars and the curve share a vertical axis. It is there for the picture, not because the counts stopped mattering.

The first time this session imports scipy, that library downloads. That can take a moment.


```python
# A normal bump on each of the twelve scores, added up, using the default width.
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import gaussian_kde

# The same twelve scores as the histogram. One row of one measurement.
weekly_quiz = np.array([4, 5, 6, 6, 7, 8, 4, 5, 6, 7, 8, 18], dtype=float)
# The default estimate. For one measurement, factor is n to the power -1/5.
kde = gaussian_kde(weekly_quiz)
# The factor the library chose, and the power it should match.
print(kde.factor)
print(12 ** (-1 / 5))

# A grid to draw the curve on. The bumps are not limited to the quiz's mark of 10.
xs = np.linspace(0, 22, 200)
ys = kde(xs)

fig, ax = plt.subplots(figsize=(6.8, 3.8), layout="constrained")
# density=True rescales the bars so their area is about 1. The height is not a count.
ax.hist(
    weekly_quiz,
    bins=[4, 6, 8, 10, 12, 14, 16, 18, 20],
    density=True,
    color="#d0d0d0",
    edgecolor="white",
)
# The sum of the bumps, drawn on the grid.
ax.plot(xs, ys, color="#1c1c1c")
ax.set_xlabel("Weekly quiz")
ax.set_ylabel("Density")
ax.set_title("A smooth curve on twelve scores")
plt.show()
```
