# Figures and axes

What belongs to the figure, and what belongs to the axes?

This notebook is in Matplotlib, in the module Figures and axes. Plotting and packages already drew one scatter, joined the points, and grew the line. Those three notebooks stay the short introduction. This course stays with the figure and goes further.

A figure is the page. An axes is the region you draw on. Matplotlib uses the word axes for one region as well as for several. `plt.subplots` returns both objects. `fig` is the figure. `ax` is the axes. The line, the labels, and the limits belong to the axes. The page that holds the axes is the figure.


## One page, one region

Five weekly quiz scores. The first plot in a session can take a moment while the plotting library downloads.

The cell prints two names, then how many axes the figure holds, then whether `fig.axes[0]` is the same object as `ax`.


```python
# One figure and the single axes on it. The line is drawn on the axes, not on the figure.
import matplotlib.pyplot as plt

# The week of the short quiz.
weeks = [1, 2, 3, 4, 5]
# The score that week, in the same order.
scores = [6, 7, 7, 8, 9]

# subplots with no arguments builds one page and one axes.
fig, ax = plt.subplots()
# Join the scores in week order, and mark each week.
# The colour argument is spelled color, which is Matplotlib's spelling.
ax.plot(weeks, scores, marker="o", color="#1c1c1c")
# The horizontal label belongs to the axes.
ax.set_xlabel("Week")
# The vertical label belongs to the axes.
ax.set_ylabel("Quiz score")
# The title belongs to the axes as well.
ax.set_title("Weekly quiz")
# The class name of the page.
print(type(fig).__name__)
# The class name of the region.
print(type(ax).__name__)
# How many regions the page holds.
print(len(fig.axes))
# The first region on the page is the axes subplots returned.
print(fig.axes[0] is ax)
# Show the figure under this cell.
plt.show()
```

## Limits belong to the axes

The same scores, with the axes extended so the line does not fill the frame. `set_xlim` and `set_ylim` change the region. They do not add a second page. This cell uses `plt` from the cell above.


```python
# The same quiz, with the axes extended past the first and last points.
# The week numbers again.
weeks = [1, 2, 3, 4, 5]
# The same scores.
scores = [6, 7, 7, 8, 9]

# A new page and a new axes.
fig, ax = plt.subplots()
# The same line as the cell above.
ax.plot(weeks, scores, marker="o", color="#1c1c1c")
# Show one week of empty space on either side.
ax.set_xlim(0, 6)
# Start the score axis at zero, so the height is the score.
ax.set_ylim(0, 10)
# Name the axes.
ax.set_xlabel("Week")
ax.set_ylabel("Quiz score")
ax.set_title("The same quiz, with the limits set")
# Show the figure under this cell.
plt.show()
```
