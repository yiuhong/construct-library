# Saving a figure

What sets the pixel size of a saved figure?

This notebook is in Matplotlib, in the module Polish and export. `fig.savefig` writes the figure to a file. The figure size is in inches. `dpi` is the dots per inch. Without `bbox_inches="tight"`, the pixel width is the width in inches times `dpi`, and the pixel height is the height in inches times `dpi`.

This figure is 6 inches wide and 3 inches tall. At `dpi=100` the PNG is 600 pixels wide. At `dpi=150` a second PNG of the same figure is 900 pixels wide. The inches do not change. The pixel count does.

`bbox_inches="tight"` crops the spare margin. The pixel count is then no longer just inches times `dpi`. The files in this notebook do not use that argument.

The files are written in the notebook's folder in this browser. Open the file browser and look in `matplotlib` for `questionnaire-counts.png` and `questionnaire-counts-fine.png`. Download from there if you want a copy. Nothing is uploaded.


## Write two PNG files

The bars are the same three response counts as in Bar charts. The first plot in a session can take a moment while the plotting library downloads. Saving does not replace `plt.show`: the cell still draws the figure under the cell, and it also writes the files.


```python
# Save the same figure twice. The inches stay put. The dpi changes the pixel count.
import matplotlib.pyplot as plt

# Yes, Not sure, and No.
labels = ["Yes", "Not sure", "No"]
counts = [8, 3, 4]

# Six inches wide and three inches tall.
fig, ax = plt.subplots(figsize=(6, 3), layout="constrained")
ax.bar(labels, counts, color="#4a4a4a")
ax.set_ylim(0, 10)
ax.set_xlabel("Response")
ax.set_ylabel("Count")
ax.set_title("Was the example clear?")

# 6 inches times 100 dpi is 600 pixels wide. 3 inches times 100 is 300 pixels tall.
fig.savefig("questionnaire-counts.png", dpi=100)
# The same inches at a finer dpi: 6 times 150 is 900 pixels wide.
fig.savefig("questionnaire-counts-fine.png", dpi=150)
# Show the figure under this cell as well as saving it.
plt.show()
```
