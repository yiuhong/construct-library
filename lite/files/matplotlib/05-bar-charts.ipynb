# Bar charts

When does a bar show a count honestly?

This notebook is in Matplotlib, in the module Common plots. A bar chart is for a count, or for another amount that starts at zero. The height is that amount. If the vertical axis is cut off so that it starts above zero, a small gap looks larger than it is. The counts have not changed.

The responses are from one questionnaire item: was the example clear? Eight students said yes, three were not sure, and four said no.


## From zero, and cut off

The left panel starts at zero. The right panel starts at 2. Both panels use the same three counts. The first plot in a session can take a moment while the plotting library downloads.


```python
# The same three counts, drawn from zero and then with the axis cut off below 2.
import matplotlib.pyplot as plt

# The three responses, in the order the item offered them.
labels = ["Yes", "Not sure", "No"]
# How many students chose that response.
counts = [8, 3, 4]

# Two panels, so the cut-off axis can sit beside the honest one.
fig, axes = plt.subplots(1, 2, figsize=(8, 3.4), layout="constrained")

# The left panel includes zero. The bar height is the count.
axes[0].bar(labels, counts, color="#4a4a4a")
axes[0].set_ylim(0, 10)
axes[0].set_xlabel("Response")
axes[0].set_ylabel("Count")
axes[0].set_title("From zero")

# The right panel starts at 2. The counts are unchanged, but the gaps look larger.
axes[1].bar(labels, counts, color="#4a4a4a")
axes[1].set_ylim(2, 10)
axes[1].set_xlabel("Response")
axes[1].set_ylabel("Count")
axes[1].set_title("Cut off below 2")

# The total is the sum of the three counts.
print(sum(counts))
plt.show()
```

## Long item names

`barh` draws the bars from a vertical list, which leaves room for a longer label. These are still counts, and the horizontal axis still starts at zero. This cell uses `plt` from the cell above.

Fifteen students answered. That is a small item, not a survey result.


```python
# Horizontal bars for the same counts. The length is still the count.
labels = ["Yes", "Not sure", "No"]
counts = [8, 3, 4]

fig, ax = plt.subplots(figsize=(6.4, 3.2), layout="constrained")
# barh takes the labels on the vertical axis and the counts as lengths.
ax.barh(labels, counts, color="#4a4a4a")
# Zero stays at the left, so the length is the count.
ax.set_xlim(0, 10)
ax.set_xlabel("Count")
ax.set_ylabel("Response")
ax.set_title("Was the example clear?")
plt.show()
```
