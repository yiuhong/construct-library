# Box plots

What does the box show that the median alone does not?

This notebook is in Matplotlib, in the module Common plots. Two seminars each have seven test scores. Both medians are 6. The boxes are not the same.

The line in the box is the median. The box runs from the lower quartile to the upper quartile, the middle half of the scores, using Matplotlib's default quartile rule. The whiskers run out to the last score that still sits within 1.5 box-heights of either end of the box. That 1.5 rule is Matplotlib's default. A score beyond the whisker would be drawn as its own point. On these two lists the whiskers reach the lowest and highest scores, so no separate point is drawn.

Seven scores, and a default rule that this notebook does not compare with another quartile rule, are a sketch of the lists. A published box plot would name the quartile rule and the sample size. A different rule can move the ends of the box.


## Two seminars

Morning scores `[4, 5, 5, 6, 6, 7, 8]`. Afternoon scores `[3, 4, 5, 6, 9, 9, 10]`. The cell prints the lower quartile, the median, and the upper quartile for each list. The first plot in a session can take a moment while the plotting library downloads.


```python
# Two box plots. Both medians are 6. The boxes are not the same height.
import matplotlib.pyplot as plt
import numpy as np

# Seven morning scores.
morning = [4, 5, 5, 6, 6, 7, 8]
# Seven afternoon scores. The median matches the morning list. The spread does not.
afternoon = [3, 4, 5, 6, 9, 9, 10]

# Lower quartile, median, and upper quartile, with NumPy's default percentile.
print([float(value) for value in np.percentile(morning, [25, 50, 75])])
print([float(value) for value in np.percentile(afternoon, [25, 50, 75])])

fig, ax = plt.subplots(figsize=(6.2, 3.8), layout="constrained")
# tick_labels names the two boxes. The default whisker is 1.5 box-heights.
ax.boxplot([morning, afternoon], tick_labels=["Morning", "Afternoon"])
ax.set_ylabel("Test score")
ax.set_title("Same median, different boxes")
plt.show()
```

## What the printout is saying

Morning prints `[5.0, 6.0, 6.5]`. The box runs from 5 to 6.5, and the line is at 6. Afternoon prints `[4.5, 6.0, 9.0]`. The box runs from 4.5 to 9, and the line is again at 6. The afternoon middle half is wider. The medians do not show that.

Those quartile numbers come from Matplotlib's default, which uses the same percentile rule as the printout. They are a description of these fourteen scores. They are not a finding about a seminar.
