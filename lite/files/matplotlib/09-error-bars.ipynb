# Error bars

What does the line through a mean add?

This notebook is in Matplotlib, in the module Common plots. Three topics, four quiz scores each. The point is the mean. The line is the standard error: the sample standard deviation, divided by the square root of the sample size. Here that sample size is 4.

The line is not a confidence interval, and it is not the range of the four scores. It is one summary of how far the mean moved inside this small list. Four scores is too small to report as a result about a topic. A fuller figure would say whether the line is a standard deviation, a standard error, or an interval, and it would use more than four rows.


## Three means

Memory is `[5, 6, 6, 7]`. Attention is `[4, 5, 7, 8]`. Language is `[6, 7, 7, 8]`. Memory and attention both have mean 6. Attention's scores sit further from that mean, so its line is longer. Language has mean 7 and the same standard error as memory, 0.408 when rounded to three decimal places. Attention's standard error rounds to 0.913.

The first plot in a session can take a moment while the plotting library downloads.


```python
# A point for each topic mean, and a line for the standard error of four scores.
import matplotlib.pyplot as plt

def standard_error(values):
    # The sample size.
    n = len(values)
    # The mean of this topic.
    mean = sum(values) / n
    # The sample variance: squared deviations, divided by n - 1.
    spread = sum((value - mean) ** 2 for value in values) / (n - 1)
    # The standard deviation, then divided by the square root of n.
    return (spread ** 0.5) / (n ** 0.5)

# Four scores on each topic.
memory = [5, 6, 6, 7]
attention = [4, 5, 7, 8]
language = [6, 7, 7, 8]
topics = [memory, attention, language]
names = ["Memory", "Attention", "Language"]

# One mean and one standard error per topic.
means = [sum(values) / len(values) for values in topics]
errors = [standard_error(values) for values in topics]
# Print the means, then the standard errors rounded to three decimal places.
print(means)
print([round(value, 3) for value in errors])

fig, ax = plt.subplots(figsize=(6.4, 3.6), layout="constrained")
# fmt="o" draws the mean as a point. yerr draws the line above and below it.
ax.errorbar(names, means, yerr=errors, fmt="o", color="#1c1c1c", capsize=4)
ax.set_ylabel("Mean quiz score")
ax.set_title("Four scores per topic")
# Start at zero so the means keep their height.
ax.set_ylim(0, 10)
plt.show()
```
