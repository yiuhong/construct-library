# Histograms

What does the height of a histogram bar count?

This notebook is in Matplotlib, in the module Common plots. Descriptive statistics already counted bins by hand. Here `ax.hist` does the count. The height is how many scores fell in the bin. The width of the bin is a choice. A wide bin can hide a pile that a narrower bin shows.

The scores are `[4, 5, 5, 6, 6, 6, 7, 7, 8, 9]`. With edges at 4, 6, 8, and 10, the bins are `[4, 6)`, `[6, 8)`, and `[8, 10]`. A score that lands on an edge goes into the bin on the right, except the last edge, which is included. So 6 belongs to the middle bin, and 8 belongs to the last bin.


## Three bins

The cell prints the three counts, `[3, 5, 2]`. The first plot in a session can take a moment while the plotting library downloads.


```python
# Count ten test scores into three bins, then draw one bar per bin.
import matplotlib.pyplot as plt

# Ten scores on one short test.
scores = [4, 5, 5, 6, 6, 6, 7, 7, 8, 9]
# The bin edges. The last edge is included. The others are the start of a bin.
edges = [4, 6, 8, 10]

fig, ax = plt.subplots(figsize=(6.4, 3.4), layout="constrained")
# hist returns the counts, the edges, and the bars. Keep the counts.
counts, drawn_edges, bars = ax.hist(scores, bins=edges, color="#4a4a4a", edgecolor="white")
ax.set_xlabel("Test score")
ax.set_ylabel("Count")
ax.set_title("Bins of width 2")
# The three bin counts, as ordinary integers.
print([int(count) for count in counts])
plt.show()
```

## Narrower bins

The same scores, with a bin for each integer from 4 up to 10. The three 6s become their own bar. In the wide bins they were part of a bar of five, together with the two 7s. This cell uses `plt` from the cell above.

Ten scores will move around if you change the edges. A histogram of a short list is a picture of that list and of the edges you chose.


```python
# The same scores, one bin per integer, so a pile of equal scores can show.
scores = [4, 5, 5, 6, 6, 6, 7, 7, 8, 9]
# Edges at every integer from 4 to 10.
edges = [4, 5, 6, 7, 8, 9, 10]

fig, ax = plt.subplots(figsize=(6.4, 3.4), layout="constrained")
counts, drawn_edges, bars = ax.hist(scores, bins=edges, color="#4a4a4a", edgecolor="white")
ax.set_xlabel("Test score")
ax.set_ylabel("Count")
ax.set_title("Bins of width 1")
print([int(count) for count in counts])
plt.show()
```
