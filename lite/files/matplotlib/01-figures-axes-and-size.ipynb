# Figures, axes, and size

What belongs to the figure, what belongs to the axes, and what does the size change?

You will make one page and one region, set the limits on that region, and then change the page size in inches. By the end you should be able to say which object holds the line, and what `figsize` measures.

A later plot is only as readable as the page it sits on. The line, the labels, and the limits belong to the axes. The size of the page is a property of the figure. A long label needs room on that page.

Plotting and packages scattered the Monday rows and joined revision hours to the weekly quiz. This lesson stays with the Monday quiz, and names the page and the region the line sits on. Two panels on one page are the next lesson.


## The page and the region

A figure is the page. An axes is the region you draw on. Matplotlib uses the word axes for one region as well as for several. `plt.subplots` returns both objects. `fig` is the figure. `ax` is the axes.

Six Monday scores, in weekly quiz class order: 4, 5, 6, 6, 7, and 8. The quiz is marked out of 10. The cell prints the class name of the page, the class name of the region, how many axes the figure holds, and whether `fig.axes[0]` is the same object as `ax`.

The first plot in a session can take a moment while the plotting library downloads.


```python
# One figure and the single axes on it. The Monday quiz is drawn on the axes, not on the figure.
import matplotlib.pyplot as plt

# Monday rows, in weekly quiz class order.
rows = [1, 2, 3, 4, 5, 6]
# The weekly quiz on those rows. Marked out of 10.
weekly_quiz = [4, 5, 6, 6, 7, 8]

# subplots with no arguments builds one page and one axes.
fig, ax = plt.subplots()
# Join the scores in row order, and mark each row.
# The colour argument is spelled color, which is Matplotlib's spelling.
ax.plot(rows, weekly_quiz, marker="o", color="#1c1c1c")
# The horizontal label belongs to the axes.
ax.set_xlabel("Monday row")
# The vertical label belongs to the axes.
ax.set_ylabel("Weekly quiz")
# The title belongs to the axes as well.
ax.set_title("Monday weekly quiz")
# The class name of the page.
print(type(fig).__name__)
# The class name of the region.
print(type(ax).__name__)
# How many regions the page holds.
print(len(fig.axes))
# The first region on the page is the axes subplots returned.
print(fig.axes[0] is ax)
# Show the figure under this cell.
plt.show()
```

## Limits belong to the axes

The same six scores, with the axes extended so the line does not fill the frame. `set_xlim` and `set_ylim` change the region. They do not add a second page. The vertical axis starts at zero, so the height is the score. This cell uses `plt` from the cell above.


```python
# The same Monday quiz, with the axes extended past the first and last points.
# The row numbers again.
rows = [1, 2, 3, 4, 5, 6]
# The same six scores.
weekly_quiz = [4, 5, 6, 6, 7, 8]

# A new page and a new axes.
fig, ax = plt.subplots()
# The same line as the cell above.
ax.plot(rows, weekly_quiz, marker="o", color="#1c1c1c")
# Show empty space on either side of the six rows.
ax.set_xlim(0, 7)
# Start the score axis at zero, so the height is the score.
ax.set_ylim(0, 10)
# Name the axes.
ax.set_xlabel("Monday row")
ax.set_ylabel("Weekly quiz")
ax.set_title("The same quiz, with the limits set")
# Show the figure under this cell.
plt.show()
```

## Width and height in inches

`figsize` is the figure, in inches: width, then height. The axes sits inside that page. Here the page is 8 inches wide and 3 inches tall, which leaves room for a label on every row of the class.

The labels are positions in the weekly quiz class, Monday 1 to 6 and then Thursday 1 to 6. The heights are revision hours: 1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, and 4. The cell prints the width and the height that were requested. This cell uses `plt` from the cell above.


```python
# A wide, short page. figsize is the figure, in inches: width, then height.
# Twelve row labels, in weekly quiz class order.
labels = [
    "Monday 1", "Monday 2", "Monday 3", "Monday 4", "Monday 5", "Monday 6",
    "Thursday 1", "Thursday 2", "Thursday 3", "Thursday 4", "Thursday 5", "Thursday 6",
]
# Revision hours on those same rows.
revision_hours = [1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, 4]

# Eight inches wide and three inches tall, with room reserved for the labels.
fig, ax = plt.subplots(figsize=(8, 3), layout="constrained")
# One point per row, joined so the class order is easy to follow.
ax.plot(labels, revision_hours, marker="o", color="#1c1c1c")
# Turn the row labels so they do not collide.
for label in ax.get_xticklabels():
    label.set_rotation(30)
    label.set_ha("right")
# Name what the horizontal axis lists.
ax.set_xlabel("Row in the weekly quiz class")
# Name the measurement, including the unit.
ax.set_ylabel("Revision hours")
ax.set_title("A wide figure")
# The width and the height that were requested, in inches.
print([float(value) for value in fig.get_size_inches()])
# Show the figure under this cell.
plt.show()
```

## Room for the labels

A small page leaves less room for those twelve labels. The first figure in this cell does not reserve space for them. The second uses `layout="constrained"`, which moves the axes inwards so the text stays on the page. The revision hours do not change. Both pages are 3.4 inches wide and 3.2 inches tall. This cell uses `plt` from the cell above.


```python
# A small page, first without a layout pass and then with room reserved for the labels.
# The same twelve row labels.
labels = [
    "Monday 1", "Monday 2", "Monday 3", "Monday 4", "Monday 5", "Monday 6",
    "Thursday 1", "Thursday 2", "Thursday 3", "Thursday 4", "Thursday 5", "Thursday 6",
]
# The same revision hours.
revision_hours = [1, 2, 3, 3, 4, 5, 2, 2, 3, 3, 4, 4]

# A nearly square page. No layout argument, so the labels use the default margins.
fig, ax = plt.subplots(figsize=(3.4, 3.2))
ax.plot(labels, revision_hours, marker="o", color="#1c1c1c")
for label in ax.get_xticklabels():
    label.set_rotation(30)
    label.set_ha("right")
ax.set_xlabel("Row in the weekly quiz class")
ax.set_ylabel("Revision hours")
ax.set_title("No layout pass")
plt.show()

# The same small page. layout="constrained" moves the axes inwards so the text stays on the page.
fig, ax = plt.subplots(figsize=(3.4, 3.2), layout="constrained")
ax.plot(labels, revision_hours, marker="o", color="#1c1c1c")
for label in ax.get_xticklabels():
    label.set_rotation(30)
    label.set_ha("right")
ax.set_xlabel("Row in the weekly quiz class")
ax.set_ylabel("Revision hours")
ax.set_title("Room for the labels")
plt.show()
```
